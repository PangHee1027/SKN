# weather.csv를 이용한 데이터 탐색

## 데이터 불러오기

```python
import pandas as pd
file_path = "./실습데이터/weather.csv"
df = pd.read_csv(file_path)
display(df)
```

## 요약 통계량 구하기

```python
df_stats = df.describe()
display(df_stats)
```

## temp 컬럼에 대한 평균값 구하기

```python
avg_temp = df_stats.loc["mean", "temp"]
print(avg_temp)
```

## hum 컬럼에 대한 최소값 구하기

```python
min_hum = df_stats.loc["min", "hum"]
print(min_hum)
```

## hum 컬럼에 대한 최대값 구하기

```python
max_hum = df_stats.loc["max", "hum"]
print(max_hum)
```

## atm 컬럼에 대한 중간값 구하기

```python
med_atm = df_stats.loc["50%", "atm"]
print(med_atm)
```

## speed 컬럼에 대한 표준 편차 구하기

```python
std_speed = df_stats.loc["std", "speed"]
print(std_speed)
```

## 상관 분석

```python
# 상관 행렬 구하기
corr_matrix = df.corr()
display(corr_matrix)
```

```python
# 온도(temp) 컬럼과 습도(hum) 컬럼의 상관 관계 분석
print(corr_matrix.loc["temp", "hum"])
```
