## 나. 중복되는 데이터를 그룹별로 모아서 처리하기
### 1. groupby()
#### 가) 중복된 데이터를 그룹별로 분할
#### 나) 분할된 각 그룹에 대하여 별도로 데이터를 처리하거나 그룹별 통계량을 확인하고자 할 때 사용
### 2. groupby(by).func()
#### 가) by=[co1, …] : 중복 여부를 확인하여 그룹별로 묶을 컬럼 지정
#### 나) func(함수) : 그룹별로 묶은 데이터에 대한 특정 함수 적용 예) sum(), mean()

```python
channel_rev = df2.groupby(by="Channel")["Revenue"].mean()
print(channel_rev)
```

### 3. groupby(by).agg(새로운_컬럼명=('원본 컬럼명', '집계 함수'))
  * 기능 : 각 그룹별로 다양한 함수를 동시에 적용

```python
channel_summary = df2.groupby('Channel').agg(
    total_cost = ('Cost', 'sum'),
    average_cost = ('Cost', 'mean'),
    total_revenue = ("Revenue", 'sum'),
    average_revenue = ("Revenue", 'mean')
)

print(channel_summary)
```

## 다. 이상치(outlier) 처리하기
### 1. 이상치 : 수집된 데이터 중에서 아주 작은 값이나 매우 큰 값
### 2. 이상치 측정 : IQR 이용
#### 가) 사분위수 : 순서대로 정렬된 데이터를 4구간으로 균등하게 분할 할 때 기준이 되는 수
#### 나) 백분위수 : 순서대로 정렬된 데이터를 균등하게 100 등분하는 경우 기준이 되는 수
#### 다) IQR(Interquartile range) : 사분위수의 75% 지점의 값과 25% 지점의 값 차이(Q3 - Q1)
#### 라) 정상 범위 데이터의 최소값(min) = Q1 – (1.5 * IQR)
#### 마) 정상 범위 데이터의 최대값(max) = Q3 + (1.5 * IQR)
#### 바) 이상치 : min, max를 벖어난 값
#### 라) 분위수(quantile) 구하기: quantile(q) 함수
##### 1) q : 원하는 백분위수 설정
##### 2) df.quantile(q) / df["col"].quantile(q)

```python
q1 = df3.quantile(q=0.25)
q2 = df3.quantile(q=0.5)
q3 = df3.quantile(q=0.75)
print(q1)
print("-" * 80)
print(q2)
print("-" * 80)
print(q3)
```

```python
iqr = q3 - q1
min_value = q1 - (iqr * 1.5)
max_value = q3 + (iqr * 1.5)
print(min_value)
print("-" * 80)
print(max_value)
```

```python
print(df3 < min_value)
print("-" * 80)
print(df3 > max_value)
print("-" * 80)
outlier_bool = (df3 < min_value) | (df3 > max_value)
print(outlier_bool)
```

```python
print(outlier_bool.any())
print("-" * 80)
has_outlier = outlier_bool.any(axis=1)
print(has_outlier)
```

```python
output = ~has_outlier
print(output)
print("-" * 80)
arr_bool = output.values
print(arr_bool)
```

### 3. 이상치 처리 방법
#### 가) 제거
  * 전체 수치형 컬럼에 대한 Q1, Q3, IQR 한 번에 계산
  * 각 컬럼별 정상 범위 최소값/최대값 계산
  *  데이터프레임 전체에 대해 논리 연산자 적용, 이상치 여부 판단 → 블리언 DataFrame 생성
  * 어떤 컬럼에서든 이상치가 하나라도 있는 행 식별
  * 블리언 인덱싱을 통한 이상치가 없는 행만 필터링 후 인덱스 재설정


```python
df3_normal = df3[arr_bool]
print(df3_normal)
```

#### 나) 대체
  * 정상 범위 최소값 미만은 정상 범위 최소값으로, 정상 범위 최대값 초과는 정상 범위 최소값으로 일괄 대체
  * clip 함수 사용: df_clipped = df.clip(lower=min_val, upper=max_val, axis=1)
