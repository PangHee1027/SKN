```python
import pandas as pd
file_path = "./실습데이터/ad_performance.csv"
df = pd.read_csv(file_path)
display(df)
```

# 데이터 탐색
## 가. 통계로 요약하기
### 1. 요약(기술) 통계량 구하기
#### 가) 요약 통계량
  * 데이터 분포의 특징을 나타내는 통계량
  * 데이터 분석의 과정에서 데이터의 전체적인 특징과 분포를 빠르게 파악하기 위해 사용
  * 구현 방법 : df.describe()

```python
df_stats = df.describe()
display(df_stats.round(1))
```

#### 나) 숫자(Numeric) 데이터의 경우 (기본 동작)
  * count : 데이터 개수 (NaN 값 제외)
  * mean : 평균값
  * std : 표준편차 (데이터가 퍼져있는 정도)
  * min : 최솟값
  * 25% : 1사분위수 (하위 25% 지점의 값)
  * 50% : 중앙값 (median, 2사분위수)
  * 75% : 3사분위수 (하위 75% 지점의 값)
  * max : 최댓값

### 2. 요약 통계량으로부터 평균 구하기

```python
avg_cost = df_stats.loc["mean", "Cost"]
print(f"{avg_cost:.1f}")
print("-" * 80)
avg_cost2 = df["Cost"].mean()
print(f"{avg_cost2:.1f}")
```

### 3. 요약 통계량으로부터 최솟값, 최댓값 구하기

```python
min_rev = df_stats.loc["min", "Revenue"]
max_rev = df_stats.loc["max", "Revenue"]
print(min_rev, max_rev)
print("-" * 80)
min_rev1, max_rev1 = min(df["Revenue"]), max(df["Revenue"])
print(min_rev1, max_rev1)
```

### 4. 요약 통계량으로부터 분위수(quantile) 구하기

```python
rev_25 = df_stats.loc["25%", "Revenue"]
rev_50 = df_stats.loc["50%", "Revenue"]
rev_75 = df_stats.loc["75%", "Revenue"]
print(rev_25, rev_50, rev_75)
print("-" * 80)
rev_25_1 = df["Revenue"].quantile(q=0.25)
rev_50_1 = df["Revenue"].quantile(q=0.5)
rev_75_1 = df["Revenue"].quantile(q=0.75)
print(rev_25_1, rev_50_1, rev_75_1)
print("-" * 80)
rev_50_2 = df["Revenue"].median()
print(rev_50_2)
```

### 5. 요약 통계량으로부터 표준편차(standard deviation) 구하기
#### 가) 표준편차 : 데이터들이 평균값에서 평균적으로 얼마나 떨어져 있는가?
  * 데이터의 '일관성' 또는 변동성을 보여주는 지표
  * 표준편차가 작다: 데이터들이 평균값 주변에 옹기종기 모여있다. (성과가 꾸준하고 예측 가능)
  * 표준편차가 크다: 데이터들이 평균값에서 멀리, 넓게 흩어져 있다. (성과가 들쭉날쭉하고 예측하기 힘듦)

```python
std_rev = df_stats.loc["std", "Revenue"]
print(std_rev)
print("-" * 80)
std_rev1 = df["Revenue"].std()
print(std_rev1)
```

```python
count_rev = df_stats.loc["count", "Revenue"]
print(count_rev)
print("-" * 80)
count_rev1 = len(df["Revenue"])
print(count_rev1)
print("-" * 80)
count_rev2 = df["Revenue"].count()
print(count_rev2)
```

## 나. 상관관계 분석(Correlation Analysis)
### 1. 개념 : 두 변수 사이에 연관성의 존재 여부와 연관성의 강도를 측정하여 분석하는 방법
### 2. 상관관계의 종류
#### 가) 양(+)의 상관관계
  * 한 변수의 값이 증가할 때 다른 변수의 값도 증가한다
  * 강도에 따라 강한 양의 상관관계, 약한 양의 상관관계가 있다
#### 나) 음(-)의 상관관계
  * 한 변수의 값이 증가할 때 다른 변수의 값은 감소한다
  * 강도에 따라 강한 음의 상관관계, 약한 음의 상관관계가 있다
#### 다) 상관계수(피어슨 상관계수) : 두 변수 갂에 존재하는 선형 관계의 정도를 수량화, +1 ~ -1
#### 라) 구현방법
  * df.corr(numeric_only=True) : 변수들 간의 상관 계수를 표현한 데이터프레임 생성 → 상관 행렬

```python
df_correlation = df.corr(numeric_only = True)
display(df_correlation)
```

```python
import plotly.express as px
file_path1 = "./실습데이터/tips.csv"
tips_df = pd.read_csv(file_path1)
display(tips_df)
```

### 4. 주로 사용되는 시각화 도구
#### 가) 직선 그래프
##### 1) 개념: 두 개 이상의 데이터 포인트를 직선으로 연결하여 연속적인 변화와 추세를 시각적으로 표현하는 그래프
##### 2) px.line(data_frame, x, y, color)
##### 3) 하이퍼파라미터
  * data_frame : 데이터프레임
  * x, y : 시각화 할 컬럼
  * color : 새로운 변수를 추가하여 색상으로 구분

```python
sort_by_bill = tips_df.sort_values(by = "total_bill")
fig1 = px.line(sort_by_bill, "total_bill", "tip", color="day")
fig1.write_html("./lines.html")
fig1.show()
```

#### 나) 막대 그래프
##### 1) 개념 : 범주형 컬럼에 대해 범주별 수치 데이터의 크기(총합, 평균 등)를 직사각형 막대로 표현하는 그래프
##### 2) 구현 방법
  * 수직 막대 : px.bar(data_frame, x, y)
  * 수평 막대: px.bar(data_frame, x, y, orientation='h')
##### 3) 하이퍼파라미터
  * data_frame : 데이터프레임
  * x, y
    * 수직형: x가 범주형 컬럼, y가 수치형 컬럼
    * 수평형: x가 수치형 컬럼, y가 범주형 컬럼
  * orientation: 'v' (수직, 기본값), 'h' (수평)

```python
fig2 = px.bar(tips_df.groupby(by="day")["total_bill"].mean())
fig2.show()
```
