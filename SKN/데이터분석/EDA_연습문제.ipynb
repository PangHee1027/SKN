```python
import pandas as pd
import plotly.express as px

file_path = "./실습데이터/tips.csv"
df = pd.read_csv(file_path)
display(df)
```

# 1. 데이터 정렬과 선 그래프(Line Plot)
  * 내용 : tip 컬럼을 기준으로 데이터를 내림차순으로 정렬한 뒤, px.line을 사용하여 tip과 total_bill의 관계를 나타내세요.
  * 조건 : 실습 파일에서 사용한 sort_values와 px.line의 기본 구조를 활용하세요.

```python
df_tip_sort = df.sort_values("tip", ascending=False)
fig1 = px.line(df_tip_sort, "tip", "total_bill")
fig1.show()
```

# 2. 그룹화 연산과 수직 막대 그래프(Bar Chart)
  * 내용 : 성별(sex)에 따른 tip의 평균값을 구하고, 이를 데이터프레임으로 변환(reset_index)한 뒤 수직 막대 그래프로 시각화하세요.

```python
tip_by_sex = df.groupby(by = "sex")["tip"].mean()
df_tip_by_sex = tip_by_sex.reset_index()
df_tip_by_sex = df_tip_by_sex.rename(columns={"tip":"avg_tip"})
fig2 = px.bar(df_tip_by_sex, "sex", "avg_tip", color="sex")
fig2.show()
```

# 3. 카테고리 정렬과 수평 막대 그래프(Horizontal Bar Chart)
  * 내용 : 요일(day)별 total_bill의 평균값을 구하여 데이터프레임으로 만든 뒤, 수평 막대 그래프(orientation='h')로 시각화하세요.
  * 조건 : 요일의 순서를 ['Thur', 'Fri', 'Sat', 'Sun']으로 직접 지정(category_orders)하세요.

```python
bill_by_day = df.groupby(by = "day")["total_bill"].mean()
df_bill_by_day = bill_by_day.reset_index()
fig3 = px.bar(df_bill_by_day, x="total_bill", y="day", orientation="h", category_orders=['Thur', 'Fri', 'Sat', 'Sun'])
fig3.show()
```

# 4. 빈도 분석과 원 그래프(Pie Chart)
  * 내용 : 식사 시간대(time)별 데이터의 빈도 비율을 구하고, 이를 도넛 형태(hole=0.3)의 원그래프로 시각화하세요.
  * 조건 : value_counts(normalize=True)를 사용하여 비율을 계산하세요.

```python
count_time = df["time"].value_counts(normalize=True)
df_count_time = count_time.reset_index()
fig4 = px.pie(df_count_time, names = "time", values="proportion", hole = 0.3)
fig4.show()
```

# 5. 데이터 분포 분석과 히스토그램(Histogram)
  * 내용 : tip 금액의 구간별 빈도수를 히스토그램으로 나타내되, 흡연 여부(smoker)에 따라 막대의 색상이 구분되도록 설정하세요.

```python
fig5 = px.histogram(df, x = "tip", color="smoker", barmode="group")
fig5.show()
```

# 6. 상자 그림(Box Plot)을 통한 분포 비교
  * 내용 : 요일(day)별 total_bill의 분포를 상자 그림으로 시각화하세요.
  * 조건 : 요일 순서를 ['Thur', 'Fri', 'Sat', 'Sun']으로 정렬하고, 수평 방향으로 그래프를 생성하세요.

```python
fig6 = px.box(df, "total_bill", "day", category_orders=['Thur', 'Fri', 'Sat', 'Sun'], orientation="h")
fig6.show()
```

# 7. 상관 관계 시각화(Scatter Plot)
  * 내용: total_bill과 tip의 관계를 산점도로 나타내되, 성별(sex)에 따라 점의 색상이 다르게 표시되도록 설정하세요.

```python
fig7 = px.scatter(df, "total_bill", "tip", color="sex")
fig7.show()
```

# 8. 상관 분석과 히트맵(Heatmap)
  * 내용 : 데이터프레임의 모든 수치형 컬럼 간 상관 계수를 계산하여 상관 행렬을 만들고, 이를 px.imshow를 사용하여 히트맵으로 시각화하세요.
  * 조건 : 상관 계수 수치가 그래프 위에 표시되도록(text_auto) 설정하세요.

```python
corr_matrix = df.corr(numeric_only=True)
fig8 = px.imshow(corr_matrix, text_auto=".2f", color_continuous_scale="Viridis")
fig8.show()
```
