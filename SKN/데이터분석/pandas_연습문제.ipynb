```python
import pandas as pd
file_path='./실습데이터/ad_performance.csv'
```

# 1. 데이터 불러오기
## pd.read_csv()를 사용하여 ad_performance.csv 파일을 불러와 변수 df에 저장하고, display 함수를 이용하여 전체 데이터를 확인하는 코드를 작성하세요.

```python
df = pd.read_csv(file_path)
display(df)
```

# 2. 인덱스 확인
## 불러온 데이터프레임(df)에서 전체 컬럼명(Columns) 정보만 추출하여 출력하세요.

```python
info_columns = df.columns
print(info_columns)
```

# 3. 특정 컬럼 선택 (Fancy Indexing)
## df에서 'Channel', 'Age_Group', 'Revenue' 세 개의 컬럼만 선택하여 새로운 데이터프레임 df_sub를 생성하세요.

```python
df_sub = df[['Channel', 'Age_Group', 'Revenue']]
print(df_sub)
```

# 4. 새로운 컬럼 추가
## 클릭당 비용을 의미하는 'CPC' 컬럼을 추가하세요. (공식: Cost / Clicks)

```python
df['CPC'] = df['Cost'] / df['Clicks']
display(df['CPC'])
```

# 5. 빈도수 계산
## Age_Group' 컬럼에 있는 각 항목별(20s, 30s 등) 데이터의 빈도수를 확인하세요

```python
channel_counts = df['Age_Group'].value_counts()
print(channel_counts)
```
