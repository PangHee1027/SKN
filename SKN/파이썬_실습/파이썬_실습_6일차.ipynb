## 2. 사용자 정의 함수 + 파일 읽기('r') + try-except FileNotFoundError 융합
### 파일 경로 file_path를 매개변수로 받는 함수 read_file_safely를 정의하세요. with open(..., mode='r')과 .read() 함수를 사용해 파일 전체 내용을 읽어 반환하되, 존재하지 않는 파일명 입력으로 FileNotFoundError가 발생할 경우 프로그램이 중단되지 않고 "파일을 찾을 수 없습니다."라는 안내 문자열을 반환하도록 명시적 예외 처리를 적용하는 코드를 작성하세요.

```python
def read_file_safety(fp) :
    try :
        with open(fp, "r", encoding = "utf-8") as f :
            lines = f.read()
            return lines
    except FileNotFoundError:
        return "파일을 찾을 수 없습니다."

print(read_file_safety("표류.txt"))
```

## 3. 사용자 정의 함수 + 리스트 인덱싱 + 다중 명시적 예외 처리 융합
### 리스트 자료형 데이터 data_list와 접근할 인덱스 index를 매개변수로 받는 함수 parse_and_divide를 작성하세요. 함수 내부에서 100 // int(data_list[index]) 연산을 수행하여 정수 몫을 반환합니다. 단, 다음 3가지 예외 상황에 대해 각각 지정된 메시지를 반환하도록 try-except 블록을 구성하세요.
  * IndexError: "인덱스 범위 초과"
  * ValueError: "숫자 변환 불가"
  * ZeroDivisionError: "0으로 나눌 수 없음"


```python
def parse_and_divide(dl, index) :
    try:
        return 100 // int(dl[index])
    except IndexError :
        print("인덱스 범위 초과")
    except ValueError :
        print("숫자 변환 불가")
    except ZeroDivisionError :
        print("0으로 나눌 수 없음")

data_list = [0, 1, 2, 3, 4]
data_list.append("a")

for i in range(len(data_list) + 1) :
    print(parse_and_divide(data_list, i))
```

## 4. 파일 읽기(f.readlines()) + 문자열 .strip(), .split() + 딕셔너리 수집 + 포괄적 예외 처리(Exception as e) 융합
### 파일 경로 file_path를 입력받아 성적 정보를 딕셔너리로 변환하는 함수 parse_scores_file을 작성하세요.
  * with open(..., mode='r')과 f.readlines()를 활용해 파일의 각 줄을 리스트로 읽어옵니다.
  * for문으로 각 줄을 순회하며 .strip()으로 공백/줄바꿈을 제거하고, 쉼표(,)를 기준으로 .split(",")하여 이름과 점수(정수형 int 변환)를 딕셔너리 score_dict에 Key-Value로 저장합니다.
  * 파일 읽기 및 처리 과정 전체를 try-except Exception as e: 블록으로 감싸 예외 발생 시 "프로그램 실행 중 예측하지 못한 오류가 발생했습니다." 및 오류 유형을 출력하고(print(e)), None을 반환하도록 작성하세요.

```python
file_path = "C:/Users/playdata2/work/SKN/파이썬_실습/실습파일/score.txt"

def parse_scores_file(fp) :
    try :
        with open(fp, mode = 'r', encoding = "utf-8") as f :
            lines = f.readlines()
            score_dict = {}
            for i in range(len(lines)) :
                cleand_line = lines[i].strip()
                if cleand_line :
                    name, score = cleand_line.split(",")
                    score_dict[name] = score
            return score_dict
    except Exception as e :
        print(f"프로그램 실행 중 예측하지 못한 오류가 발생했습니다 : {e}")
        return None
        
print(parse_scores_file(file_path))       
```

# 가. 클래스
## 1. 정의: 기본 자료형(int, float, str, list, dict 등)만으로는 표현하기 어려운 현실 세계의 데이터를 표현하기 위해, 기존에 없던 새로운 종류의 데이터 타입(Custom Data Type)을 직접 정의하는 것입니다.
### 가) 클래스 정의

```python
# 1. 'Customer(고객)'이라는 이름의 커스텀 자료형 '틀(설계도)' 만들기
class Customer:
    # 이 자료형이 만들어질 때 가질 세부 데이터(속성)를 정한다
    def __init__(self, name, age, gender):
        self.name = name    # 고객 이름 데이터
        self.age = age      # 고객 나이 데이터
        self.gender = gender # 고객 성별 데이터
    
    # 이 자료형만 사용할 수 있는 전용 기능(메서드)을 정한다
    def purchase(self, price, product_name):
        print(f"[{self.name}] 고객님이 {price}원 상당의 {product_name}를 구매하셨습니다.")

# 2. 'Customer' 클래스 이용해 구체적인 데이터 생성하기 (인스턴스화)
client = Customer("김철수", 24, "남성")
```

### 나) 변수에서 데이터 꺼내 쓰기 및 기능(메서드) 실행
  * 데이터 꺼내기

```python
print(client.name)  # 결과: 김철수
print(client.age)   # 결과: 24
```

  * 전용 기능(메소드) 실행하기

```python
client.purchase(900000, "아이패드")
# 결과: [김철수] 고객님이 900000원 상당의 아이패드를 구매하셨습니다.
```

# 나. 클래스 상속
## 1. 상속을 사용하는 이유
  * 코드 재사용성: 공통된 기능을 부모 클래스에 미리 구현해 두면, 자식 클래스에서 중복 코드를 작성할 필요 없이 그대로 재사용 가능
  * 유지보수성: 공통 로직에 수정이 필요할 경우 부모 클래스 한 곳만 수정하면 되므로 관리가 편리
  * 확장성: 기존의 원본 코드를 건드리지 않고, 자식 클래스에서 안전하게 새로운 기능을 추가 가능
## 2. 상속 기본 문법

```python
class 부모클래스:
    # 부모의 속성과 메서드 정의
    pass

class 자식클래스(부모클래스):
    # 부모의 기능을 물려받고, 추가적인 속성이나 메서드 정의
    pass
```

```python
# 1. 부모 클래스 (Parent / Super Class)
class Customer:
    def __init__(self, name, age, gender):
        self.name = name       # 고객 이름
        self.age = age         # 고객 나이
        self.gender = gender   # 고객 성별
    
    def purchase(self, price, product_name):
        print(f"[{self.name}] 고객님이 {price}원 상당의 {product_name}를 구매하셨습니다.")

# 2. 자식 클래스 (Child / Sub Class) - Customer 클래스를 상속 받음
class VIPCustomer(Customer):
    def __init__(self, name, age, gender, discount_rate):
        # super()를 통해 부모 클래스의 __init__ 메서드 호출 (이름, 나이, 성별 초기화)
        super().__init__(name, age, gender)
        self.discount_rate = discount_rate  # VIP 전용 할인율 속성 추가
    
    # 메서드 오버라이딩 (Method Overriding): 부모에게 물려받은 기능을 자식에 맞게 재정의
    def purchase(self, price, product_name):
        discounted_price = int(price * (1 - self.discount_rate))
        print(f"[VIP 고객] {self.name}님은 {int(self.discount_rate * 100)}% 할인 적용되어 "
              f"{discounted_price}원에 {product_name}를 구매하셨습니다.")

# --- 실행 및 테스트 ---

# 일반 고객 객체 생성
normal_customer = Customer("김철수", 24, "남성")
normal_customer.purchase(100000, "일반 이어폰") 
# 출력: [김철수] 고객님이 100000원 상당의 일반 이어폰를 구매하셨습니다.

print("-" * 50)

# VIP 고객 객체 생성 (할인율 20% 적용)
vip_customer = VIPCustomer("이지은", 29, "여성", 0.2)
vip_customer.purchase(100000, "프리미엄 헤드폰") 
# 출력: [VIP 고객] 이지은님은 20% 할인 적용되어 80000원에 프리미엄 헤드폰를 구매하셨습니다.
```

## 3. 핵심 포인트
### 가) super().__init__(...): 자식 클래스에서 부모 클래스의 속성들을 초기화할 때 사용합니다. 부모의 코드를 그대로 가져와 재사용하는 대표적인 방법
### 나) 메서드 오버라이딩(Method Overriding): 부모 클래스로부터 물려받은 purchase 메서드를 자식 클래스에서 입맛에 맞게 덮어씌워 재정의
