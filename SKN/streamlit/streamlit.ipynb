# streamlit
## 가. Streamlit 개요
### 1. 정의: 복잡한 웹 개발 언어(HTML, CSS, JavaScript)나 백엔드 서버 지식 없이 순수 파이썬만으로 웹사이트를 만들 수 있는 도구
### 사용 목적: 파이썬 변수와 함수 몇 줄만으로 단 몇 분 만에 완성도 높은 대시보드 웹 생성
### 기본 사용법
  * 설치: pip install streamlit (터미널 입력)
  * 임포트: import streamlit as st (파이썬 코드 상단 작성)
  * 함수 호출: st.func()
  * 실행: streamlit run 파일명.py (웹 브라우저 실행)
## 나. 주요 기능 및 구현 함수
### 1. 텍스트 출력
  * st.title(): 가장 큰 메인 제목 생성
  * st.header(): 대분류 주제 제목 생성
  * st.subheader(): 중분류 세부 제목 생성
  * st.write(): 문자, 숫자, 딕셔너리 등 무엇이든 출력하는 만능 함수 (파이썬의 print() 역할)
  * st.caption(): 화면 하단의 작은 주석이나 부연 설명 작성
### 2. 사용자 입력 위젯 (Interactive UI)
  * 선택형 입력
    * st.button(): 클릭 시 True/False 반환
    * st.checkbox(): 체크 여부(True/False) 확인
    * st.radio(): 여러 개 중 단 하나만 선택하는 라디오 버튼
    * st.selectbox(): 드롭다운 메뉴로 하나만 선택
    * st.multiselect(): 여러 개를 동시에 선택 (파이썬 리스트 반환)
  * 값 입력
    * st.text_input(): 문자열 입력 창
    * st.number_input(): 숫자 입력 창 (정수/실수 반환)
### 3. 데이터 표출 및 차트 생성
딕셔너리나 리스트 데이터를 웹 화면에 시각적으로 전달
  * st.metric(): 매출액, 회원 수 같은 핵심 성과 지표(KPI)와 변화량(Delta)을 하이라이트
  * st.table(): 파이썬 데이터를 깔끔한 웹 표 형태로 출력
  * st.bar_chart(), st.line_chart(): 막대그래프나 선그래프 즉시 생성
### 4. 레이아웃 및 컨테이너
화면 공간을 효율적으로 분할하여 깔끔한 UI를 제공
  * st.sidebar(): 화면 왼쪽에 메뉴나 설정창 생성
  * st.columns(): 화면을 가로로 다단 컬럼 분할
  * st.tabs(): 탭을 눌러 서로 다른 내용을 전환
  * st.expander(): 긴 설명글을 접었다가 클릭할 때만 펼치게 생성
### 5. 세션 상태 (st.session_state) 메모리
  * 개념: 웹 브라우저가 열려 있는 동안 파이썬 변수 값을 기억하고 보존하는 메모리 도구
  * 필요성: Streamlit은 버튼 클릭 등 상호작용 발생 시 스크립트를 처음부터 다시 실행(새로고침)하므로 일반 변수는 초기화됩니다. 데이터를 유지하려면 st.session_state를 사용
  * 규칙: 존재하지 않는 키를 불러오면 에러가 나므로 if 'key' not in st.session_state:로 존재 여부를 먼저 검사
### 6. 데이터 CRUD (동기화)
  * 데이터의 상태(생성, 조회, 수정, 삭제) 변화를 실시간으로 웹 화면에 일치(동기화) 시킴
  * 리스트나 딕셔너리를 활용하여 입력값 추가(append), 조회, 수정, 삭제 기능을 구현
### 7. 입력 폼 (st.form)
  * 개념: 여러 개의 입력 위젯을 하나로 묶어, [제출 버튼]을 누르는 순간에만 데이터가 전송되고 화면이 새로고침되도록 제어하는 기능
  * 목적: 입력창을 칠 때마다 매번 화면이 새로고침되며 앱이 버벅거리는 현상을 방지 (st.form_submit_button 사용)
### 8. 환경(색상, 폰트) 설정 (config.toml)
  * 개념: 웹 앱의 글꼴, 배경색, 메인 강조 색상 등 전체 디자인 테마를 통합 관리하는 파일
  * 경로 규칙: 반드시 폴더 최상단에 .streamlit 폴더를 만들고 그 안에 config.toml 파일로 생성해야 인식됩니다 ([내 프로젝트 폴더] > [.streamlit] > [config.toml]).
  * 주요 설정 항목 ([theme])
    * primaryColor: 강조 포인트 색상
    * backgroundColor: 메인 화면 배경 색상
    * secondaryBackgroundColor: 사이드바 등 보조 배경 색상
    * textColor: 기본 텍스트 색상
    * font: 기본 글꼴 스타일 (sans serif, serif, monospace)
