# 가. 기본 자료형, 출력

## 1. 기본 자료형 (단일 값 저장)
  * 숫자형 (int, float): 계산을 위한 정수 및 실수 (예: 나이, 금액)

  * 문자열 (str): 따옴표(', ")로 감싼 글자 (예: 이름, 주소)

  * 불리언 (bool): 참(True)과 거짓(False)의 논리 값


```python
# 기본 출력, 기본 자료형
a = 1
print(a)
print(2026)
print('Hello python')
print("안녕 파이썬")
```

## 2. 변수와 입출력, 네이밍 규칙
  * 변수: 데이터(값)가 저장되는 메모리 공간의 이름
  * 입출력: 키보드로 입력받고, 모니터로 출력 (print() 함수 사용).
  * 문자열 출력 (f-string): f'text {변수}' 형태로 문자열 내에 변수 값을 직접 삽입하여 출력.
  * 변수 네이밍 규칙:
    * 알파벳(대소문자 구분), 숫자, 언더바(_), 한글 사용 가능.
    * 숫자는 변수 이름의 맨 처음에 올 수 없으며,
    이름 사이 공백은 불가능.

```python
# f-string (문자열 내 변수 삽입)
height = 181
print(f'저의 키는 {height}cm 입니다.')

name = '이재현'
print(f'제 이름은 {name}입니다.')
```

# 나. 묶음 자료형 (List, Tuple, Dictionary)

## 2. 묶음 자료형 (여러 값 저장)
  * 리스트 (list): 순서(인덱스)가 존재하며, 데이터 추가·수정·삭제가 자유로운 대표 묶음 자료형

  * 튜플 (tuple): 순서가 존재하나, 한 번 생성하면 수정할 수 없는 읽기 전용(Read-Only) 자료형

  * 딕셔너리 (dict): Key와 Value를 1:1로 쌍을 이루어 저장하는 매핑형 자료형

## 1. 리스트

### 가) 리스트 기본
  * 생성 예시
    * 빈 리스트: a = []
    * 숫자/문자 혼합 및 중첩 리스트: e = [1, 2, ['My', 'name']]

```python
# --- 리스트 (List) ---
a = []
b = [1, 2, 3]
c = ['My', 'name', 'is', 'MH']
d = [1, 2, 'My', 'name']
e = [1, 2, ['My', 'name']]
print(a)
print(b)
print(c)
print(d)
print(e)
```

### 나) 리스트 인덱싱
  * 인덱싱 (Indexing): 리스트[인덱스] 형태로 특정 위치의 값을 추출 (0부터 시작).

```python
# 인덱싱 및 슬라이싱
list1 = [1, 2, 3, ['a', 'b', 'c']]
print(list1[3][1])  # 'b' 출력
```

### 다) 리스트 슬라이싱
  * 슬라이싱 (Slicing): 리스트[start:stop] 형태로 start부터 stop - 1 위치까지의 범위를 추출.주요 

```python
list2 = [0, 10, 20, 30, 40]
print(list2[0:3])   # [0, 10, 20]
print(list2[3:])    # [30, 40]
```

### 라) 리스트 메소드
  * 리스트.append(값): 리스트의 맨 뒤에 새로운 값을 추가.
  * len(리스트): 모든 자료형에 공통 적용되는 내장 함수로, 리스트 원소의 전체 개수를 반환.

```python
# 리스트 전용 메서드
list5 = [0, 1, 2, 3, 4]
list5.append(5)     # [0, 1, 2, 3, 4, 5]
print(len(list5))   # 원소 개수 확인
```
