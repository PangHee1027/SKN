# 가. 함수
## 1. 함수의 정의
  * 입력 받은 데이터를 가지고 사젂에 정의된 기능을 수행하고 결과값을 전달하는 코드의 집합
## 2. 함수 구조

```python
def add(x, y):
 sum = x + y
 return sum
```

## 3. 함수 호출
  * 함수 실행 : 함수 이름 뒤의 소괄호 안에 필요핚 인수(argument) 입력 후 실행

```python
result = add(3, 5)
print(result)
```

## 4. 함수의 종류
### 가) 사용자 정의함수
* 사용자가 직접 정의하고 호출해서 사용

```python
def multi(x, y):
 sum = x * y
 return sum

result1 = multi(3, 5)
print(result1)

print("-" * 80)

result2 = multi(y = 5, x = 3) 
print(result2)
```

* 리스트 전용 함수
  * append : 값 추가
  * extend : 리스트 연결
  * len : 성분 원소의 개수 확인


```python
list1 = [0, 1, 2, 3, 4]
list1.append(5)
print(list1)

print("-" * 80)

list1.append(6)
print(list1)
```

```python
list2 = [7, 8, 9, 10]
list1.extend(list2)
print(list1)
```

```python
list3 = [11, 12, 13, 14]
list1 = list1 + list3
print(list1)
```

```python
list4 = [0, 1, 2]
print(len(list3))

print("-" * 80)

list5 = ["a", "b", "c", "d", "e", "f"]
print(len(list4))
```

* 딕셔너리 전용 함수
  * keys : key만 가져오기
  * values : value만 가져오기
  * items : key, value 쌍으로 가져오기
  * get : key로 value 가져오기

```python
dict_data = {'name': '박보검', 'age': 31}

# 전체 key 추출
keys_data = dict_data.keys()
print(f'전체 key 확인: {keys_data}')

# 전체 value 추출
values_data = dict_data.values()
print(f'전체 value 확인: {values_data}')
```

```python
dict_data = {'name': '박보검', 'age': 31, 'phone': '010-1234-1234'}

# 전체 key, value 쌍 추출
items_data = dict_data.items()
print(f'전체 key, value 쌍 확인: {items_data}')

# get() 함수로 value 추출
name = dict_data.get('name')
print(name)
```
