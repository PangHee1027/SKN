# 가. 함수
## 1. 함수의 정의
  * 입력 받은 데이터를 가지고 사젂에 정의된 기능을 수행하고 결과값을 전달하는 코드의 집합
## 2. 함수 구조

```python
def add(x, y):
 sum = x + y
 return sum
```

## 3. 함수 호출
  * 함수 실행 : 함수 이름 뒤의 소괄호 안에 필요핚 인수(argument) 입력 후 실행

```python
result = add(3, 5)
print(result)
```

## 4. 함수의 종류
### 가) 사용자 정의함수
* 사용자가 직접 정의하고 호출해서 사용

```python
def multi(x, y):
 sum = x * y
 return sum

result1 = multi(3, 5)
print(result1)

print("-" * 80)

result2 = multi(y = 5, x = 3) 
print(result2)
```

  * python에서 정의 → 호출만 하면 사용 가능

* 리스트 전용 함수
  * append() : 값 추가
  * extend() : 리스트 연결
  * len() : 성분 원소의 개수 확인


```python
list1 = [0, 1, 2, 3, 4]
list1.append(5)
print(list1)

print("-" * 80)

list1.append(6)
print(list1)
```

```python
list2 = [7, 8, 9, 10]
list1.extend(list2)
print(list1)
```

```python
list3 = [11, 12, 13, 14]
list1 = list1 + list3
print(list1)
```

```python
list4 = [0, 1, 2]
print(len(list3))

print("-" * 80)

list5 = ["a", "b", "c", "d", "e", "f"]
print(len(list4))
```

* 딕셔너리 전용 함수
  * keys() : key만 가져오기
  * values() : value만 가져오기
  * items() : key, value 쌍으로 가져오기
  * get() : key로 value 가져오기

```python
dict_data = {'name': '박보검', 'age': 31}

# 전체 key 추출
keys_data = dict_data.keys()
print(f'전체 key 확인: {keys_data}')

# 전체 value 추출
values_data = dict_data.values()
print(f'전체 value 확인: {values_data}')
```

```python
dict_data = {'name': '박보검', 'age': 31, 'phone': '010-1234-1234'}

# 전체 key, value 쌍 추출
items_data = dict_data.items()
print(f'전체 key, value 쌍 확인: {items_data}')

# get() 함수로 value 추출
name = dict_data.get('name')
print(name)
```

* 문자열 전용 함수
  * capitalize() : 문자열의 첫번째 문자를 대문자로 변환
  * upper() : 문자열을 대문자로 변환
  * lower() : 문자열을 소문자로 변환
  * title() : 문장에서 각 단어의 첫벆째 문자를 대문자로 변환
  * split() : 입력한 문자를 기준으로 문자열을 분리하여 리스트의 형태로 출력
  join() : 리스트의 형태로 분리된 문자열을 입력한 문자를 기준으로 연결
  * replace() : 특정 문자(열)를 다른 문자(열)로 치환
  * strip() : 문자열의 시작과 끝에 있는 공백과 줄 바꿈 기호 제거

```python
# capitalize(): 첫 글자만 대문자로 변환
string1 = 'python'
result1 = string1.capitalize()
print(result1)  # 출력: Python
```

```python
# upper(): 전체를 대문자로 변환
result2 = string1.upper()
print(result2)  # 출력: PYTHON
```

```python
# lower(): 전체를 소문자로 변환
string2 = 'PYTHON'
result3 = string2.lower()
print(result3)  # 출력: python
```

```python
# title(): 각 단어의 첫 글자를 대문자로 변환
string3 = 'just do it!'
result4 = string3.title()
print(result4)  # 출력: Just Do It!
```

```python
# split(): 공백/빈칸을 기준으로 문자열 분리
example1 = 'zero one two three'
result5 = example1.split()
print(result5)  # 출력: ['zero', 'one', 'two', 'three']

# split(): 특정 구분자(",")를 기준으로 문자열 분리
example2 = 'python, jquery, javascript'
result6 = example2.split(",")
print(result6)  # 출력: ['python', 'jquery', 'javascript']
```

```python
# join(): 리스트 요소 사이에 공백(1칸)을 넣어 연결
colors = ['red', 'blue', 'green', 'yellow']
result = ' '.join(colors)
print(result)  # 출력: red blue green yellow

# join(): 리스트 요소 사이에 특정 문자(", ")를 넣어 연결
result = ', '.join(colors)
print(result)  # 출력: red, blue, green, yellow
```

```python
string4 = "Life is too short"

print(f'replace 함수 사용 전 문자열 출력 : \n{string4}')
print('-' * 80)

# replace() 적용
result = string4.replace("Life", "My leg")

# 결과 확인
print(f'replace 함수 사용 후 문자열 출력 : \n{result}')
```

```python
string5 = "\n Hello world, Python! \n"

print(f'strip 함수 사용 전 문자열 출력 : \n{string5}')
print('-' * 80)

# strip() 적용
result = string5.strip()

# 결과 확인
print(f'strip 함수 사용 후 문자열 출력 : \n{result}')
```
