## 1. 두 수의 곱을 구하는 사용자 정의 함수
### 두 개의 매개변수 x, y를 전달받아 곱셈 연산 결과를 반환하는 사용자 정의 함수 multiply를 정의하고, multiply(4, 5)를 호출하여 결과를 출력하는 코드를 작성하세요.

```python
def multifly(x, y) :
    result = x * y
    return result

print(multifly(4, 5))
```

## 2. 키워드 인수를 활용한 함수 호출
### 두 매개변수 a, b를 받아 뺄셈 결과(a - b)를 반환하는 함수 subtract를 정의하세요. 그 후 키워드 인수 기법을 사용하여 a=10, b=3을 직접 지정하여 호출한 결과를 출력하는 코드를 작성하세요.

```python
def subtract(x, y) :
    result = x - y
    return result

print(subtract(10, 3))
```

## 3. 거듭제곱을 계산하는 사용자 정의 함수
### 밑(base)과 지수(exponent) 두 개를 매개변수로 받아 제곱 연산자(**)를 이용해 거듭제곱을 계산하고 반환하는 함수 power를 정의하고, power(3, 3)을 호출한 결과를 출력하는 코드를 작성하세요.


```python
def power(x, y) :
    result = x ** y
    return result

print(power(3, 3))
```

## 4. 두 문자열을 연결하여 반환하는 함수
### 두 개의 문자열 str1과 str2를 매개변수로 받아 두 문자열 사이에 공백 한 칸(" ")을 두고 연결한 문자열을 반환하는 함수 concat_words를 정의하고, "Hello"와 "World"를 전달하여 호출 결과를 출력하는 코드를 작성하세요.


```python
def concat_words(str1, str2) :
    result = str1 + " " + str2
    return result

print(concat_words("Hello", "World"))
```

## 23. 사용자 정의 함수 + 문자열 join 융합
### 단어들의 리스트 words와 연결 구분자 sep를 매개변수로 받아서, sep를 기준으로 리스트 원소들을 하나의 문자열로 결합하여 반환하는 함수 custom_join을 정의하세요. ['apple', 'banana', 'orange']와 ", "를 인수로 전달하여 호출한 결과를 출력하는 코드를 작성하세요.

## 24. 사용자 정의 함수 + 나눗셈 몫(//)과 나머지(%) 융합
### 두 정수 num1, num2를 매개변수로 받아 몫 연산자(//)와 나머지 연산자(%)로 나눗셈의 몫과 나머지를 구해 튜플 형태 (몫, 나머지)로 반환하는 함수 get_quotient_remainder를 정의하세요. 23과 5를 전달하여 호출한 결과를 출력하는 코드를 작성하세요.

```python
def get_quotient_remainder(num1, num2) :
    return (num1 // num2, num1 % num2)

print(get_quotient_remainder(23, 5))
```

# 가. 반복문
## 1. 개요
  * 정의 : 특정 명령(코드 블록)을 지정된 조건이나 횟수만큼 반복해서 실행하도록 제어하는 명령문
  * 필요성 : 동일한 작업을 여러 번 수행해야 할 때 코드를 일일이 작성하지 않고 효율적으로 반복 처리하기 위해 사용
  * 종류 : for문, while문

## 2. for 문
  * 정의: 지정된 자료구조에서 요소를 하나씩 가져와 일정 횟수만큼 코드 블록을 반복 실행
  * 특징: 리스트, 튜플, 문자열, 딕셔너리 등 순회 가능한(Iterable) 자료구조에서 순차적으로 데이터를 추출할 때 사용

```python
list_food = ["햄버거", "치킨", "피자"]
for food in list_food:
    print(food)
```

```python
hi = "안녕하세요"
for s in hi:
    print(s)
```

```python
# 1부터 5까지 출력하기
for i in range(1, 6):
    print(i)
```

```python
# 1부터 10까지의 합 구하기
total = 0
for i in range(1, 11):
    total += i

print("1부터 10까지의 합:", total)
```

## 3. while 문
  * 정의: 조건식(Condition Expression)이 참(True)인 동안 코드 블록을 계속 반복 실행
  * 특징
    * 반복 횟수가 명확하지 않을 때 주로 사용
    * 반복 실행 전 조건을 평가하는 선(先) 조건 평가, 후(後) 코드 실행하는 방식
    * 조건문이 계속 참으로 유지되어 무한 루프에 빠지지 않도록 반복 제어 변수의 변화를 주거나 종료 조건을 명시해야 함
    * 반복 제어 (break) : reak 키워드를 사용하면 조건과 상관없이 현재 실행 중인 while 반복문을 즉시 종료

```python
# 초기 값 설정
count = 1 

# 조건 평가
while count <= 5:
    print(count)
    count = count + 1 # 조건이 거짓이 되도록 변화 부여
```

```python
while True:
    user_input = input("사용자: ")
    if user_input.lower() in ["exit", "quit"]:
        print(f"{user_input.lower()}이/가 입력되어 종료합니다.")
        break
    else:
        print("사용자 입력:", user_input)
```

# 나. 조건문
## 1. 개요
  * 조건문은 주어진 조건식의 참(True) 또는 거짓(False) 여부에 따라 프로그램의 실행 흐름을 제어하는 문법
## 2. 핵심 키워드 및 문법
  * 파이썬에서는 if, elif, else 키워드를 사용하여 조건문을 구성
### 가) if 문
  * 가장 기본적인 조건문으로, 조건식이 True일 경우에만 블록 내부의 실행 문장을 수행

```python
score = 95

if score >= 90:
    print("축하합니다! 합격입니다.")

print("검사가 완료되었습니다.")  # if문 외부 코드 (항상 실행됨)
```

### 나) if - else 문
  * 조건식이 True이면 if 블록의 문장1을 실행하고, False이면 else 블록의 문장2를 실행

```python
num = -5

if num >= 0:
    print("양수 또는 0입니다.")
else:
    print("음수입니다.")
```

### 다) if - elif - else 문
  * 여러 개의 조건을 단계별로 검사할 때 사용
  * 첫 번째 조건식이 False일 경우 다음 elif 조건식을 확인하며, 모든 조건이 False일 때 마지막 else 블록을 실행

```python
score = 85

if score >= 90:
    print("A 학점")
elif score >= 80:
    print("B 학점")
elif score >= 70:
    print("C 학점")
else:
    print("F 학점")
```

```python
is_member = True
age = 20

if is_member:
    if age >= 19:
        print("성인 회원입니다. 입장할 수 있습니다.")
    else:
        print("청소년 회원입니다.")
else:
    print("비회원입니다. 로그인 후 이용해주세요.")
```

## 4. 사용자 정의 함수 + 딕셔너리 .items() + if-else + 산술 연산자 융합
### 상품명과 가격이 담긴 딕셔너리(price_dict)와 할인율(discount_rate)을 매개변수로 받는 함수 apply_discount를 정의하세요. 가격이 10000원 이상인 상품만 할인율을 적용하여 정수(int) 할인가로 계산하고, 10000원 미만인 상품은 기존 가격 그대로 유지하여 새로운 딕셔너리로 반환하는 코드를 작성하세요.

```python
def apply_discount(dict, dr) :
    new_dict = {}
    items = dict.items()
    for i in items :
        value = i[1]
        if value >= 10000 :
            new_dict[i[0]] = value * dr * 0.01
        else :
            new_dict[i[0]] = value * 1.0
    return new_dict

price_dict = {"a" : 15000, "b" : 9000}
discount_rate = 50

print(apply_discount(price_dict, discount_rate))
```

## 11. 사용자 정의 함수 + 딕셔너리 .get() + 리스트 인덱싱 + 몫 연산자 // + if-else 융합
### 학생 정보 딕셔너리(student_info = {'name': '김철수', 'scores': [90, 85, 80]})를 매개변수로 받는 함수 check_student_pass를 정의하세요. 함수 내부에서 점수 리스트의 3개 과목 합을 구하고 3으로 나눈 몫(//)으로 평균 정수를 구하세요. 평균이 80점 이상이면 "[name]님 합격 (평균: [avg]점)", 그렇지 않으면 "[name]님 불합격 (평균: [avg]점)" 문자열을 반환하는 코드를 작성하세요.

```python
student_info = {'name': '김철수', 'scores': [90, 85, 80]}

def check_student_pass(si) :
    sum = 0
    for i in si["scores"] :
        sum += i
    result = sum // 3
    return f"[{si["name"]}]님 합격 (평균: [{result}]점)" if result >= 80 else f"[{si["name"]}]님 불합격 (평균: [{result}]점)"

print(check_student_pass(student_info))
```

## 14. 사용자 정의 함수 + 딕셔너리 .values() + for 반복문 융합
### 장바구니 딕셔너리(cart_dict)를 매개변수로 전달받는 함수 calculate_total_price를 정의하세요. 함수 내에서 .values() 함수와 for 반복문을 사용하여 모든 상품 가격을 누적 합산한 총 금액을 반환하고, 이를 호출하여 출력하는 코드를 작성하세요.

```python
def calculate_total_price(cd) :
    price_list = cd.values()
    sum = 0
    for i in price_list :
        sum += i
    return sum

cart_dict = {"a" : 5000, "b" : 7000, "c" : 2000, "d" : 30000}

print(calculate_total_price(cart_dict))
```

## 16. 딕셔너리 .get() + if-else 조건문 + 리스트 .append() 융합
### 그룹 데이터 딕셔너리 group_data = {'A': ['철수'], 'B': ['영희']}가 있습니다. .get('C')를 이용해 Key 'C'가 존재하는지 확인한 후, Key가 없으면(None이면) 'C' 키에 ['민수'] 리스트를 새로 할당하고, Key가 이미 존재하면 해당 리스트에 .append('민수')를 수행하도록 조건을 작성한 뒤 최종 딕셔너리를 출력하세요.
