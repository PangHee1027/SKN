# 가. 예외 처리
## 1. 예외 : 프로그램 실행 중에 발생하는 에러로, 예외가 발생하면 프로그램이 비정상 종료
## 2. 예외 처리 : 예외 발생 시 프로그램이 중단되지 않고 정상적으로 계속 실행될 수 있도록 제어하는 코드를 작성하는 것
### 가) 주요 구문
  * try: 예외가 발생할 가능성이 있는 코드를 실행합니다.
  * except: 예외가 발생했을 때 처리할 코드를 작성합니다.
## 3. 명시적 예외 처리 
  * except에 예외의 이름을 지정해서 해당 예외가 발생했을 때만 프로그램이 중단되지 않도록 처리
  * IndexError (리스트 인덱스 범위 초과)
    * 리스트의 범위를 벗어난 인덱스에 접근할 때 발생하는 예외

```python
numbers = [10, 20, 30]

try:
    print(numbers[5])  # 존재하지 않는 5번 인덱스 접근
except IndexError:
    print("리스트의 인덱스 범위를 벗어났습니다.")
```

  * NameError (존재하지 않는 변수 호출)
    * 정의되지 않거나 존재하지 않는 변수를 사용하려고 할 때 발생하는 예외

```python
try:
    print(undefined_variable)  # 선언되지 않은 변수 호출
except NameError:
    print("존재하지 않는 변수를 호출했습니다.")
```

  * ZeroDivisionError (0으로 숫자를 나눔)
    * 숫자를 0으로 나누려고 할 때 발생하는 예외

```python
def div_ten(x):
    try:
        return 10 / x
    except ZeroDivisionError:
        return "0으로 나눌 수 없습니다."

# 정상 작동
print(div_ten(2))  # 결과: 5.0

# 예외 발생 및 처리
print(div_ten(0))  # 결과: 0으로 나눌 수 없습니다.
```

  * ValueError (변환할 수 없는 값 사용)
    * 함수의 인자로 올바른 타입이지만 부적절한 값을 전달할 때 발생하는 예외

```python
try:
    num = int("abc")  # 문자를 정수로 변환 시도
except ValueError:
    print("변환할 수 없는 문자나 숫자 타입입니다.")
```

  * FileNotFoundError (존재하지 않는 파일 열기)
    * 지정한 경로에 파일이 존재하지 않을 때 발생하는 예외

```python
try:
    with open("not_exists.txt", "r") as f:
        content = f.read()
except FileNotFoundError:
    print("존재하지 않는 파일을 호출했습니다.")
```

## 4. 포괄적 예외 처리
  * 발생할 수 있는 모든 오류를 한번에 처리하는 방식

```python
def div_ten_all(x):
    try:
        return 10 / x
    except Exception as e:
        return f"에러가 발생했습니다: {e}"

print(div_ten_all(0))  # 결과: 에러가 발생했습니다: division by zero
```

# 나. 파일 처리
## 1. 개요 : 파이썬에서 파일을 다루는 과정은 크게 파일 열기(모드 설정) $\rightarrow$ 내용 읽기 또는 쓰기 $\rightarrow$ 파일 닫기의 순서
## 2. 열기 모드
  * r (Read, 읽기 모드): 기존에 존재하는 파일을 읽기만 할 때 사용
  * w (Write, 쓰기 모드): 새로운 파일을 생성하거나 내용을 작성할 때 사용
    * 주의: 만약 동일한 이름의 파일이 이미 존재할 경우, 기존 파일을 무시하고 내용을 새로 입력(덮어쓰기)
  * a (Add, 추가모드): 파일의 마지막에 새로운 내용을 추가할 때 사용
