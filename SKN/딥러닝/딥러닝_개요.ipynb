# 딥러닝 (Deep Learning)

## 가. 딥러닝 개요
### 1. **딥러닝(Deep Learning)의 정의**
* 인간의 뇌 신경망 구조를 모방한 인공신경망(Artificial Neural Network)을 기반으로, 대량의 데이터에서 스스로 특징을 학습하는 머신러닝의 한 분야

### 2. **머신러닝 vs 딥러닝 비교**
| 비교 항목 | 머신러닝 (Machine Learning) | 딥러닝 (Deep Learning) |
| :--- | :--- | :--- |
| **특징 추출** | 사람이 직접 정의 (Feature Engineering) | 모델이 데이터로부터 스스로 학습 |
| **데이터 의존도** | 적은 양의 데이터로도 유의미한 성능 확보 가능 | 데이터가 많을수록 성능이 극대화됨 |
| **학습 시간** | 수초~수 시간 내 학습 완료 (빠름) | 수일~수 주일 이상 소요 (무거움) |
| **연산 자원** | 일반적인 CPU 기반 연산 가능 | GPU/TPU 등의 병렬 연산 장치 필수 |
| **대표 알고리즘** | 결정 트리, 로지스틱 회귀 등 | CNN, RNN, LSTM, Transformer, BERT 등 |

## 나. 딥러닝 핵심 기술 및 오픈 프레임워크
### 1. **주요 알고리즘**
* **CNN (합성곱 신경망)**: 이미지의 공간적 구조를 유지하며 유의미한 특징을 추출하는 필터 기반 알고리즘 (컴퓨터 비전 분야)
* **RNN / LSTM (순환 신경망)**: 데이터의 순서와 시간적 흐름을 기억하기 위해 은닉층의 출력을 다시 입력으로 순환시키는 구조 (시계열 데이터 및 초기 자연어 처리)
* **Transformer / BERT**: Self-Attention 메커니즘을 기반으로 문장 내 단어 간 관계를 동시에 파악 (거대 언어 모델 LLM)

### 2. **오픈 프레임워크 비교**
* **TensorFlow & Keras (구글 주도)**: 직관적인 코드 구조를 제공하여 입문 및 프로토타이핑에 매우 유리
* **PyTorch (메타 주도)**: 글로벌 표준

## 다. 퍼셉트론 (Perceptron)
### 1. **퍼셉트론의 개념과 의미**
* **퍼셉트론(인공 뉴런) 정의**: 다수의 입력 신호를 받아 하나의 신호를 출력하는 인공 뉴런
* **가중치(Weight)**: 뉴런의 출력 신호를 낼지 말지 결정하기 위해 입력 신호에 곱하는 계수
* **편향(Bias)**: 현대 연산 표준에 맞춰 도입된 수식의 상수항

### 2. **수학적 정의 및 의의**
* **수식**: $y = w_1x_1 + w_2x_2 + \dots + w_nx_n + b$
* **이진 분류기**: 0 또는 1을 판별하며 이진 분류의 '결정 경계(Decision Boundary)'로 작동하는 선형 모델
* **빌딩 블록**: 퍼셉트론 수십억 개가 유기적으로 연결된 현대 심층신경망(DNN)의 구조적 모태

## 라. 논리 게이트 (Logic Gate)와 단층 퍼셉트론의 한계
### 1. **논리 게이트 개념과 종류**
* **개념**: 컴퓨터의 두뇌(CPU)를 구성하는 가장 작은 단위의 스위치로, 특정 조건 만족 시 신호를 통과시키거나 차단함
* **종류**:
  * **AND (논리곱)**: 모든 입력이 1일 때만 1 출력 (**선형 분리 가능**)
  * **OR (논리합)**: 입력 중 하나라도 1이면 1 출력 (**선형 분리 가능**)
  * **NAND (부정논리곱)**: AND 연산 결과를 뒤집어 모든 입력이 1일 때만 0 출력 (**선형 분리 가능**)
  * **XOR (배타적논리합)**: 입력값이 서로 다를 때만 1 출력 (**선형 분리 불가능**)

### 2. **XOR 문제와 인공지능의 역사 (제1차 AI 겨울)**
* XOR 게이트는 데이터 분포가 대각선으로 엇갈려 있어 단 하나의 '직선(단층 퍼셉트론)'으로는 0과 1을 분리할 수 없음
* 1969년 마빈 민스키 교수가 이를 수학적으로 증명하면서 인공지능 연구가 중단된 **'제1차 인공지능의 겨울'**이 도래함
* 이후 층을 여러 개 쌓은 **다층 퍼셉트론(MLP)**과 **오차역전파** 알고리즘이 등장하며 이 한계를 극복하고 현대 딥러닝으로 발전하는 터닝 포인트가 됨

```python
# 퍼셉트론 실습

import numpy as np
```

```python
def AND(x1, x2) :
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7

    output = np.sum(x * w) + b

    return 1 if output >= 0 else 0

print(f"x1 = 1, x2 = 1 -> {AND(1, 1)}")
print(f"x1 = 1, x2 = 0 -> {AND(1, 0)}")
print(f"x1 = 0, x2 = 1 -> {AND(0, 1)}")
print(f"x1 = 0, x2 = 0 -> {AND(0, 0)}")
```

```python
def OR(x1, x2) :
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.2

    output = np.sum(x * w) + b

    return 1 if output >= 0 else 0

print(f"x1 = 1, x2 = 1 -> {OR(1, 1)}")
print(f"x1 = 1, x2 = 0 -> {OR(1, 0)}")
print(f"x1 = 0, x2 = 1 -> {OR(0, 1)}")
print(f"x1 = 0, x2 = 0 -> {OR(0, 0)}")
```

```python
def NAND(x1, x2) :
    x = np.array([-x1, -x2])
    w = np.array([0.5, 0.5])
    b = 0.7

    output = np.sum(x * w) + b

    return 1 if output >= 0 else 0

print(f"x1 = 1, x2 = 1 -> {NAND(1, 1)}")
print(f"x1 = 1, x2 = 0 -> {NAND(1, 0)}")
print(f"x1 = 0, x2 = 1 -> {NAND(0, 1)}")
print(f"x1 = 0, x2 = 0 -> {NAND(0, 0)}")
```

```python
def NAND(x1, x2) :
    output = -AND(x1, x2)

    return 1 if output >= 0 else 0

print(f"x1 = 1, x2 = 1 -> {NAND(1, 1)}")
print(f"x1 = 1, x2 = 0 -> {NAND(1, 0)}")
print(f"x1 = 0, x2 = 1 -> {NAND(0, 1)}")
print(f"x1 = 0, x2 = 0 -> {NAND(0, 0)}")
```

```python
def XOR(x1, x2) :
    output = AND(NAND(x1, x2), OR(x1, x2)) 

    return 1 if output > 0 else 0

print(f"x1 = 1, x2 = 1 -> {XOR(1, 1)}")
print(f"x1 = 1, x2 = 0 -> {XOR(1, 0)}")
print(f"x1 = 0, x2 = 1 -> {XOR(0, 1)}")
print(f"x1 = 0, x2 = 0 -> {XOR(0, 0)}")
```

```python
for x1, x2 in[(0, 0), (0, 1), (1, 0), (1, 1)] :
    print(f"AND({x1}, {x2}) = {AND(x1, x2)}")
    print(f"OR({x1}, {x2}) = {OR(x1, x2)}")
    print(f"NAND({x1}, {x2}) = {NAND(x1, x2)}")
    print(f"XOR({x1}, {x2}) = {XOR(x1, x2)}")
    print("-" * 80)
```

## 마. 활성화 함수 (Activation Function)의 개념과 필요성
### 1. **활성화 함수 정의**
* 입력 신호의 총합($w_1x_1 + w_2x_2 + b$)을 받아, 다음 뉴런으로 보낼 출력 신호의 형태로 변환하고 결정하는 함수

### 2. **활성화 함수가 반드시 필요한 이유**
* **선형 연산의 한계 극복**: 퍼셉트론의 내부 연산은 공간에 직선을 긋는 선형 연산이므로, 활성화 함수 없이 층만 깊게 쌓아도 결국 하나의 큰 선형 함수($y = Wx + B$)가 됨
* **비선형 데이터 처리**: 퍼셉트론이 계산한 직선 신호를 활성화 함수로 구부리고 꺾어 줌으로써, 실타래처럼 꼬인 현실의 비선형 데이터를 정교하게 분류하는 곡면 경계선을 만들어 냄

## 바. 주요 활성화 함수 비교
### 1. **계단 함수 (Step Function)**
* **정의**: 입력값이 0을 넘으면 1을 출력하고, 0 이하이면 0을 출력하는 가장 기본적인 이분법적 함수 ($h(x) = \begin{cases} 0 & (x \le 0) \\ 1 & (x > 0) \end{cases}$)
* **한계**: 0인 지점에서 미분이 불가능하고, 그 외 구간에서는 기울기가 0이 되어 오차를 수정하는 경사하강법을 적용할 수 없음

```python
def step_function(input_arr) :
    return np.where(input_arr >= 0, 1, 0)
```

```python
def perceptron1(x1, x2) :
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7
    output = np.sum(x * w) + b
    return output
```

```python
result = []
for x1, x2 in ((0, 0), (1, 0), (0, 1), (1, 1)) :
    result.append(perceptron1(x1, x2))

print(step_function(np.array(result)))
```

### 2. **시그모이드 함수 (Sigmoid Function)**
* **정의**: 입력에 따라 출력이 0과 1 사이의 S자 곡선 형태로 부드럽게 변하는 함수 ($\phi(z) = \frac{1}{1 + e^{-z}}$)
* **특징과 의의**: 출력을 '확률(Probability)'로 해석할 수 있으며, 부드러운 미분이 가능하여 오차역전파 학습의 기틀을 마련함
* **한계**: 입력값이 너무 크거나 작으면 기울기가 0에 수렴하는 **포화(Saturation) 현상**이 발생하여 **기울기 소실(Vanishing Gradient)** 문제를 야기함

```python
def sigmoid(input):
  return 1 / (1 + np.exp(-input))
```

```python
def perceptron2(x1, x2) :
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7
    output = np.sum(x * w) + b
    return output
```

```python
result = []
for x1, x2 in ((0, 0), (1, 0), (0, 1), (1, 1)) :
    result.append(perceptron2(x1, x2))

print(sigmoid(np.array(result)))
```

### 3. **ReLU 함수 (Rectified Linear Unit)**
* **정의**: 입력이 0을 넘으면 그대로 통과시키고, 0 이하이면 0으로 차단하는 함수 ($R(z) = \max(0, z)$)
* **표준인 이유**: 
  * **경사 소실 문제 해결**: 양수 영역에서 기울기가 항상 1로 유지되어 층을 깊게 쌓아도 오차가 잘 전달됨
  * **압도적인 연산 속도**: 복잡한 지수 연산 없이 단순 비교만 수행하므로 속도가 빠름

```python
def ReLU(input) :
    return max(0, input)
```

```python
def perceptron3(x1, x2) :
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7
    output = np.sum(x * w) + b
    return output
```

```python
result = []
for x1, x2 in ((0, 0), (1, 0), (0, 1), (1, 1)) :
    result.append(ReLU(perceptron3(x1, x2)))

print(result)
```

## 사. 다차원 배열의 계산과 신경망 Layer
### 1. **행렬 곱 / 벡터 내적 (Matrix Multiplication / Vector Dot Product)**
* 신경망 계산을 다차원 배열(행렬)로 묶어 처리하면, 개별 수식을 일일이 계산하는 것보다 단 한 번의 행렬 곱셈 연산으로 층(Layer) 전체의 계산을 효율적으로 끝낼 수 있다.
* **행렬 곱의 일반화 조건**: 행렬 A의 크기가 $(m, k)$이고 행렬 B의 크기가 $(k, n)$일 때, 두 행렬의 곱 $C = A \cdot B$의 결과 크기는 $(m, n)$이 된다.
* **신경망의 연산 공식**: 입력 벡터 $X$, 가중치 행렬 $W$, 편향 벡터 $B$가 주어졌을 때의 출력 레이어 연산 공식은 $Y = X \cdot W + B$이다.

```python
# 다차원 배열의 계산 실습 (행렬 곱 / 신경망 Layer 내적)
import numpy as np

# 입력 데이터 X와 가중치 W, 편향 B 정의
X = np.array([1, 2])                  # 입력 특성 수 d = 2
W = np.array([[1, 0, 1], [2, 3, 1]])  # 가중치 행렬 (2, 3)
B = np.array([1, 2, 3])               # 편향 벡터 (3,)

# 행렬 내적 및 편향 덧셈 연산
Y = np.dot(X, W) + B

print(f"입력 행렬 X Shape : {X.shape}")
print(f"가중치 행렬 W Shape : {W.shape}")
print(f"편향 벡터 B Shape : {B.shape}")
print(f"최종 출력 Y : {Y}")
print(f"최종 출력 Y Shape : {Y.shape}")
```

## 아. 출력층(Output Layer)의 역할과 설계
### 1. **출력층의 정의**
* 인공신경망의 가장 마지막 단계로, 앞선 은닉층들을 거쳐 정제되어 흘러온 신호들을 최종 종합하여 원하는 예측 결과물을 생성해 내는 곳
* 딥러닝의 문제 유형에 따라 출력층에 적용해야 하는 활성화 함수와 노드의 구조가 완전히 달라짐

### 2. **문제 유형에 따른 출력층 매핑 규칙**
* **회귀 (Regression) $\rightarrow$ 항등 함수 (Identity Function) 사용**: 연속적인 수치(주택 가격, 기온 등)를 왜곡 없이 그대로 맞히는 문제
* **이진 분류 (Binary Classification) $\rightarrow$ 시그모이드 함수 (Sigmoid Function) 사용**: 둘 중 하나를 판별하여 '정답일 확률'을 도출하는 문제
* **다중 클래스 분류 (Multi-class Classification) $\rightarrow$ 소프트맥스 함수 (Softmax Function) 사용**: 세 개 이상의 후보군 중 가장 정답 같은 대상을 맞히는 문제

## 자. 주요 출력층 활성화 함수
### 1. **항등 함수 (Identity Function)**
* **정의**: 입력된 신호를 아무런 변형이나 필터링 없이 그대로 다음 단계로 출력하는 함수 ($f(x) = x$)

### 2. **시그모이드 함수 (Sigmoid Function)**
* **정의**: 은닉층에서 넘어온 최종 결과를 0과 1 사이의 S자 곡선 값으로 압축하는 함수 ($h(x) = \frac{1}{1 + e^{-x}}$)

### 3. **소프트맥스 함수 (Softmax Function)**
* **정의**: $n$개의 여러 출력 노드가 있을 때, 각 노드의 지수 함수 값을 전체 노드 지수 함수의 총합으로 나누어 정규화하는 함수 ($y_k = \frac{\exp(a_k)}{\sum_{i=1}^{n} \exp(a_i)}$)
* **2대 핵심 특성**: 출력값 범위의 정규화($0 < y_k < 1$) 및 총합의 법칙(총합은 항상 1)

```python
# 주요 출력층 활성화 함수 구현

# 1. 항등 함수
def identity_function(x):
    return x

# 2. 시그모이드 함수
def sigmoid_output(x):
    return 1 / (1 + np.exp(-x))

# 3. 소프트맥스 함수 (오버플로 대책 적용)
def softmax(a):
    c = np.max(a)
    exp_a = np.exp(a - c)
    sum_exp_a = np.sum(exp_a)
    y = exp_a / sum_exp_a
    return y
```

## 차. 출력층의 뉴런 수 정하기 (Design Rule) 및 실습
* **이진 분류**: 스팸 메일 판별 등 $\rightarrow$ 출력층 노드 수: **1개** (시그모이드 함수 연결)
* **다중 분류**: MNIST 손글씨 숫자(0~9) 등 $\rightarrow$ 출력층 노드 수: **10개** (소프트맥스 함수 연결)
* **회귀**: 주택 가격 예측 등 $\rightarrow$ 출력층 노드 수: **1개** (항등 함수 연결 또는 없음)

```python
# 출력층 설계하기 실습

# 1. 이진 분류 출력층 모의 테스트 (노드 1개, 시그모이드)
binary_input = np.array([1.5])
binary_pred = sigmoid_output(binary_input)
print(f"이진 분류 예측 확률 (시그모이드) : {binary_pred[0]:.4f}")

# 2. 다중 분류 출력층 모의 테스트 (노드 3개, 소프트맥스)
multi_input = np.array([2.5, 1.0, -1.5])
multi_pred = softmax(multi_input)
print(f"다중 분류 예측 확률 분포 (소프트맥스) : {multi_pred}")
print(f"출력 확률의 총합 : {np.sum(multi_pred)}")
```
