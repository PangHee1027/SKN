# 딥러닝 (Deep Learning)

## 가. 딥러닝 개요
### 1. **딥러닝(Deep Learning)의 정의**
* 인간의 뇌 신경망 구조를 모방한 인공신경망(Artificial Neural Network)을 기반으로, 대량의 데이터에서 스스로 특징을 학습하는 머신러닝의 한 분야

### 2. **머신러닝 vs 딥러닝 비교**
| 비교 항목 | 머신러닝 (Machine Learning) | 딥러닝 (Deep Learning) |
| :--- | :--- | :--- |
| **특징 추출** | 사람이 직접 정의 (Feature Engineering) | 모델이 데이터로부터 스스로 학습 |
| **데이터 의존도** | 적은 양의 데이터로도 유의미한 성능 확보 가능 | 데이터가 많을수록 성능이 극대화됨 |
| **학습 시간** | 수초~수 시간 내 학습 완료 (빠름) | 수일~수 주일 이상 소요 (무거움) |
| **연산 자원** | 일반적인 CPU 기반 연산 가능 | GPU/TPU 등의 병렬 연산 장치 필수 |
| **대표 알고리즘** | 결정 트리, 로지스틱 회귀 등 | CNN, RNN, LSTM, Transformer, BERT 등 |

## 나. 딥러닝 핵심 기술 및 오픈 프레임워크
### 1. **주요 알고리즘**
* **CNN (합성곱 신경망)**: 이미지의 공간적 구조를 유지하며 유의미한 특징을 추출하는 필터 기반 알고리즘 (컴퓨터 비전 분야)
* **RNN / LSTM (순환 신경망)**: 데이터의 순서와 시간적 흐름을 기억하기 위해 은닉층의 출력을 다시 입력으로 순환시키는 구조 (시계열 데이터 및 초기 자연어 처리)
* **Transformer / BERT**: Self-Attention 메커니즘을 기반으로 문장 내 단어 간 관계를 동시에 파악 (거대 언어 모델 LLM)

### 2. **오픈 프레임워크 비교**
* **TensorFlow & Keras (구글 주도)**: 직관적인 코드 구조를 제공하여 입문 및 프로토타이핑에 매우 유리
* **PyTorch (메타 주도)**: 글로벌 표준

## 다. 퍼셉트론 (Perceptron)
### 1. **퍼셉트론의 개념과 의미**
* **퍼셉트론(인공 뉴런) 정의**: 다수의 입력 신호를 받아 하나의 신호를 출력하는 인공 뉴런
* **가중치(Weight)**: 뉴런의 출력 신호를 낼지 말지 결정하기 위해 입력 신호에 곱하는 계수
* **편향(Bias)**: 현대 연산 표준에 맞춰 도입된 수식의 상수항

### 2. **수학적 정의 및 의의**
* **수식**: $y = w_1x_1 + w_2x_2 + \dots + w_nx_n + b$
* **이진 분류기**: 0 또는 1을 판별하며 이진 분류의 '결정 경계(Decision Boundary)'로 작동하는 선형 모델
* **빌딩 블록**: 퍼셉트론 수십억 개가 유기적으로 연결된 현대 심층신경망(DNN)의 구조적 모태

## 라. 논리 게이트 (Logic Gate)와 단층 퍼셉트론의 한계
### 1. **논리 게이트 개념과 종류**
* **개념**: 컴퓨터의 두뇌(CPU)를 구성하는 가장 작은 단위의 스위치로, 특정 조건 만족 시 신호를 통과시키거나 차단함
* **종류**:
  * **AND (논리곱)**: 모든 입력이 1일 때만 1 출력 (**선형 분리 가능**)
  * **OR (논리합)**: 입력 중 하나라도 1이면 1 출력 (**선형 분리 가능**)
  * **NAND (부정논리곱)**: AND 연산 결과를 뒤집어 모든 입력이 1일 때만 0 출력 (**선형 분리 가능**)
  * **XOR (배타적논리합)**: 입력값이 서로 다를 때만 1 출력 (**선형 분리 불가능**)

### 2. **XOR 문제와 인공지능의 역사 (제1차 AI 겨울)**
* XOR 게이트는 데이터 분포가 대각선으로 엇갈려 있어 단 하나의 '직선(단층 퍼셉트론)'으로는 0과 1을 분리할 수 없음
* 1969년 마빈 민스키 교수가 이를 수학적으로 증명하면서 인공지능 연구가 중단된 **'제1차 인공지능의 겨울'**이 도래함
* 이후 층을 여러 개 쌓은 **다층 퍼셉트론(MLP)**과 **오차역전파** 알고리즘이 등장하며 이 한계를 극복하고 현대 딥러닝으로 발전하는 터닝 포인트가 됨

```python
# 퍼셉트론 실습

import numpy as np
```

```python
def AND(x1, x2) :
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7

    output = np.sum(x * w) + b

    return 1 if output >= 0 else 0

print(f"x1 = 1, x2 = 1 -> {AND(1, 1)}")
print(f"x1 = 1, x2 = 0 -> {AND(1, 0)}")
print(f"x1 = 0, x2 = 1 -> {AND(0, 1)}")
print(f"x1 = 0, x2 = 0 -> {AND(0, 0)}")
```

```python
def OR(x1, x2) :
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.2

    output = np.sum(x * w) + b

    return 1 if output >= 0 else 0

print(f"x1 = 1, x2 = 1 -> {OR(1, 1)}")
print(f"x1 = 1, x2 = 0 -> {OR(1, 0)}")
print(f"x1 = 0, x2 = 1 -> {OR(0, 1)}")
print(f"x1 = 0, x2 = 0 -> {OR(0, 0)}")
```

```python
def NAND(x1, x2) :
    x = np.array([-x1, -x2])
    w = np.array([0.5, 0.5])
    b = 0.7

    output = np.sum(x * w) + b

    return 1 if output >= 0 else 0

print(f"x1 = 1, x2 = 1 -> {NAND(1, 1)}")
print(f"x1 = 1, x2 = 0 -> {NAND(1, 0)}")
print(f"x1 = 0, x2 = 1 -> {NAND(0, 1)}")
print(f"x1 = 0, x2 = 0 -> {NAND(0, 0)}")
```

```python
def NAND(x1, x2) :
    output = -AND(x1, x2)

    return 1 if output >= 0 else 0

print(f"x1 = 1, x2 = 1 -> {NAND(1, 1)}")
print(f"x1 = 1, x2 = 0 -> {NAND(1, 0)}")
print(f"x1 = 0, x2 = 1 -> {NAND(0, 1)}")
print(f"x1 = 0, x2 = 0 -> {NAND(0, 0)}")
```

```python
def XOR(x1, x2) :
    output = AND(NAND(x1, x2), OR(x1, x2)) 

    return 1 if output > 0 else 0

print(f"x1 = 1, x2 = 1 -> {XOR(1, 1)}")
print(f"x1 = 1, x2 = 0 -> {XOR(1, 0)}")
print(f"x1 = 0, x2 = 1 -> {XOR(0, 1)}")
print(f"x1 = 0, x2 = 0 -> {XOR(0, 0)}")
```

```python
for x1, x2 in[(0, 0), (0, 1), (1, 0), (1, 1)] :
    print(f"AND({x1}, {x2}) = {AND(x1, x2)}")
    print(f"OR({x1}, {x2}) = {OR(x1, x2)}")
    print(f"NAND({x1}, {x2}) = {NAND(x1, x2)}")
    print(f"XOR({x1}, {x2}) = {XOR(x1, x2)}")
    print("-" * 80)
```

## 마. 활성화 함수 (Activation Function)의 개념과 필요성
### 1. **활성화 함수 정의**
* 입력 신호의 총합($w_1x_1 + w_2x_2 + b$)을 받아, 다음 뉴런으로 보낼 출력 신호의 형태로 변환하고 결정하는 함수

### 2. **활성화 함수가 반드시 필요한 이유**
* **선형 연산의 한계 극복**: 퍼셉트론의 내부 연산은 공간에 직선을 긋는 선형 연산이므로, 활성화 함수 없이 층만 깊게 쌓아도 결국 하나의 큰 선형 함수($y = Wx + B$)가 됨
* **비선형 데이터 처리**: 퍼셉트론이 계산한 직선 신호를 활성화 함수로 구부리고 꺾어 줌으로써, 실타래처럼 꼬인 현실의 비선형 데이터를 정교하게 분류하는 곡면 경계선을 만들어 냄

## 바. 주요 활성화 함수 비교
### 1. **계단 함수 (Step Function)**
* **정의**: 입력값이 0을 넘으면 1을 출력하고, 0 이하이면 0을 출력하는 가장 기본적인 이분법적 함수 ($h(x) = \begin{cases} 0 & (x \le 0) \\ 1 & (x > 0) \end{cases}$)
* **한계**: 0인 지점에서 미분이 불가능하고, 그 외 구간에서는 기울기가 0이 되어 오차를 수정하는 경사하강법을 적용할 수 없음

```python
def step_function(input_arr) :
    return np.where(input_arr >= 0, 1, 0)
```

```python
def perceptron1(x1, x2) :
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7
    output = np.sum(x * w) + b
    return output
```

```python
result = []
for x1, x2 in ((0, 0), (1, 0), (0, 1), (1, 1)) :
    result.append(perceptron1(x1, x2))

print(step_function(np.array(result)))
```

### 2. **시그모이드 함수 (Sigmoid Function)**
* **정의**: 입력에 따라 출력이 0과 1 사이의 S자 곡선 형태로 부드럽게 변하는 함수 ($\phi(z) = \frac{1}{1 + e^{-z}}$)
* **특징과 의의**: 출력을 '확률(Probability)'로 해석할 수 있으며, 부드러운 미분이 가능하여 오차역전파 학습의 기틀을 마련함
* **한계**: 입력값이 너무 크거나 작으면 기울기가 0에 수렴하는 **포화(Saturation) 현상**이 발생하여 **기울기 소실(Vanishing Gradient)** 문제를 야기함

```python
def sigmoid(input):
  return 1 / (1 + np.exp(-input))
```

```python
def perceptron2(x1, x2) :
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7
    output = np.sum(x * w) + b
    return output
```

```python
result = []
for x1, x2 in ((0, 0), (1, 0), (0, 1), (1, 1)) :
    result.append(perceptron2(x1, x2))

print(sigmoid(np.array(result)))
```

### 3. **ReLU 함수 (Rectified Linear Unit)**
* **정의**: 입력이 0을 넘으면 그대로 통과시키고, 0 이하이면 0으로 차단하는 함수 ($R(z) = \max(0, z)$)
* **표준인 이유**: 
  * **경사 소실 문제 해결**: 양수 영역에서 기울기가 항상 1로 유지되어 층을 깊게 쌓아도 오차가 잘 전달됨
  * **압도적인 연산 속도**: 복잡한 지수 연산 없이 단순 비교만 수행하므로 속도가 빠름

```python
def ReLU(input) :
    return max(0, input)
```

```python
def perceptron3(x1, x2) :
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7
    output = np.sum(x * w) + b
    return output
```

```python
result = []
for x1, x2 in ((0, 0), (1, 0), (0, 1), (1, 1)) :
    result.append(ReLU(perceptron3(x1, x2)))

print(result)
```

## 사. 다차원 배열의 계산과 신경망 Layer
### 1. **행렬 곱 / 벡터 내적 (Matrix Multiplication / Vector Dot Product)**
* 신경망 계산을 다차원 배열(행렬)로 묶어 처리하면, 개별 수식을 일일이 계산하는 것보다 단 한 번의 행렬 곱셈 연산으로 층(Layer) 전체의 계산을 효율적으로 끝냄
* **행렬 곱의 일반화 조건**: 행렬 A의 크기가 $(m, k)$이고 행렬 B의 크기가 $(k, n)$일 때, 두 행렬의 곱 $C = A \cdot B$의 결과 크기는 $(m, n)$
* **신경망의 연산 공식**: 입력 벡터 $X$, 가중치 행렬 $W$, 편향 벡터 $B$가 주어졌을 때의 출력 레이어 연산 공식은 $Y = X \cdot W + B$

```python
# 다차원 배열의 계산 실습 (행렬 곱 / 신경망 Layer 내적)
import numpy as np

# 입력 데이터 X와 가중치 W, 편향 B 정의
X = np.array([1, 2])                  # 입력 특성 수 d = 2
W = np.array([[1, 0, 1], [2, 3, 1]])  # 가중치 행렬 (2, 3)
B = np.array([1, 2, 3])               # 편향 벡터 (3,)

# 행렬 내적 및 편향 덧셈 연산
Y = np.dot(X, W) + B

print(f"입력 행렬 X Shape : {X.shape}")
print(f"가중치 행렬 W Shape : {W.shape}")
print(f"편향 벡터 B Shape : {B.shape}")
print(f"최종 출력 Y : {Y}")
print(f"최종 출력 Y Shape : {Y.shape}")
```

## 아. 출력층(Output Layer)의 역할과 설계
### 1. **출력층의 정의**
* 인공신경망의 가장 마지막 단계로, 앞선 은닉층들을 거쳐 정제되어 흘러온 신호들을 최종 종합하여 원하는 예측 결과물을 생성해 내는 곳
* 딥러닝의 문제 유형에 따라 출력층에 적용해야 하는 활성화 함수와 노드의 구조가 완전히 달라짐

### 2. **문제 유형에 따른 출력층 매핑 규칙**
* **회귀 (Regression) $\rightarrow$ 항등 함수 (Identity Function) 사용**: 연속적인 수치(주택 가격, 기온 등)를 왜곡 없이 그대로 맞히는 문제
* **이진 분류 (Binary Classification) $\rightarrow$ 시그모이드 함수 (Sigmoid Function) 사용**: 둘 중 하나를 판별하여 '정답일 확률'을 도출하는 문제
* **다중 클래스 분류 (Multi-class Classification) $\rightarrow$ 소프트맥스 함수 (Softmax Function) 사용**: 세 개 이상의 후보군 중 가장 정답 같은 대상을 맞히는 문제

## 자. 주요 출력층 활성화 함수
### 1. **항등 함수 (Identity Function)**
* **정의**: 입력된 신호를 아무런 변형이나 필터링 없이 그대로 다음 단계로 출력하는 함수 ($f(x) = x$)

### 2. **시그모이드 함수 (Sigmoid Function)**
* **정의**: 은닉층에서 넘어온 최종 결과를 0과 1 사이의 S자 곡선 값으로 압축하는 함수 ($h(x) = \frac{1}{1 + e^{-x}}$)

### 3. **소프트맥스 함수 (Softmax Function)**
* **정의**: $n$개의 여러 출력 노드가 있을 때, 각 노드의 지수 함수 값을 전체 노드 지수 함수의 총합으로 나누어 정규화하는 함수 ($y_k = \frac{\exp(a_k)}{\sum_{i=1}^{n} \exp(a_i)}$)
* **2대 핵심 특성**: 출력값 범위의 정규화($0 < y_k < 1$) 및 총합의 법칙(총합은 항상 1)

```python
# 주요 출력층 활성화 함수 구현

# 1. 항등 함수
def identity_function(x):
    return x

# 2. 시그모이드 함수
def sigmoid_output(x):
    return 1 / (1 + np.exp(-x))

# 3. 소프트맥스 함수 (오버플로 대책 적용)
def softmax(a):
    c = np.max(a)
    exp_a = np.exp(a - c)
    sum_exp_a = np.sum(exp_a)
    y = exp_a / sum_exp_a
    return y
```

## 차. 출력층의 뉴런 수 정하기 (Design Rule) 및 실습
* **이진 분류**: 스팸 메일 판별 등 $\rightarrow$ 출력층 노드 수: **1개** (시그모이드 함수 연결)
* **다중 분류**: MNIST 손글씨 숫자(0~9) 등 $\rightarrow$ 출력층 노드 수: **10개** (소프트맥스 함수 연결)
* **회귀**: 주택 가격 예측 등 $\rightarrow$ 출력층 노드 수: **1개** (항등 함수 연결 또는 없음)

```python
# 출력층 설계하기 실습

# 1. 이진 분류 출력층 모의 테스트 (노드 1개, 시그모이드)
binary_input = np.array([1.5])
binary_pred = sigmoid_output(binary_input)
print(f"이진 분류 예측 확률 (시그모이드) : {binary_pred[0]:.4f}")

# 2. 다중 분류 출력층 모의 테스트 (노드 3개, 소프트맥스)
multi_input = np.array([2.5, 1.0, -1.5])
multi_pred = softmax(multi_input)
print(f"다중 분류 예측 확률 분포 (소프트맥스) : {multi_pred}")
print(f"출력 확률의 총합 : {np.sum(multi_pred)}")
```

## 카. 손실 함수 (Loss Function)의 개념과 학습의 목적
### 1. **손실 함수의 정의와 역할**
* **손실(Loss)의 본질**: 딥러닝 모델이 예측한 값($y$)과 실제 데이터가 가진 정답 레이블($t$) 간의 오차 크기
* **나쁨의 지표**: 모델의 현재 성능이 얼마나 '나쁜가'를 정량적으로 나타내는 지표로, 수많은 오차의 크기를 단 하나의 명확한 스칼라(실수) 값으로 계산함
* **딥러닝의 목적**: 이 손실 함수의 출력값을 최소화(0에 가깝게) 만드는 최적의 가중치 매개변수 조합을 찾아내는 것이 바로 딥러닝 **'학습(Learning)'**의 본질이다.

### 2. **문제 유형별 대표 손실 함수**
* **회귀 문제 $\rightarrow$ 평균 제곱 오차 (MSE, Mean Squared Error)**:
  * 수식: $E = \frac{1}{n} \sum_{i=1}^{n} (y_i - t_i)^2$
  * 특징: 오차의 부호를 제거하고, 오차가 클수록 제곱 연산에 의해 기하급수적으로 큰 페널티를 부여함
* **분류 문제 $\rightarrow$ 교차 엔트로피 오차 (CEE, Cross-Entropy Error)**:
  * 수식: $E = -\sum_{k} t_k \log(y_k)$
  * 특징: 정답 레이블의 분포와 모델이 예측한 확률 분포 사이의 수학적 거리를 계산함 (원-핫 인코딩 적용)

```python
# 손실 함수(Loss Function) 실습

# 1. 평균 제곱 오차 (MSE) 구현
def mean_squared_error(y, t):
    return 0.5 * np.sum((y - t)**2)

# 2. 교차 엔트로피 오차 (CEE) 구현
def cross_entropy_error(y, t):
    delta = 1e-7  # 로그에 0이 들어가 마이너스 무한대가 되는 것을 방지
    return -np.sum(t * np.log(y + delta))

# 모의 데이터 테스트
t_sample = np.array([0, 1, 0])  # 정답 (고양이)
y_sample = np.array([0.1, 0.7, 0.2])  # 모델 예측 확률

print(f"MSE 손실 값 : {mean_squared_error(y_sample, t_sample):.4f}")
print(f"CEE 손실 값 : {cross_entropy_error(y_sample, t_sample):.4f}")
```

## 타. 데이터 규모에 따른 학습 방식: 배치 vs 미니배치
손실 함수를 통해 오차를 구한 뒤 모델의 매개변수(가중치와 편향)를 업데이트할 때, **"한 번에 얼마나 많은 데이터를 보고 손실을 계산할 것인가?"**에 따라 크게 두 가지 학습 방식으로 나뉨

### 1. **배치 학습 방식 (Batch Learning)**
* **개념**: 컴퓨터 메모리가 허용하는 한, 가지고 있는 전체 데이터 $N$개의 손실 값을 몽땅 구한 뒤, 이들의 평균 손실을 내어 매개변수를 딱 1번 업데이트하는 방식
* **한계점**: 전체 데이터를 모두 반영하므로 데이터가 수백만 개에 달하는 빅데이터 환경에서는 컴퓨터 메모리가 버티지 못하고 터지거나, 한 번 업데이트하는 데 시간이 너무 오래 걸림

### 2. **미니배치 학습 방식 (Mini-batch Learning, 현대 딥러닝의 표준)**
* **개념**: 전체 데이터 $N$개 중 임의의 작은 묶음인 $n$개(배치 크기)를 무작위로 추출하여, 그 $n$개 데이터 손실의 '평균'을 구한 뒤 매개변수를 기민하게 업데이트하는 방식
* **작동 프로세스**:
  1. 전체 데이터 수 $N$개 중 미니배치 크기 $n$개를 무작위로 추출하여 평균 손실을 구하고 매개변수를 업데이트함
  2. 이 과정을 $N / n$번 반복하여 데이터를 소진하며, 마지막 남은 잔여 데이터까지 처리하면 전체 데이터에 대해 1회 학습을 완료한 상태인 **1 에포크(Epoch)**가 됨

## 파. 미분과 편미분 (Differentiation & Partial Derivative)
### 1. **딥러닝에서 미분이 반드시 필요한 이유**
* 손실 함수의 값을 줄이기 위해 **특정 가중치($W$)를 아주 조금 바꿨을 때 손실 함수의 값이 늘어나는지 줄어드는지** 판별하는 유일한 도구가 미분(기울기)이다.
* 미분 값이 양수(+)이면 가중치를 $(-)$ 방향으로, 음수(-)이면 가중치를 $(+)$ 방향으로 조절한다.

### 2. **수치 미분 (Numerical Differentiation)**
* 복잡한 미분 공식을 컴퓨터에게 일일이 가르치지 않고, 아주 미세한 차이($h$)를 대입하여 수치적인 근사치로 고속 계산하는 방식
* 수식: $f'(x) \approx \frac{f(x + h) - f(x - h)}{2h}$

### 3. **편미분 (Partial Derivative)**
* 입력 변수가 여러 개인 다변수 함수에서, 관심 있는 단 하나의 변수만 미분하고 나머지는 모두 고정된 상수로 취급하는 방식 (기호: $\frac{\partial f}{\partial x}$)
* 각 가중치별 편미분을 모두 구해 1차원 배열로 묶은 것을 **기울기(Gradient)**라고 한다.

```python
# 수치 미분 함수 구현 실습
def numerical_diff(f, x):
    h = 1e-4  # 0.0001
    return (f(x + h) - f(x - h)) / (2 * h)

# 예시 함수: f(x) = x^2
def function_1(x):
    return x**2

# x = 3에서의 미분값(기울기) 계산
grad = numerical_diff(function_1, 3.0)
print(f"수치 미분 결과 (x=3.0에서의 기울기): {grad}")
```

## 하. 계산 그래프와 오차역전파 (Backpropagation)
### 1. **계산 그래프와 국소적 계산**
* **순방향 연산 (Forward Pass)**: 입력층에서 출력층 방향으로 순서대로 연산하여 최종 결과(손실값)를 구하는 과정
* **역방향 연산 (Backward Pass)**: 출력층에서 입력층 방향으로 거꾸로 이동하며 기울기를 계산하는 과정
* **국소적 계산**: 전체 신경망이 거대해도 개별 노드는 자신에게 입력되고 출력되는 좁은 범위의 연산만 처리하면 됨

### 2. **연쇄법칙 (Chain Rule)**
* 합성함수의 미분은 구성하는 개별 함수의 미분(국소적 미분)들을 차례대로 모두 곱해서 구할 수 있다는 법칙 ($\frac{\partial y}{\partial x} = \frac{\partial y}{\partial u} \times \frac{\partial u}{\partial x}$)
* **오차역전파의 본질**: 뒤에서 건너온 오차 신호에 활성화 함수의 미분과 입력값을 곱해주는 것만으로 모든 은닉층 가중치의 기울기를 일괄 산출

## 거. 최적화 (Optimization)와 경사하강법 (Gradient Descent)
### 1. **최적화의 정의**
* 손실 함수의 값(오차)이 최소가 되는 최적의 가중치를 찾아가는 과정

### 2. **확률적 경사 하강법 (SGD, Stochastic Gradient Descent)**
* **공식**: $W_{new} = W_{old} - \eta \frac{\partial L}{\partial W}$
* **작동 방식**: 현재 위치에서 기울기가 감소하는 내리막길 방향으로, 설정한 보폭(학습률 $\eta$)만큼 정직하게 한 걸음씩 걸어 내려가는 방식
* **치명적 한계 (로컬 미니멈)**:
  * 가짜 최소인 지역 극소점(Local Minimum) 바닥에 도달하면 주변이 모두 오르막이라 기울기가 0이 됨
  * 보폭이 0이 되어 학습이 완전히 멈추고 전역 최소점(Global Minimum)에 도달하지 못하는 문제 발생

### 3. **개선된 최적화 알고리즘**
* **모멘텀 (Momentum)**: 운동량(관성) 개념을 도입하여 가짜 골짜기를 관성의 힘으로 치고 넘어가도록 개선
* **AdaGrad**: 학습이 많이 진행된 가중치는 보폭을 촘촘히 줄이고, 학습이 부족한 가중치는 보폭을 넓혀 맞춤형 조율 수행
* **Adam**: 모멘텀과 AdaGrad의 장점을 융합하여 현대 딥러닝에서 가장 널리 사용되는 표준 최적화 알고리즘

```python
# 확률적 경사 하강법(SGD) 업데이트 모의 구현 실습

def gradient_descent(w, lr, grad):
    # W_new = W_old - (learning_rate * 기울기)
    w_new = w - (lr * grad)
    return w_new

# 테스트 변수 설정
w_old = 10.0          # 현재 가중치
learning_rate = 0.1   # 학습률 (보폭)
current_grad = 12.0   # 현재 가중치에서의 손실 함수 기울기

w_updated = gradient_descent(w_old, learning_rate, current_grad)
print(f"초기 가중치: {w_old}")
print(f"업데이트된 가중치 (1회 반복 후): {w_updated}")
```

## 너. 계산 그래프와 국소적 계산 (Computational Graph & Local Computation)
### 1. **계산 그래프(Computational Graph)의 정의**
* **개념**: 복잡한 수학 계산 과정을 노드(Node, 연산 기호)와 에지(Edge, 데이터가 흐르는 선)로 이루어진 네트워크 그래프로 시각화하여 표현한 것
* **순방향 연산 (Forward Pass)**: 입력층에서 시작하여 출력층 방향으로 순서대로 연산을 진행하여 최종 결과(예측값 또는 손실값)를 구해내는 과정
* **역방향 연산 (Backward Pass)**: 출력층에서 시작하여 입력층 방향(반대 방향)으로 거꾸로 이동하며 기울기(Gradient)를 계산하는 과정

### 2. **국소적 계산 (Local Computation)**
* **개념**: 전체 신경망이 아무리 거대하고 수천만 개의 연산이 얽혀 있어도, 개별 노드는 "자신에게 입력되는 데이터"와 "자신이 출력하는 데이터"라는 아주 좁은 범위(국소적)의 연산만 처리하는 방식
* **의의**: 예를 들어 '더하기(+)' 연산 노드는 전체 과정을 전혀 알 필요 없이, 오직 자신에게 입력된 두 숫자인 4,000과 200만 가지고 4,200이라는 결과를 내보내면 되므로 복잡한 연산을 효율적으로 단순화할 수 있음

```python
# 계산 그래프 순방향 연산 실습 (사과 구매 예시)
# 조건: 1개에 100원인 사과를 2개 사고, 10%의 소비세 부과

apple_price = 100
apple_count = 2
tax_rate = 1.1

# 1단계 국소적 계산: 사과 가격 계산
total_apple_price = apple_price * apple_count

# 2단계 국소적 계산: 소비세 반영 최종 금액 계산
final_payment = total_apple_price * tax_rate

print(f"사과 구매 순방향 연산 결과 (소비세 포함 최종 금액): {final_payment}원")
```

## 더. 합성함수(Composite Function)의 이해와 연쇄법칙
### 1. **합성함수란 무엇인가?**
* 여러 개의 함수가 사슬처럼 줄지어 엮여서, 한 함수의 출력이 다음 함수의 입력으로 이어지는 구조의 함수
* 딥러닝 모델의 실체가 바로 이 합성함수임 (입력층 $\rightarrow$ 은닉층 함수 $\rightarrow$ 출력층 함수 $\rightarrow$ 손실 함수)

### 2. **연쇄법칙 (Chain Rule)**
* **정의**: 합성함수의 미분은 합성함수를 구성하는 개별 함수의 미분(국소적 미분)들을 차례대로 모두 곱해서 구할 수 있다는 법칙
* **공식**: $\frac{\partial y}{\partial x} = \frac{\partial y}{\partial u} \times \frac{\partial u}{\partial x}$
* **오차역전파의 본질**: 단계별 미분 결과를 곱해주면 중간 변수 $u$가 약분되듯, 출력층에서 뒤로 오차를 보내는(역전파) 것만으로 모든 은닉층 가중치의 기울기를 일괄 산출할 수 있음

## 러. 순방향 연산과 역방향 연산
### 1. **순방향 연산 (Forward Pass, Left to Right)**
* 마트에서 계산이 일어날 때는 원가($x$)가 입력되어 가격($u$)이 되고, 세금이 붙어 최종 금액($y$)이 결정됨
* 왼쪽(입력)에서 오른쪽(출력)으로 흘러가는 과정
* 흐름: 원가 ($x$) $\rightarrow$ [마진 노드] $\rightarrow$ 가격 ($u$) $\rightarrow$ [소비세 노드] $\rightarrow$ 최종 금액 ($y$)

### 2. **기울기의 계산과 역방향 연산 (Backward Pass, Right to Left)**
* 연쇄법칙 공식: $\frac{\partial y}{\partial x} = \frac{\partial y}{\partial u} \times \frac{\partial u}{\partial x}$
* 맨 오른쪽 끝의 결과물에서 구한 미분 결과($\frac{\partial y}{\partial u}$)를 왼쪽으로 보내고, 국소적 미분값($\frac{\partial u}{\partial x}$)과 곱하면 최종 결과에 도달함
* 이처럼 미분의 결과를 출력(오른쪽)에서 입력(왼쪽) 방향으로 거꾸로 전달하는 행위를 **역전파(Backpropagation)**라고 부름

## 머. 사과 예시와 인공신경망의 1:1 대응 및 최종 목표
### 1. **사과 예시와 인공신경망의 대응**
* 사과 예시: 원가($x$) $\rightarrow$ 판매 가격($u$) $\rightarrow$ 최종 결제 금액($y$)
* 인공신경망: 은닉층 퍼셉트론 입력 신호 총합 ($z = w \times x + b$) $\rightarrow$ 활성화 함수 ($a = \sigma(z)$) $\rightarrow \dots \rightarrow$ 최종 손실값($L$)
* 인공신경망의 본질은 수많은 퍼셉트론들이 사슬처럼 엮인 하나의 거대한 **'합성함수'**이다.

### 2. **우리가 찾아야 하는 최종 목표**
* 딥러닝 학습의 목표는 최종 오차인 손실 함수($L$)를 0에 가깝게 줄이는 것이며, 모든 퍼셉트론의 가중치($W$)를 최적의 값으로 업데이트해야 함 ($W_{new} = W_{old} - \eta \frac{\partial L}{\partial W}$)
* 즉, 우리가 구하고자 하는 최종 계산값은 특정 퍼셉트론의 **'손실($L$)에 대한 가중치($W$)의 기울기($\frac{\partial L}{\partial W}$)'**이다.

### 3. **은닉층 가중치 기울기 도출 (오차역전파)**
* **공식**: $\frac{\partial L}{\partial w} = \frac{\partial L}{\partial a} \times \frac{\partial a}{\partial z} \times \frac{\partial z}{\partial w}$
* 출력층에서 건너온 오차($\frac{\partial L}{\partial a}$), 활성화 함수 미분($\frac{\partial a}{\partial z}$), 입력값($\frac{\partial z}{\partial w} = x$)을 곱해주어 일괄 산출함

## 버. 최적화(Optimization)와 확률적 경사 하강법 (SGD)
### 1. **최적화의 정의**
* **최적화**: 손실 함수의 값(오차)이 최소가 되는 최적의 가중치를 찾는 과정
* **최적화 알고리즘**: 최적의 가중치를 찾아가는 구체적인 방법과 수단 (기본: 확률적 경사 하강법)

### 2. **확률적 경사 하강법 (SGD, Stochastic Gradient Descent)**
* **공식**: $W_{new} = W_{old} - \eta \frac{\partial L}{\partial W}$
* **작동 방식**: 현재 위치에서 기울기가 감소하는 내리막길 방향으로, 설정한 보폭(학습률 $\eta$)만큼 정직하게 한 걸음씩 걸어 내려가는 방식
* **치명적 한계 (로컬 미니멈 문제)**:
  * 가짜 최소인 지역 극소점(Local Minimum) 바닥에 도달하면 주변이 모두 오르막이라 순간적으로 가중치의 기울기가 0이 됨
  * 보폭이 0이 되어 학습이 완전히 멈추고, 반대편의 진짜 목적지인 전역 최소점(Global Minimum)을 눈앞에 둔 채 가짜 골짜기에 갇히게 됨

## 서. 개선된 최적화 알고리즘: 모멘텀과 AdaGrad
### 1. **모멘텀 (Momentum)**
* **개념**: '운동량' 또는 '관성'을 뜻하는 물리학 용어로, 가중치가 이동할 때 과거에 이동했던 방향과 속도를 기억하여 가속도를 붙여주는 알고리즘
* **수식**: 
  * $v_{new} = \alpha v_{old} - \eta \frac{\partial L}{\partial W}$
  * $W_{new} = W_{old} + v_{new}$
* **효과**: 지역 극소점 바닥에서 기울기가 0이 되더라도 과거에서 누적된 속도(관성)의 힘으로 가짜 골짜기를 탈출하여 언덕을 넘어감

### 2. **AdaGrad**
* **기본 원리**: 학습이 많이 진행된 가중치는 학습률을 강하게 감소시켜 보폭을 촘촘히 줄이고, 학습이 부족한 가중치는 초기 학습률을 유지하여 최소 지점에 빠르게 도달하도록 가중치별 맞춤형 조율을 수행함
* **치명적 한계**: 로컬 미니멈까지 오는 동안 학습 시간이 오래 걸리면, 기울기 제곱을 누적하는 주행거리 계기판($h$) 값이 무한히 커져 보폭이 완전히 0이 되어버리고 학습이 종료됨

## 어. 현대 딥러닝의 표준: Adam
### 1. **두 알고리즘의 완벽한 융합**
* 쇠구슬의 묵직한 관성(Momentum)과 가중치별 보폭 자율 조정(AdaGrad)의 장점만을 영리하게 담은 딥러닝 최우선 표준 최적화 알고리즘

### 2. **AdaGrad 단점의 완벽한 극복 (기억 상실 기능)**
* 과거의 기울기 제곱 누적 기록을 일정 비율로 계속 깎아내어(감쇠시켜) 자연스럽게 제거함
* 최근에 생성되어 누적되는 최신 기울기 위주로 반영하기 때문에, 누적 수치가 무한대로 폭발하지 않음
* 따라서 로컬 미니멈처럼 경사가 완만한 지점에 도달해도 유효 학습률이 0이 되지 않고 안정적으로 탈출 가능함

## 저. PyTorch를 이용한 딥러닝 모델 개발 기초
### 1. **`nn.Linear(in_features, out_features)`**
* 은닉층과 출력층을 만드는 핵심 도구
* `in_features`: 이전 단계에서 들어오는 데이터(입력 특성)의 개수
* `out_features`: 현재 계층에 배치할 새로운 인공 뉴런(퍼셉트론)의 개수

### 2. **`nn.Sequential()`**
* 여러 층을 순서대로 쌓아 하나로 묶어주는 컨테이너 도구

## 처. PyTorch 분류(Classification) 모델 설계, 학습, 평가
### 1. **모델 설계 및 손실 함수/옵티마이저**
* **이진 분류**: `nn.BCEWithLogitsLoss()` (시그모이드 + 이진 교차 엔트로피 결합, 출력 노드 1개)
* **다중 분류**: `nn.CrossEntropyLoss()` (소프트맥스 + 다중 교차 엔트로피 결합, 출력 노드 = 클래스 개수)
* **옵티마이저**: `optim.Adam(model.parameters(), lr=0.01)`

### 2. **학습 루프 (Training Loop: 5대 핵심 공식)**
1. `optimizer.zero_grad()`: 이전 루프의 기울기 초기화
2. `outputs = model(X_train)`: 순전파(Forward Pass) 진행
3. `loss = criterion(outputs, y_train)`: 손실 함수를 통한 오차 계산
4. `loss.backward()`: 역방향 연산(Backpropagation)을 통한 기울기 계산
5. `optimizer.step()`: 가중치 갱신(Step)

### 3. **모델 평가**
* `model.eval()` 및 `torch.no_grad()` 블록을 사용하여 미분 계산을 끄고 추론 진행 후 정확도(Accuracy) 측정

## 커. PyTorch 회귀(Regression) 모델 설계, 학습, 평가
### 1. **회귀 모델의 특징**
* 연속적인 수치값 1개를 예측하므로 출력층 노드는 **1개**로 설정함
* 회귀 손실 함수로 평균 제곱 오차(`nn.MSELoss()`)를 사용하여 예측값과 실제값의 차이를 제곱 평균함

### 2. **모델 평가 지표**
* **MSE (평균 제곱 오차)**: 오차의 제곱 평균
* **RMSE (제곱근 평균 제곱 오차)**: MSE에 루트를 씌워 오차를 원래의 단위로 되돌린 지표
