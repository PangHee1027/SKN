# 환경 설정

```python
# 필요한 라이브러리 / 모듈 / 클래스 / 함수 임포트
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import root_mean_squared_error
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
```

# 데이터 불러오기

```python
# 파일 경로 설정
file_path='./데이터/house_price.csv'

# DataFrame 생성
df = pd.read_csv(file_path)

# 결과 확인
display(df)
```

# 학습용과 평가용 데이터 생성

```python
# 75:25의 비율로 학습용과 평가용 데이터 생성

## X_data 생성: 정답 레이블 제거
X_data = df.drop(columns=['medv'])
display(X_data)

print('-'*80)

## y_data 생성: 정답 레이블 인덱싱
y_data = df['medv']
print(y_data)

## train_test_split() 함수 호출
X_train, X_test, y_train, y_test = train_test_split(
    X_data,
    y_data,
    test_size=0.25,
    random_state=0
)
```

```python
# 학습용 데이터 확인: 인덱스 확인
print(f'X_train의 인덱스: \n{X_train.index}')
print('-'*80)
print(f'y_train의 인덱스: \n{y_train.index}')
```

# 전처리: 표준화

```python
# 표준화 도구 생성
scaler = StandardScaler()
```

```python
# 학습용과 평가용 데이터 변환: 학습용 기준(평균, 표준편차)
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 변환의 결과 확인: pandas 자료형(DataFrame) -> 넘파이 배열
print(X_train_scaled)
print('-'*80)
print(X_test_scaled)
```

```python
# Pytorch Tensor 변환 및 디바이스로 이동
"""
# 넘파이 배열 -> pytorch 학습에 적합한 형태의 자료형(Tensor)으로 변환
"""
X_train_tensor = torch.FloatTensor(X_train_scaled)
X_test_tensor = torch.FloatTensor(X_test_scaled)
y_train_tensor = torch.FloatTensor(y_train.values)
```

```python
# pytorch용 표준 데이터셋 생성 -> 미니 배치 생성 준비 단계
train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
```

```python
# DataLoader 생성

"""
# 학습용 데이터: 379개 -> batch_size=16 -> 미니 배치:24개 -> 미니 배치 단위 학습 가능
"""

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
```

# MLP 회귀 모델 생성

```python
# 입력 특성(컬럼) 수 추출
input_dim = X_train_tensor.shape[1]
print(f'입력 데이터 특성의 수: {input_dim}')

print('-'*80)

# nn.Sequential 클래스 사용: 입력층 -> 은닉층1 -> 은닉층2 -> 출력층 구성
model = nn.Sequential(
    nn.Linear(input_dim, 64), # [입력층 -> 은닉층1]: 입력 13개, 인공 뉴런 64개로 구성
    nn.ReLU(),                # [활성화 함수]: 비선형 연산(음수 -> 0)
    nn.Linear(64, 32),        # [은닉층2]: 입력 64, 인공 뉴런 32개로 구성
    nn.ReLU(),                # [활성화 함수]: 비선형 연산(음수 -> 0)
    nn.Linear(32, 1)          # [출력층]: 입력 32, 출력 뉴런 1개로 구성 -> 주택 가격(예측)
)

print(f'생성된 모델 아키텍처: \n{model}')
```

# 손실 함수 설정

```python
# 회귀 문제의 손실 함수: 평균 제곱 오차(MSE)
criterion = nn.MSELoss()
```

# 최적화 알고리즘 설정

```python
# 최적화: Adam 적용(SGD 개선)
optimizer = optim.Adam(model.parameters(), lr=0.005)
```

# 모델 학습

```python
# 학습 횟수 설정
epochs = 200

# 모델을 학습 모드로 전환
model.train()

# for 반복문 이용 학습 진행
for epoch in range(epochs):

    ## 손실값 초기화
    epoch_loss = 0.0

    for batch_x, batch_y in train_loader:
        # 1) 기울기 초기화(이전 학습의 결과물 삭제)
        optimizer.zero_grad()

        # 2) 순전파 연산 (출력층 뉴런 입력의 총합 -> 예측값)
        predictions = model(batch_x)

        # 3) 손실 계산 (batch_y를 (batch_size, 1) 모양으로 맞춰줌)
        loss = criterion(predictions, batch_y.unsqueeze(1))

        # 4) 오차역전파 (가중치별 기울기 산출)
        loss.backward()

        # 5) 가중치 매개변수 갱신
        optimizer.step()

        # 손실값 업데이트: pytorch Tensor 자료형 파이썬 숫자형으로 변환
        epoch_loss += loss.item()

    if (epoch + 1) % 20 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Train MSE Loss: {epoch_loss/len(train_loader):.4f}")
```

# 모델 평가

```python
# 모델을 평가 모드로 전환
model.eval()

# 평가용 데이터를 이용한 예측: 평가 시에는 기울기 계산 비활성화
with torch.no_grad():
    # 출력 뉴런의 입력 신호의 총합(예측값) 도출
    test_preds = model(X_test_tensor).cpu().numpy()

# 성능 평가: rmse 계산
rmse_mlp = root_mean_squared_error(y_test, test_preds)

# 결과 확인
print(f"[보스톤 주택 가격 예측] PyTorch MLP 평가 RMSE: {rmse_mlp:.4f}")
```
