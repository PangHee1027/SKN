## 1. 딕셔너리

### 가) 딕셔너리 개요

* Key-Value 쌍 구조: 데이터가 {Key: Value}의 형태(1쌍)로 저장

* Key의 중복 불가: Key는 유일해야 하며 중복될 수 없으나, Value는 중복이 가능

* Key를 통한 데이터 접근: 순서(인덱스)가 따로 없으며, 인덱스 대신 Key를 이용하여 데이터를 가져옴

```python
a = {}
b = {"name" : "JH"}
c = {1 : "이재현", 2 : "백수"}
print(a)
print(b)
print(c)
```

### 나) 딕셔너리 생성

중괄호 {}를 사용하며, Key와 Value는 콜론 :으로 구분

* a = {}

* b = {"name" : "JH"}

* c = {1 : "이재현", 2 : "백수"}

```python
my_info = {"name" : "이재현", "age" : 26, "phone" : "010-4131-1852"}
print(my_info)
```

### 다) 딕셔너리 값 가져오기

# 나. 기본 자료형 (계속)

## 1. 숫자형 자료형
### 1) 정수형 (int, Integer): 소수점이 없는 수 (양의 정수, 음의 정수, 0)
  * 생활 예시: 나이, 제품 수량, 방문자 수
### 2) 실수형 (float, Floating-point): 소수점이 포함된 수
  * 생활 예시: 키/몸무게, 시력, 환율, 할인율
### 3) 연산자
  * 기본 사칙 연산 (+, -, *, /): 두 숫자형 데이터 간의 가감승제 연산 수행


```python
num1 = 123
num2 = 1.2
summ = num1 + num2
sub = num1 - num2
multi = num1 * num2
div = num1 / num2
print(f"{num1} + {num2} = {summ}")
print(f"{num1} - {num2} = {sub}")
print(f"{num1} * {num2} = {multi}")
print(f"{num1} / {num2} = {div}")
```

  * 특수 연산자 (**, //, %)
    * 제곱 연산자 ( ** ): x의 y제곱 계산, 예) 3의 4제곱 (3**4)
    * 나머지 연산자 (%): 나눗셈 후 나머지만 추출
    * 몫 연산자 (//): 나눗셈 후 소수점 아래를 버리고 몫(정수)만 추출

```python
power = num1 ** num2
remain = num1 % num2
quot = num1 // num2
print(f"{num1} ** {num2} = {power}")
print(f"{num1} % {num2} = {remain}")
print(f"{num1} // {num2} = {quot}")
```

## 2. 문자열 자료형
### 가) 개요
  * 하나 이상의 연속된 문자로 구성된 데이터 형태
  * 각 문자마다 위치(인덱스)를 가짐
### 나) 생성 방식
  * 단일 줄: 따옴표(' 또는 ")로 감싸서 생성
  * 여러 줄: 작은따옴표/큰따옴표 3개(''' 또는 """)를 사용하여 줄 바꿈이 포함된 문자열을 생성


```python
string1 = 'a'
string2 = 'apple'
string3 = "banana"
multi_line = """
this is line number1
this is line number2
this is line number3
this is line number4
"""
print(string1, string2, string3)
print(multi_line)
```

### 다) 인덱싱, 슬라이싱
  * string[0]처럼 위치 번호를 지정해 특정 위치의 문자를 추출
  * string[start:stop] 형태로 원하는 범위의 문자열을 잘라냄


```python
string4 = "abcdefg"
print(string4[3])
print(string4[3:6])
```

### 라) 연산
  * "+" 연산자를 사용해 문자열끼리 연결 가능

```python
string5 = "hijklmnop"
print(string4 + string5)
```

  * "*" 연산자를 사용해 문자열 반복 가능

```python
string6 = "안녕"
print(string6 * 3)
print("-" * 80)
```


## 2. 숫자형 사칙 연산 변수 num1에 15, num2에 4를 할당하고, 두 수의 덧셈, 뺄셈, 곱셈, 나눗셈 결과를 각각 출력하는 코드를 작성하세요.

```python
num1 = 15
num2 = 4
sum = num1 + num2
sub = num1 - num2
multi = num1 * num2
div = num1 / num2
print(f"{num1} + {num2} = {sum}")
print(f"{num1} - {num2} = {sub}")
print(f"{num1} * {num2} = {multi}")
print(f"{num1} / {num2} = {div}")
```

## 3. 제곱 연산자(**) 활용 변수 base에 2, exponent에 10을 저장한 뒤, 제곱 연산자(**)를 사용하여 2의 10제곱 값을 계산하여 출력하는 코드를 작성하세요.


```python
base, exponent = 2, 10
print(f"{base}^{exponent} = {base ** exponent}")
```

## 4. 몫(//)과 나머지(%) 연산자 17을 5로 나누었을 때의 몫과 나머지를 각각 몫 연산자(//)와 나머지 연산자(%)를 이용해 구하고 출력하는 코드를 작성하세요.

```python
num1, num2 = 17, 5
quot = 17 // 5
remain = 17 % 5
print(f"{num1} // {num2} = {quot}")
print(f"{num1} % {num2} = {remain}")
```
