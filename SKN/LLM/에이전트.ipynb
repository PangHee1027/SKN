## 나. OpenAI API 사용법과 토큰 (Token)
### 1. **OpenAI API 개요 및 필요성**
* **필요성**: 모든 기능을 직접 구현할 필요 없이, OpenAI의 대규모 언어 모델(LLM)을 가져와 기업 내 전용 Agentic AI 애플리케이션에 결합
* **OpenAI**: 인공지능 연구 개발 기업으로 ChatGPT를 개발
* **OpenAI API**: 텍스트 생성(사용자 요청에 대한 응답), 코드 생성 등 다양한 기능 제공
* **API 키 관리**: 민감한 정보는 코드와 분리하여 `.env` 파일에 안전하게 저장 (예: `OPENAI_API_KEY="sk-..."`)

### 2. **토큰(Token)과 서브 워드 토큰화 (Subword Tokenization)**
* **핵심 단위**: 토큰(Token)
* **서브 워드 토큰화 개념**: 단어를 고정된 단위로 자르지 않고, 의미를 가진 더 작은 단위인 '서브워드'로 분리하는 방식
* **필요성**: 컴퓨터가 신조어를 포함한 수억 개의 다양한 문장을 적은 양의 단어 조각 조합으로 이해할 수 있도록 지원
* **핵심 아이디어**: 자주 사용되는 단어는 하나의 토큰으로 처리하고, 자주 사용되지 않는 단어나 신조어는 여러 개의 서브워드로 분리 (예: Chatbot $\rightarrow$ Chat + bot)

### 3. **영어와 한글 토큰 처리 비교**
| 구분 | 영어 | 한글 | 차이점 |
| :--- | :--- | :--- | :--- |
| **1 토큰당 글자 수** | 평균 4글자 (알파벳 기준) | 평균 1~2글자 (공백 포함) | 한글은 음절이나 자모 단위로 쪼개짐 |
| **단어당 토큰 수** | 1단어 $\approx 0.75$ 토큰 | 1단어 $\approx 2\sim3$ 토큰 | 한글은 조사와 어미 때문에 토큰이 증가함 |
| **비용 및 속도** | 기준점 (1배) | 약 2.5배~3배 부담 | 한글 텍스트 처리가 훨씬 비싸고 느림 |

## 다. 프롬프트 엔지니어링 (Prompt Engineering)
### 1. **프롬프트 엔지니어링의 중요성**
* **프롬프트**: 생성형 AI 모델에게 내리는 명령어 또는 질문
* **프롬프트 엔지니어링**: 효과적인 프롬프트 작성 방법을 연구하여 AI 모델 성능을 극대화하고 원하는 결과를 얻기 위해 필수적임
* **기법 비교**:
  * **Standard Prompting**: 질문에 대한 직접적인 결과 도출
  * **Chain-of-Thought (CoT) Prompting**: 단계별 사고 과정(풀이 과정)을 함께 유도하여 추론 정확도 향상

### 2. **효과적인 프롬프트 작성 방법**
* **구체적인 지시**: 명확하고 자세하게 요구사항 명시 (예: "소설을 써줘" $\rightarrow$ "등장인물 3명, 갈등 구조를 가진 로맨스 소설 1000자 내외로 써줘")
* **역할 부여**: AI 모델에게 특정 역할 부여 (예: "너는 IT 전문가야. IT 트렌드를 요약해줘.")
* **맥락 제시**: 필요한 배경 정보 제공 (예: "최근 경제 상황을 고려하여, 2026년 하반기 IT 시장 전망을 분석해줘.")
* **예시 활용**: 원하는 결과물의 예시 제시
* **반복 및 개선**: 결과물을 확인하며 프롬프트 수정 및 개선

# LangChain 프레임워크 활용

## 가. LangChain 개요
### 1. **LangChain의 정의와 목표**
* **정의**: 대규모 언어 모델(LLM)을 기반으로 한 AI 애플리케이션 개발을 빠르고 효율적으로 완성해주는 프레임워크
* **목표**: LLM의 추론 능력과 외부 데이터 소스 또는 기타 계산 도구를 결합하여 더욱 강력하고 확장 가능한 애플리케이션 생성

### 1. **Prompts (프롬프트 관리)**

* **ChatPromptTemplate**: 시스템, 사용자, 어시스턴트 등 역할별 메시지 구조를 체계적으로 관리

### 2. **LLMs와 ChatModels**
* **LLMs**: 텍스트를 입력받아 텍스트를 완성하는 모델 인터페이스
* **ChatModels**: 메시지 리스트를 입력받아 대화형 응답을 생성하는 모델 인터페이스 (OpenAI, Anthropic 등 다양한 공급자 지원)

### 3. **Output Parsers (출력 파서)**
* **개념**: LLM의 텍스트 출력을 파이썬 객체나 특정 구조(JSON, Pydantic 모델 등)로 변환하여 후속 처리가 용이하도록 지원

## 다. Chains와 Memory
### 1. **Chains (체인)**
* **개념**: LangChain의 핵심 빌딩 블록으로, 여러 구성 요소(프롬프트 $\rightarrow$ 모델 $\rightarrow$ 파서)를 파이프라인 형태로 순차적 연결
* **LCEL (LangChain Expression Language)**: 파이프(`|`) 연산자를 사용하여 선언적이고 직관적으로 체인을 조합하는 최신 문법

### 2. **Memory (메모리)**
* **필요성**: LLM은 기본적으로 상태가 없는(Stateless) 특성이 있으므로, 이전 대화 맥락을 기억하기 위해 필요
* **주요 종류**:
  * `ConversationBufferMemory`: 대화 기록을 버퍼에 그대로 유지
  * `ConversationSummaryMemory`: 대화 기록이 길어질 경우 핵심 내용을 요약하여 유지

````python
chat_history = []

print("대화형 AI 챗봇이 시작되었습니다!. \n")


# 종료 조건이 충족되기전까지 계속 실행 -> while 반복문
while True:
    # 사용자 입력 함수 호출 -> 사용자 입력 생성
    user_input = input("사용자 입력: ")

    # 종료 조건 설정: 사용자 입력이 ["q", "quit", "exit", "종료"]에 속하면 종료
    if user_input.lower() in ["q", "quit", "exit", "종료"]:
        break

    # chain을 호출하기 전, 사용자 질문(입력)을 chat_history에 추가한다.
    """
    # user_input -> 자료형: 문자열 -> LangChain 활용 -> HumanMessage 자료형
    """
    chat_history.append(HumanMessage(content=user_input))

    # invoke() 함수 호출, AI 답변 생성
    response = chatbot_chain.invoke({"chat_history":chat_history})
    print(f"AI 챗봇 답변: {response}\n")

    # LLM 답변 생성이 끝난 후, LLM의 답변을 AIMessage 클래스를 이용, chat_history 추가
    chat_history.append(AIMessage(content=response))
````

````python
# 필요한 함수 / 클래스 임포트
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_tavily import TavilySearch
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, ToolMessage
from langchain_core.output_parsers import StrOutputParser
````

## 도구 설정

````python
# 검색 도구 생성 클래스 호출
search_tool = TavilySearch(
    max_results=3,
    search_depth='advanced'
)
````

````python
# 사용할 도구 목록 생성: 에이전트(LLM)의 선택을 돕는다.
tools = [search_tool]
print(f'도구 목록: {tools}')
````

## LLM  생성

````python
# ChatOpenAI 함수 호출
llm = ChatOpenAI(model='gpt-5-nano')
````

````python
# LLM에 도구 바인딩(binding)

"""
1. 구현: llm.bind_tools()
2. 입력 값: 도구 목록 -> tools = [search_tool]
3. bind_tools(tools)
   1) LLM과 외부 도구(tools)를 연결, LLM이 필요에 따라 이 도구들을 호출할 수 있도록 만드는 핵심 기능
   2) LLM에게 도구 이름과 설명 제공
   3) LLM에게 특정 작업을 수행하기 위한 "능력"을 부여하고, 해당 능력이 필요할 때 자율적으로 도구를 사용하도록 유도
"""

## bind_tools() 함수 호출
llm_with_tools = llm.bind_tools(tools)
print(llm_with_tools)
````

## 프롬프트 생성

````python
"""
# MessagesPlaceholder 클래스: 프롬프트 내에 대화 기록의 목록(리스트) 입력 -> 시스템의 메모리 기능
   1) 예시: [HumanMessage(content="서울 날씨 어때?), AIMessage(content="", tool_calls=[{"name":"tavily_search"}]), ToolMessage(content="서울 날씨 흐림, 기온 10"))]
   2) 매개 변수 variable_name: 대화 히스토리를 저장하는 리스트 변수 
"""

# 시스템 프롬프트
system_prompt="""
너는 사용자의 질문에 정확하고 객관적인 답변을 제공하는 전문 AI 에이전트야.
1. 일반적인 대화나 이미 알고 있는 보편적인 지식만으로 충분히 답변할 수 있는 질문에는 웹 검색 도구를 호출하지 말고 네 내부 지식을 활용해 답변해.
2. 최신 정보, 실시간 트렌드, 또는 정확한 사실 확인이 필요한 질문에는 웹 검색 도구를 적극 호출하고, 검색 결과에 기반하여 환각(Hallucination) 없이
사실만을 정리해서 답변해.
그리고 모든 답변은 반드시 한국어로 작성해야 해.
"""

# 프롬프트 생성
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", system_prompt),
        MessagesPlaceholder(variable_name="messages")
    ]
)

print(prompt)
````

## chain 생성

````python
# 1차 체인: 사용자 요청 분석 -> 도구 호출(Tool Call) 판단용 체인 -> 결과: AI 답변 
tool_calling_chain = prompt | llm_with_tools
````

````python
# 2차 체인: 검색 결과를 포함한 최종 텍스트 답변 생성용 체인

## StrOutputParser를 통해 AIMessage에서 순수 문자열만 추출 -> AIMessage.content 추출 
output_parser = StrOutputParser()

## chain 구성
final_anwser_chain = prompt | llm | output_parser
````

## AI 에이전트 실행

````python
# 대화 히스토리를 저장할 빈 리스트 생성
messages = []

print("AI 에이전트 가동...")

print('-'*80)

# 사용자 입력 생성
# user_input = "2026년 생성형 AI 시장 트렌드 및 최신 AI 에이전트 프레임워크 동향을 간단 명료하게 정리해 줘."
user_input = "효과적인 IT 서비스 개발 방법론에 대해서 200토큰 이내로 요약해서 제시해 줘."

# chain을 호출하기 전, 사용자 질문(입력)을 HumanMessage 클래스로 포장해서 messages에 추가
messages.append(HumanMessage(content=user_input))
print(f'messages 변수의 상태: \n{messages}')

print('-'*80)

# 1차 chain 실행 -> 도구 필요 여부 판단
ai_message = tool_calling_chain.invoke({"messages":messages})
print(f'ai_message: \n{ai_message}')
````

````python
# AI 답변 분석

## 사용자에 대한 답변 -> content에 저장
output1 = ai_message.content
print(f'AI 답변: {output1}')

print('-'*80)

## 도구 호출(시스템에게 요청) -> tool_calls에 답변이 저장
output2 = ai_message.tool_calls
print(f'AI 답변: {output2}')
````

````python
# 대화 히스토리에 1차 응답(AIMessage) 기록
messages.append(ai_message)
print(f"messages 변수의 상태: \n{messages}")
````

````python
# LLM 모델: 도구 호출 횟수 확인

## ai_message.tool_calls -> 리스트 -> [{}, {}, ...]
for tool_call in ai_message.tool_calls:
    print(tool_call)
    print('-'*80)
````

````python
# 외부 검색 필요시 도구 실행, 아니면 LLM 내부 지식만으로 답변 제공 -> 조건문 작성

if ai_message.tool_calls:
    for tool_call in ai_message.tool_calls:
        # TavilySearch 검색어 추출 -> 딕셔너리 {'query':'검색 문장'}
        search_query = tool_call['args']
        print(f'검색어: {search_query}')

        # TavilySearch 검색 도구 호출, 검색 실행
        search_output = search_tool.invoke(search_query)
        # print(search_output)

        # 도구 호출 시 도구 사용에 대한 고유 id 항목 추출
        tool_call_id = tool_call['id']
        print(f'tool_call_id: {tool_call_id}')

        # MessagesPlaceholder -> 대화 메시지(누적) 리스트 작성 -> 도구 사용의 결과 저장
        messages.append(
            ToolMessage(
                content=str(search_output),
                tool_call_id=tool_call_id
            )
        ) 

        ## 2차 체인 실행 (최종 답변 생성)
        final_response = final_anwser_chain.invoke({"messages":messages})
        print('-'*80)
        print("[최종 LLM 답변]")
        print('-'*80)
        print(final_response)

else:
    # 1차 체인의 결과 -> AI 답변 -> .content 추출
    print(f"[LLM 판단] 외부 검색 없이 내부 지식만으로 답변을 제공합니다.")
    print(ai_message.content)
````

````python
# 대화 히스토리 리스트 초기화
messages = []

print("대화형 AI 챗봇이 시작되었습니다! \n")

while True:
    user_input = input("사용자: ")
    print(f'사용자 요청: {user_input}')

    if user_input.lower() in ["q", "exit", "quit", "종료"]:
        print("AI 챗봇: 대화를 종료합니다.")
        break

    # chain을 호출하기 전, 사용자 질문(입력)을 HumanMessage 클래스로 포장해서 messages에 추가
    messages.append(HumanMessage(content=user_input))
    # print(f'messages 변수의 상태: \n{messages}')

    print('-'*80)

    # 1차 chain 실행 -> 도구 필요 여부 판단 -> 1차 응답 생성
    ai_message = tool_calling_chain.invoke({"messages":messages})

    # 대화 히스토리에 1차 응답(AIMessage) 기록
    messages.append(ai_message)
    # print(f"messages 변수의 상태: \n{messages}")

    # 외부 검색 필요시 도구 실행, 아니면 LLM 내부 지식만으로 답변 제공 -> 조건문 작성

    if ai_message.tool_calls:
        for tool_call in ai_message.tool_calls:
            # TavilySearch 검색어 추출 -> 딕셔너리 {'query':'검색 문장'}
            search_query = tool_call['args']
            # print(f'검색어: {search_query}')

            # TavilySearch 검색 도구 호출, 검색 실행 -> 외부 검색 결과 생성 
            search_output = search_tool.invoke(search_query)
            # print(search_output)

            # 도구 호출 시 도구 사용에 대한 고유 id 항목 추출
            tool_call_id = tool_call['id']
            # print(f'tool_call_id: {tool_call_id}')

            # MessagesPlaceholder -> 대화 메시지(누적) 리스트 작성 -> 도구 사용의 결과 저장
            messages.append(
                ToolMessage(
                    content=str(search_output),
                    tool_call_id=tool_call_id
                )
            ) 

        ## 2차 체인 실행 (최종 답변 생성)
        final_response = final_anwser_chain.invoke({"messages":messages})
        # print('-'*80)
        print("[최종 LLM 답변]")
        print('-'*80)
        print(final_response)

    else:
        # 1차 체인의 결과 -> AI 답변 -> .content 추출
        print(f"[LLM 판단] 외부 검색 없이 내부 지식만으로 답변을 제공합니다.")
        print(ai_message.content)
````

# 24. RAG (Retrieval Augmented Generation)

## 가. 생성형 AI의 빛과 그림자
### 1. **빛 (놀라운 능력)**
* **놀라운 창의력**: 글, 그림, 코드 등 무엇이든 척척 생성
* **방대한 지식**: 세상의 온갖 정보를 학습
* **유창한 소통**: 사람처럼 자연스러운 대화 가능

### 2. **그림자 (치명적 약점)**
* **환각 (Hallucination)**: 그럴듯한 거짓말을 생성하는 문제
* **지식의 한계**: 특정 시점까지의 데이터로 학습되어 최신 정보(예: 어제 나온 신제품)를 모름
* **출처 불분명**: 어떤 근거로 답변하는지 알 수 없어 기업의 고객 응대나 중요 의사결정에 사용하기 어려움

## 나. RAG 시스템 도입 (가장 현실적인 대안)
* **RAG 정의**: 검색(Retrieval) + 증강(Augmented) + 생성(Generation)의 약자
* **개념**: AI가 답변하기 전, 기업의 내부 데이터나 최신 웹 정보 등 신뢰할 수 있는 자료를 먼저 '검색(Retrieval)'하고 이를 기반으로 답변을 생성(Generation)하여 AI의 약점을 보완하는 핵심 기술

## 다. 지능형 고객 응대 Chatbot 구현 프로세스
1. **데이터 수집 및 변환**: 텍스트 파일을 LangChain의 기본 Document 자료형으로 변환
2. **텍스트 분할**: 임베딩 모델의 입력 길이 제한을 고려하여 긴 문서를 작은 조각(chunk)으로 분할
3. **텍스트 임베딩**: 벡터화 진행
4. **벡터 저장소 및 검색 도구 생성**: 텍스트와 임베딩 벡터를 저장하는 전용 DB(VectorStore) 구축 및 검색 도구 생성
5. **프롬프트와 LLM 생성**
6. **RAG 체인(Chain) 구성**
7. **Chatbot 실행**

## 데이터 불러오기

````python
# 내용 정리

"""
1. TextLoader 클래스
   1) .txt 파일에 있는 데이터 로드 및 자료형 변환 클래스
   2) 기능1: 외부 파일(.txt 파일)로부터 텍스트 데이터 추출
   3) 기능2: 텍스트 데이터를 LangChain의 기본 자료형 Document 자료형으로 변환
2. Document 자료형
   1) 텍스트 데이터에 대한 LangChain의 기본 자료형
   2) 메타 데이터(metadata: 파일 경로 포함) + 텍스트 내용(page_content)으로 구성
3. 전용 함수: load() 함수 -> TextLoader 클래스 호출 시 지정된 파일을 encoding 방식으로 읽고 Document 자료형으로 변환      
"""

# 데이터 로드 및 변환 도구 생성
loader = TextLoader(file_path="./데이터/manual.txt", encoding="utf-8")

# load() 함수 호출, Document 자료형 생성 -> 결과물: 리스트 [Document], .txt 파일 1개당 Document 자료형 1개 생성
documents = loader.load()
print(f'데이터 로드 및 변환의 결과물: \n{documents}') 
````

## 텍스트 분할

````python
# 적절한 글자수(chunk_size)를 찾기 위한 계산 단계

## 파이썬 문법을 이용, .txt 파일의 데이터를 문자열로 변환
file_path='./데이터/manual.txt'

with open(file_path, 'r', encoding='utf-8') as f:
    manual_text = f.read()

print(manual_text)

print('-'*80)

## 데이터 상태 확인 -> 각 질문-답변 쌍 -> 문단 단위('\n\n')로 구분

qa_pairs = []

for pair in manual_text.split("\n\n"):
    qa_pairs.append(pair)  

print(qa_pairs)

print('-'*80)

## 각 QA 쌍 -> 글자수 계산
character_length=[]

for pair in qa_pairs:
    character_length.append(len(pair))

print(f'각 질문-답변 쌍의 글자수: \n{character_length}')

print('-'*80)

##  QA 쌍 중 글자 수가 가장 긴 경우의 글자수 추출
max_char_length = max(character_length)
print(f'가장 긴 QA 단락의 글자 수: {max_char_length}자')
````

````python
for text in qa_pairs:
    print(text)
    print('-'*80)
````

````python
# RecursiveCharacterTextSplitter 클래스 내용 정리
"""
1. 임베딩 모델이 한 번에 처리할 수 있는 입력 길이 제한(8,192 토큰)을 고려하여 긴 문서를 
   작은 조각(chunk)로 분할하거나 짧은 문서 조각을 서로 병합하여 사용자 질문과 유사도 계산
   시 정확도를 높일 수 있는 적절한 길이로 만든다.
2. RecursiveCharacterTextSplitter
   1) 데이터 분할 도구
   2) 분할 순서
      - 기본 설정: \n\n 단락(문단), \n문장,공백(" "), 글자 단위("")  
      - 먼저 가장 큰 단위인 \n\n(단락)을 기준으로 텍스트를 쪼갠다
      - 쪼개진 텍스트 조각이 여전히 사용자가 설정한 기준 길이(chunk_size)보다 크다면, 
        그 다음 구분자인 \n(줄바꿈/문장)으로 다시 쪼갠다
      - 설정한 크기 이하가 될 때까지 이 과정을 공백( ), 글자 단위("") 순으로 반복한다.
      - 기준 길이(chunk_size)보다 짧으면 병합
   3) Document 자료형의 텍스트 데이터를 분할
3. 매개 변수 정리
   1) chunk_size: 각 청크의 최대 문자 수
   2) chunk_overlap: 인접한 청크 간의 중복 문자 수 -> QA, 법률 -> overlap=0 / 일반 소설 -> chunk_size 10% 
4.split_documents() 함수
   1) RecursiveCharacterTextSplitter 전용 함수
   2) 입력된 Document 자료형의 텍스트 데이터를 설정 기준(chunk_size, chunk_overlap)에 따라 분할
"""

# 데이터 분할 도구 생성
text_splitter = RecursiveCharacterTextSplitter( 
    # separators=["\n\n", "\n", " ", ""]   
    chunk_size=160,
    chunk_overlap=0
)

# split_documents() 함수 호출, 분할된 문서 생성
docs = text_splitter.split_documents(documents=documents)
print(f'분할된 문서: \n{docs}')
print('-'*80)
print(f'분할된 문서의 수: {len(docs)}')
print('-'*80)
print(f'분할된 첫 번째 문서: \n{docs[0]}')
````

## 텍스트 임베딩 모델 생성

````python
# 내용 정리
"""
1. OpenAIEmbeddings: OpenAI의 임베딩 모델을 LangChain에서 사용할 수 있도록 지원하는 클래스
2. 매개 변수 정리
   1) model: 사용할 OpenAI 임베딩 모델의 이름 지정 ('text-embedding-3-small')
"""

## 모델 이름 설정
model_name = "text-embedding-3-small"

## OpenAI 임베딩 모델 생성
model = OpenAIEmbeddings(
    model=model_name
)
````

## 백터 저장소 및 검색 도구 생성

````python
# 내용 정리
"""
1. FAISS 내부 메모리 구조
   1) index: 임베딩 벡터 배열만 저장
   2) docstore: 원본 텍스트(page_content)와 메타데이터 저장

1. from_documents() 함수
   1) FAISS 전용 함수
   2) 입력: docs (분할된 문서 조각, Document의 리스트)와 embeddings (임베딩 모델)
   3) 기능: 각 Document의 텍스트를 임베딩하고 메모리에 저장 -> 결과물: 벡터 저장소(vectorstore)
"""

# 벡터 저장소(vectorstore) 생성
vectorstore = FAISS.from_documents(docs, model)

# 저장된 임베딩 백터 확인
print(vectorstore.index)

print('-'*80)

# 저장된 원본 텍스트(page_content)와 메타데이터(metadata)
print(vectorstore.docstore)
````

````python
# 검색 도구 생성

## 내용 정리
"""
# 전용 함수: as_retriever() 함수
   1) FAISS 전용 함수. 검색 도구 생성
   2) invoke 함수를 통해서 사용자 질문 입력
      - 사용자 질문을 받을 경우, 벡터 저장소에 저장된 텍스트 중에서 사용자의 질문과 가장 유사한 텍스트를 찾아주는 역할 수행
      - 매개 변수: search_kwargs={'k': n} -> 사용자 질문과 가장 유사도가 높은 n개의 텍스트 데이터를 가져오도록 설정
"""

retriever = vectorstore.as_retriever(search_kwargs={'k': 1})
print(f'검색 도구: {retriever}')
````

````python
# 검색 도구 성능 테스트
user_query = "환불 가능한가요?"
output = retriever.invoke(user_query)
print(f"검색 결과: \n{output}")
````

## RAG 체인 생성

### 프롬프트 생성

````python
# 시스템 프로프트 생성
system_prompt = """
당신은 친절한 온라인 교육 플랫폼 직원입니다.
제공된 [컨텍스트]를 바탕으로 고객의 질문에 정중하게 답변해주세요.

[답변 규칙]
1. [컨텍스트]의 내용을 바탕으로 자연스럽게 답변을 재구성하거나 그대로 활용하세요.
2. 답변 시 '답변:'이라는 표현을 사용하지 마세요.
3. [컨텍스트]에 질문에 대한 정보가 없다면, '죄송하지만 문의하신 내용에 대해서는 아는 바가 없습니다.'라고 답변하세요.

[컨텍스트]
{context}
"""

# 사용자 프롬프트 생성
user_prompt = "{question}"

# 전체 프롬프트 생성
prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", user_prompt)
])

print(prompt)
````

### LLM 생성

````python
llm = ChatOpenAI(model='gpt-5-nano')
````

### output parser 생성

````python
# LLM 출력 분석 -> AIMessage(LLM 답변).content -> 문자열 출력
output_parser = StrOutputParser()
````

### chain 생성

````python
# LCEL 방식 사용 -> "|" 사용
rag_chain = prompt | llm | output_parser
````

## Chatbot 실행

````python
# 챗봇 실행을 위한 while 반복문 정의 및 실행

## 안내문 출력
print("안녕하세요! 고객 응대 채봇입니다. 궁금한 점을 질문해주세요. (종료하려면 'exit' 입력)")

## while 조건문 작성
while True:
    try:
        ### 사용자 입력 -> 질문 생성
        user_query = input("질문: ")

        ### 종료 조건 생성
        if user_query.lower() == 'exit':
            print("채봇을 종료합니다. 이용해주셔서 감사합니다.")
            break

        ### 검색 도구를 사용하여 사용자 질문과 가장 유사한 문서(Document) 리스트 생성
        retrieved_docs = retriever.invoke(user_query)
        print(f'사용자 질문과 가장 유사한 문서 리스트: \n{retrieved_docs}')

        ### 검색된 문서 리스트로부터 page_content 추출  
        context_text = retrieved_docs[0].page_content
        print(f'답변의 근거: \n{context_text}')

        print('-'*80)

        ### rag_chain.invoke()를 호출, 사용자 질문에 대한 답변 생성
        response = rag_chain.invoke({
            'context': context_text,
            'question': user_query
        })
        print(f'chatbot 응답: \n{response}')
        print('-'*80)

    except Exception as e:
        print(f'오류가 발생합니다.: {e}')
````

## 라. 벡터 데이터베이스 (Vector DB)
* **개념**: 텍스트, 이미지 등의 비정형 데이터를 임베딩 모델을 통해 벡터로 변환하여 저장하고, 유사도 연산을 수행하는 전용 데이터베이스
* **핵심 역할**:
  * **지식 컨텍스트 수집(Retrieval)**: LLM이 답변을 생성하기 전 질문과 의미적으로 가장 가까운 문서를 초고속 검색
  * **환각(Hallucination) 방지**: 최신 데이터나 기업 내부 데이터를 답변의 근거로 제공
* **종류**: FAISS, ChromaDB

````python
# 필요한 모듈 / 함수 / 클래스 임포트
import os
from dotenv import load_dotenv
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
````

````python
# .env file -> API Key 가져오기
env_file_path='./.env'
output = load_dotenv(env_file_path)
print(output)
````

````python
# 내용 정리
"""
1. TextLoader
   1) 데이터 로드 및 변환 도구
   2) 기능 1: 외부 파일로 부터 텍스트 데이터를 추출
   3) 기능 2: 텍스트 데이터를 LangChain의 기본 자료형 Document 자료형으로 변환
2. Document 자료형:
   1) 텍스트 데이터에 대한 LangChain의 기본 자료형
   2) 메타데이터(meatdata) + 텍스트 내용(page_content)으로 구성
3. load() 함수
   1) TextLoader의 전용 함수
   2) TextLoader 생성 시 지정된 파일을 지정된 encoding 방식으로 읽고 Document 자료형으로 변환
"""

# 데이터 로드 및 변환 도구 생성
txt_file_path='./데이터/manual.txt'
loader = TextLoader(file_path=txt_file_path, encoding='utf-8')

# 기능: 전용 함수 -> load() 함수 호출 -> txt 파일(텍스트) -> Document 자료형으로 변환
documents = loader.load()
print(f'documents: {documents}')
````

````python
# 내용 정리
"""
1. 임베딩 모델이 한 번에 처리할 수 있는 입력 길이 제한(8,192 토큰)을 고려하여 긴 문서를 
   작은 조각(chunk)로 분할하거나 짧은 문서 조각을 서로 병합하여 사용자 질문과 유사도 계산
   시 정확도를 높일 수 있는 적절한 길이로 만든다.
2. RecursiveCharacterTextSplitter
   1) 데이터 분할 도구
   2) 분할 순서
      - 기본 설정: \n\n 단락(문단), \n문장,공백(" "), 글자 단위("")  
      - 먼저 가장 큰 단위인 \n\n(단락)을 기준으로 텍스트를 쪼갠다
      - 쪼개진 텍스트 조각이 여전히 사용자가 설정한 기준 길이(chunk_size)보다 크다면, 
        그 다음 구분자인 \n(줄바꿈/문장)으로 다시 쪼갠다
      - 설정한 크기 이하가 될 때까지 이 과정을 공백( ), 글자 단위("") 순으로 반복한다.
      - 기준 길이(chunk_size)보다 짧으면 병합
   3) Document 자료형의 텍스트 데이터를 분할
3. 매개 변수 정리
   1) chunk_size: 각 청크의 최대 문자 수
   2) chunk_overlap: 인접한 청크 간의 중복 문자 수 -> QA, 법률 -> overlap=0 / 일반 소설 -> chunk_size 10% 
4.split_documents() 함수
   1) RecursiveCharacterTextSplitter 전용 함수
   2) 입력된 Document 자료형의 텍스트 데이터를 설정 기준(chunk_size, chunk_overlap)에 따라 분할
"""

# 데이터 분할 도구 생성
text_splitter = RecursiveCharacterTextSplitter(chunk_size=160, chunk_overlap=0)

# split_documents() 함수 호출, 분할된 문서 생성
docs = text_splitter.split_documents(documents=documents)
print(f'분할된 문서: \n{docs}')
print('-'*80)
print(f'분할된 문서의 수: {len(docs)}')
print('-'*80)
print(f'분할된 첫 번째 문서: \n{docs[0]}')
````

````python
# OpenAI -> 임베딩 모델(API)

# 내용 정리
"""
1. OpenAIEmbeddings: OpenAI의 임베딩 모델을 LangChain에서 사용할 수 있도록 지원하는 클래스
2. 매개 변수 정리
   1) model: 사용할 OpenAI 임베딩 모델의 이름 지정 ('text-embedding-3-small')
"""

## 모델 이름 설정
model_name = "text-embedding-3-small"

## OpenAI 임베딩 모델 생성
model = OpenAIEmbeddings(
    model=model_name
)
````

````python
# 내용 정리
"""
1. 데이터 저장 테이블 
   1) embedding_metadata: 원본 텍스트, 메타 데이터 
   2) embeddings_queue: 임베딩 벡터 
2. from_documents() 함수
   1) ChromaDB 전용 함수
   2) 입력: docs (분할된 문서 조각, Document의 리스트)와 임베딩 모델
   3) 기능: 각 Document의 텍스트를 임베딩하고 메모리에 저장 -> 결과물: 벡터 저장소(vectorstore)
3. as_retriever() 함수
   1) ChromaDB 전용 함수. 검색 도구 생성
   2) invoke 함수를 통해서 사용자 질문 입력
      - 사용자 질문을 받을 경우, 벡터 저장소에 저장된 텍스트 중에서 사용자의 질문과 가장 유사한 텍스트를 찾아주는 역할 수행
      - search_kwargs={'k': n} -> 사용자 질문과 가장 유사도가 높은 n개의 텍스트 데이터를 가져오도록 설정
"""

# 백터 저장소 저장 경로 설정
chroma_path='./chroma_db'

# 이미 생성된 백터 저장소가 있다면 불러오고, 없다면 새로 생성
if os.path.exists(chroma_path) and os.listdir(chroma_path):
    print("기존 벡터 저장소를 불러옵니다.")
    # 이미 생성된 벡터 저장소를 불러올 때 (from_documents 사용 금지)
    vectorstore = Chroma(
        persist_directory=chroma_path,
        embedding_function=model  # 주의: embedding_arguments가 아닌 embedding_function 매개변수 사용
    )
else:
    print("새로운 벡터 저장소를 생성합니다.")
    # 문서가 처음이라 저장소가 없을 때 새로 생성
    vectorstore = Chroma.from_documents(
        documents=docs,
        embedding=model,
        persist_directory=chroma_path
    )
````

````python
# os.path.exists, os.listdir 기능

"""
1. os.path.exists(path)
: 지정한 경로에 파일이나 폴더가 실제로 존재하는지 확인하는 함수

2. os.listdir(=폴더)
: 지정한 경로에 있는 파일과 폴더 목록을 리스트로 반환하는 함수
"""

output1 = os.path.exists(chroma_path)
print(output1)

print('-'*80)

output2 = os.listdir(chroma_path)
print(output2)
````

````python
# 챗봇 실행을 위한 while 반복문 정의 및 실행

## 안내문 출력
print("안녕하세요! 대한민국헌법 Q&A 챗봇입니다. 헌법 조문이나 기본권 등에 대해 질문해주세요. (종료하려면 'exit' 입력)")
print('-' * 80)

## while 조건문 작성
while True:
    try:
        ## 사용자 입력(질문) 생성
        user_query = input("질문: ")

        ## 사용자 입력 -> 종료 조건 만족 -> 종료
        if user_query.lower() == 'exit':
            print("헌법 Q&A 챗봇을 종료합니다. 이용해주셔서 감사합니다.")
            break

        ## 검색 도구를 사용, 사용자 질문과 가장 유사한 Document 2개 -> 리스트 추출
        retrieved_docs = retriever.invoke(user_query)

        ## 검색 결과 -> 텍스트 추출 -> 병합 

        ### 1. 추출한 텍스트를 담을 빈 리스트 생성
        contents = []  

        ### 2. 검색된 2개의 Document 리스트 -> for문 -> Document 별 텍스트 추출 및 contents에 추가
        for doc in retrieved_docs:
            contents.append(doc.page_content)  
        
        ### 3. 문자열 전용 함수 join() 함수 -> 병합 기준: "\n\n"
        context_text = "\n\n".join(contents)

        print(f"[검색된 참조 조문 근거]:\n{context_text}")
        print('-' * 80)

        ## rag_chain.invoke() 함수 실행 -> user_query, context_text 전달
        response = rag_chain.invoke({
            'context': context_text,
            'question': user_query
        })

        print(f"chatbot 응답:\n{response}")
        print('-' * 80)

    except Exception as e:
        print(f'오류가 발생했습니다: {e}')
````

## 바. 밀집 검색(Dense Retrieval)과 희소 검색(Sparse Search)
### 1. **밀집 검색 (Dense Retrieval)**
* **개념**: 텍스트를 임베딩 모델을 통해 고차원의 연속적인 수치 벡터로 변환한 후, 코사인 유사도를 기반으로 가장 유사한 문서를 찾는 방식

### 2. **키워드 기반 희소 검색 (Sparse Search)**
* **개념**: 사용자 질의 상의 중요한 단어가 문서 분량 대비 가장 알차게 들어있는 문서를 찾는 방식
* **필요 산업**: 제조업(부품 번호, 에러 로그), 법률(법령 조문, 판례 번호), 의료/제약(약품명, 질병 코드) 등 정확한 문자열 일치가 필수적인 분야
* **BM25 알고리즘**: TF-IDF 방식을 개선하여 단어 빈도 포화도 조절, IDF, 문서 길이 정규화를 적용한 키워드 검색 알고리즘 (한글 처리를 위해 형태소 분석기 필수)
* **장단점**: 고유명사/에러 코드 검색에 정확하고 빠른 인덱싱이 가능하나, 의미적 유사성이나 동의어 파악이 불가함

````python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter as rcts
from langchain_community.retrievers import BM25Retriever
from mecab import MeCab
````

````python
text_splitter = rcts(
    chunk_size = 160,
    chunk_overlap = 0
)

docs = text_splitter.split_documents(documents)
````

````python
mecab = MeCab()
````

````python
def mecab_tokenizer(text):
    """
    Mecab 형태소 분석기를 사용하여 텍스트에서 의미 있는 형태소(명사, 용언, 숫자)만 추출합니다.
    - N: 명사 (NNG, NNP 등)
    - V: 용언 (VV: 동사, VA: 형용사 등)
    - S: 숫자 및 외국어, 기호 등 (SN, SL 등)
    """

    ## 의미 있는 형태소 추출 결과 저장할 빈 리스트 생성
    tokens = []

    ## 품사 태깅 -> 결과: [('한글', 'NNG'), ...]
    tags = mecab.pos(text)

    ## for문 정의
    for morph, pos in tags:
        if pos.startswith(('N','V','S')):
            tokens.append(morph)

    return tokens
````

## 사. 하이브리드 검색 (Hybrid Search)과 RRF
### 1. **하이브리드 검색 개요**
* **개념**: 키워드 기반의 '희소 검색(Sparse Search)'과 의미/문맥 기반의 '밀집 검색(Dense Search)'을 결합하여 양쪽의 장점을 모두 취하는 통합 검색 방식

### 2. **RRF (Reciprocal Rank Fusion / 상호 순위 융합)**
* **개념**: 서로 다른 스케일과 점수 체계를 가진 희소 검색과 밀집 검색의 결과를 각 검색기에서의 '순위(Rank)'만을 기반으로 공정하게 재정렬하여 결합하는 알고리즘
* **LangChain 구현**: `EnsembleRetriever(retrievers=[bm25_retriever, dense_retriever], weights=[0.5, 0.5])`를 통해 가중치를 부여하여 통합 순위 계산

````python
file_path="./데이터/tech_support_manual.txt"

### with open 구문, 문자열로 변환
with open(file_path, "r", encoding="utf-8") as file:
    manual_text = file.read()

## 문자열 -> split() -> 질문-답변 쌍으로 분할
qa_pairs = manual_text.split("\n\n")
print(qa_pairs)

print('-'*80)

## 각 질문-답변 쌍 별 글자수 계산

### 결과 저장 리스트 생성
character_length=[]

### for문 + len 함수 + append 함수
for pair in qa_pairs:
    character_length.append(len(pair))

print(character_length)

print('-'*80)

## 질문-답변 쌍 중 가장 글자 수가 긴 경우의 길이 추출
max_char_length = max(character_length)
print(f"가장 긴 Q&A 단락의 글자 수: {max_char_length}자")

print('-'*80)

## optimal_chunk_size 설정
optimal_chunk_size = 210
````

````python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=optimal_chunk_size,
    chunk_overlap=0
)

# 전용 함수: split_documents() 함수 호출 -> 분할된 문서 생성
docs = text_splitter.split_documents(documents)
print(f'분할된 원본 문서: \n{docs}')
print('-'*80)
print(f'분할된 문서 조각의 수: {len(docs)}')
print('-'*80)
print(f'분할된 문서 조각 예시: \n{docs[0]}')
````

````python
mecab = MeCab()
````

````python
def mecab_tokenizer(text):
    
    ## 의미 있는 형태소(명사, 용언, 숫자, 영어 및 특수 기호 등) 추출 결과 저장용 빈 리스트 생성
    tokens = []

    ## 품사 태깅 -> [('형태소', '품사'), ...]
    tags = mecab.pos(text)

    ## for문 정의
    for morph, pos in tags:
        if pos.startswith(('N', 'V', 'S')):
            tokens.append(morph)
    return tokens
````

````python
sparse_retriever = BM25Retriever.from_documents(
    documents=docs,
    preprocess_func=mecab_tokenizer,
    # 연관성이 높은 문서 추출 조건 설정
    k=2
) 
````

````python
test_query = "ERR-502 오류코드 조치법"

## invoke() 함수 호출 -> 검색 실행 -> 결과: [Document1, Document2] 
retrieved_docs = sparse_retriever.invoke(test_query)

## for문 -> 각 Document -> page_content -> 텍스트 추출
for doc in retrieved_docs:
    print(f"내용: {doc.page_content}")
    print('-'*80)
````

````python
hybrid_retriever = EnsembleRetriever(
    retrievers=[sparse_retriever, dense_retriever],
    weights=[0.5, 0.5]
)

print(f"하이브리드 검색 도구: {hybrid_retriever}")
````

````python
system_prompt = """
당신은 스마트홈 IT 기기 및 지원 서비스의 전문 기술 지원 AI 엔지니어입니다.
제공된 [참고 컨텍스트] 정보를 바탕으로 고객의 제품 문의, 오류 코드 해결법, A/S 규정 질문에 정중하고 명확하게 답변해주세요.

[답변 작성 수칙]
1. [참고 컨텍스트]에 명시된 모델명, 오류 코드, 조치 단계, 규정 조항을 정확하게 인용하세요.
2. 답변을 시작할 때 '답변:'이라는 단어를 사용하지 마세요.
3. [참고 컨텍스트]에 고객 질문에 관한 내용이 없다면, "죄송합니다. 문의하신 내용에 대한 기술 매뉴얼 정보를 찾을 수 없습니다. 고객센터(1588-0000)로 문의 부탁드립니다."라고 답변하세요.

[참고 컨텍스트]
{context}
"""

# 사용자 프롬프트 생성
user_prompt="{question}"

# 전체 프롬프트 생성
prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", user_prompt)
])

print(prompt)
````

````python
llm = ChatOpenAI(model='gpt-5-nano')
````

````python
output_parser = StrOutputParser()
````

````python
rag_chain = prompt | llm | output_parser
````

````python
print("안녕하세요! 하이브리드 RAG 기반 고객 응대 챗봇입니다. 궁금한 점을 질문해주세요. (종료하려면 'exit' 입력)")

print('-'*80)

while True:
    try:
        ## 사용자 입력(질문) 생성
        user_question = input("사용자 질문: ")

        ## 사용자 입력 -> 종료 조건 만족 -> 종료
        if user_question.lower() == 'exit':
            print("챗봇을 종료합니다. 이용해주셔서 감사합니다.")
            break

        ## 하이브리드 검색 실행 (sparse 2개 + dense 2개 -> 결과 -> RRF 적용 -> 문서별 순위)
        retrieved_docs = hybrid_retriever.invoke(user_question)
        print(f'하이브리드 검색 결과: 총 {len(retrieved_docs)}개 문서 수집')

        print('-'*80)

        ## 하이브리드 검색 실행 -> 추출된 문서 리스트 -> 문자열 추출 -> 병합: "\n\n".join([text1, text2, ...]) 

        ### 추출한 텍스트를 담을 빈 리스트 생성
        contents = []

        ### 검색된 문서 리스트를 순회하며 텍스트(page_content) 추출
        for doc in retrieved_docs:
            text=doc.page_content
            contents.append(text) 

        ### 리스트(contents)의 모든 문자열 -> join() 함수 -> "\n\n" 기준으로 연결
        context_text = "\n\n".join(contents)
        print(f'답변의 근거: \n{context_text}')

        print('-'*80)

        ## rag_chain.invoke()를 호출, 사용자 질문에 대한 답변 생성
        response = rag_chain.invoke({
            "context": context_text,
            "question": user_question
        })

        ## 결과 출력
        print(f"챗봇 응답: \n{response}")
        print('='*80)

    except Exception as e:
        print(f'오류가 발생했습니다: {e}')
````

# 25. 서울시 상권 분석 및 보고서 작성 멀티 에이전트 개발

## 가. LangChain의 한계와 LangGraph의 등장
* **LangChain의 한계**: 전체 작업 흐름이 '컨베이어 벨트'처럼 단방향 비순환 경로로 진행되어 중간 분기나 순환 실행(Self-Correction)이 어려움
* **LangGraph 정의**: 복잡한 AI 에이전트를 만들기 위한 프레임워크로, AI의 사고와 도구 사용, 결정 과정을 '순서도'처럼 명확하게 설계하고 제어
* **핵심 구성 요소**:
  * **노드 (Node)**: 작업 주체 (Python 함수 또는 에이전트)
  * **엣지 (Edge)**: 노드와 노드를 연결하는 연결선 (일반 엣지 및 조건부 엣지)
  * **상태 (State)**: 워크플로우 전체를 흐르며 데이터를 관리하는 공유 메모리이자 통신 규약 (TypedDict 또는 Pydantic BaseModel 활용)

## 나. LangGraph 조건부 엣지와 멀티 에이전트
* **조건부 엣지**: `add_conditional_edges()` 메서드를 사용하여 출발 노드, 조건 판단 함수, 결과에 따른 경로 맵핑을 연결
* **싱글 에이전트 vs 멀티 에이전트**: 복잡하고 다양한 하위 작업으로 나눌 수 있는 대규모 문제 해결에는 여러 전문 에이전트가 협력하는 멀티 에이전트 시스템이 유리

````python
# LangGraph 상태 정의: TypedDict 클래스 상속

"""
# 첫 번째 항목(key): messages -> 대화 메세지 입력
  1) 대화 메세지: HumanMessage, AIMessage, ToolMessage 
  2-1) 대화 흐름(예시): 사용자 질문 -> LLM 답변(도구 호출 요청) -> 도구 사용
  2-2) 대화 메세지 생성 순서: HumanMessage -> AIMessage -> ToolMessage
  2-3) messages 항목의 값 업데이트
       : [HumanMessage] -> [HumanMessage, AIMessage] -> [HumanMessage, AIMessage, ToolMessage]  
  3) 잘못된 값의 자료형 -> messages: list[BaseMessage] 
                      -> 결과: 기존 대화 메시지가 새로운 대화 메시지로 대체
                      -> 예시: [HumanMessage(content="안녕")] -> [AIMessage(content="반갑습니다! 무엇을 도와드릴까요?")]
# Annotated 클래스 기능
  1) 기존 타입 힌트에 부가 정보(함수, 설명 등)를 추가하는 도구
  2) 코드: Annotated[T(기존 자료형: list[BaseMessage]), 함수, 설명글]

# add_messages 함수 기능
  1) 기존 메시지 리스트와 새 메시지 리스트를 연결 
  2) [HumanMessage] + [AIMessage] -> [HumanMessage, AIMessage] -> 대화 기록 누적 

# 정리: 
  1) 목표: 모든 대화 기록을 messages(키)에 누적해서 저장
  2) 목표 구현 예시: [HumanMessage] -> [HumanMessage, AIMessage] -> [HumanMessage, AIMessage, ToolMessage] 
  3) 구현 방법: messages: Annotated[list[BaseMessage], add_messages]
"""

class AgentState(TypedDict):
    # 대화 메세지 누적 구조 생성
    messages: Annotated[list[BaseMessage], add_messages]
    # 도구 실행 결과 텍스트 저장
    tool_output: str | None
````

````python
# 내용 정리

"""
# 노드를 구성하는 함수
  1) 매개 변수의 자료형: 'AgentState'로 지정
  2) 매개 변수의 입력값 -> {'messages':[BaseMessage], 'tool_output']}
  3) 실행 결과 -> 딕셔너리로 반환
  4) LangGraph 시스템: 실행 결과를 '상태'에 반영 (업데이트) 한다.

# 진행 순서(가정)
  : 사용자 요청 -> call_llm(도구 사용 요청) 함수 -> call_tool(도구 실행) 함수 -> summarize_result(결과 요약) 함수
"""

# llm 호출 함수 정의(Type Hint 적용)
def call_llm(state: AgentState) -> dict:
    print("---LLM 호출 노드 실행---")
    print('-'*80)

    ## state = {'messages':[BaseMessage], 'tool_output']} -> 대화 히스토리 -> 'messages'키의 값 추출
    messages = state["messages"]

    ## messages 상태 (1턴) -> [HumanMessage]
    ## messages 상태 (2턴) -> [HumanMessage, AIMessage, ToolMessage, HumanMessage]

    ## if 조건문 이용, 마지막 메시지가 HumanMessage인지 확인 후 content 값 가져오기
    if messages and isinstance(messages[-1], HumanMessage):
        last_user_message = messages[-1].content
    else:
        last_user_message = ""

    print(f'사용자 입력: {last_user_message}')

    print('-'*80)

    ## 실제: LLM -> '사용자 입력' 분석 -> 도구 사용 할지 말지 판단 -> 가정: 도구 사용을 지시

    ## call_llm 함수 실행의 결과 -> 도구 사용을 지시 -> AIMessage 생성
    ## 실제는 llm 응답 결과를 반환, 여기서는 AIMessage 클래스를 사용해서 직접 결과를 생성
    return {"messages":[AIMessage(content="도구 사용을 요청합니다.")]}

# 도구 실행 함수 정의
def call_tool(state: AgentState) -> dict:
    print("---도구 노드 실행---")

    print('-'*80)

    ## 도구 실행: 여기서는 도구 실행을 시뮬레이션한다.
    print("도구 실행 중...")
    tool_result = "도구 실행 완료."

    ## 함수 실행의 결과 -> 도구 응답(ToolMessage)
    ## 도구 실행의 결과는 ToolMessage 자료형으로 생성, 
    ## 여기서는 ToolMessage 클래스를 이용하여 messages 처리
    return {
        "messages": [ToolMessage(content=tool_result, tool_call_id="searcg_tool_call_id_123")],
        "tool_output": tool_result
    }

# 결과 요약 함수 정의(llm 최종 답변 생성)
def summarize_result(state: AgentState) -> dict:
    print("---결과 요약 노드 실행---")

    print('-'*80)

    ## 상태 -> 도구 실행의 결과 추출
    tool_result = state["tool_output"]

    ## 결과 요약: 실제로는 llm이 수행
    summary = f"도구 실행 결과 '{tool_result}'를 요약했습니다."
    # print(f"\n결과 요약: {summary}")

    ## 함수 실행의 결과 -> LLM의 응답(AIMessage)
    ## 실제는 llm의 요약 결과를 반환, 여기서는 AIMessage 클래스를 사용해서 직접 결과를 생성하고 반환
    return {"messages": [AIMessage(content=summary)]}   
````

````python
# 사용자 요청 -> call_llm -> call_tool -> summarize_result -> 종료

# 그래프 생성 도구 호출, 그래프 생성
graph = StateGraph(AgentState)

# 그래프에 노드 추가 -> StateGraph 클래스 전용 함수 -> add_node()
graph.add_node("call_llm_node", call_llm)
graph.add_node("call_tool_node", call_tool)
graph.add_node("summarize_node", summarize_result)

# 시작점 설정 -> StateGraph 클래스 전용 함수 -> set_entry_point()
graph.set_entry_point("call_llm_node")

# 그래프에 엣지 추가 -> StateGraph 클래스 전용 함수 -> add_edge()

## LLM 실행 후 도구 호출
graph.add_edge("call_llm_node", "call_tool_node")

## 도구 사용 후 결과 요약
graph.add_edge("call_tool_node", "summarize_node")

## 결과 요약 후 종료
graph.add_edge("summarize_node", END)
````

````python
agent = graph.compile()
````

````python
print("=== TypedDict 기반 에이전트 실행 ===")

# 사용자 질문 생성
user_question = "nvidia 2026년도 2분기 매출액은?"

# 사용자 질문 -> '상태'에 입력 가능한 형태(규칙 준수) -> dict {"message": [HumanMessage(content=user_question)]} 
initial_state = {"messages": [HumanMessage(content=user_question)]}

# agent 실행 -> 최종 결과 생성
# 결과물 추적: HumanMessage(사용자 요청) -> AIMessage(도구 호출 요청) -> ToolMessage(도구 작동) -> AIMessage(요약 생성) 
# 최종 상태: "messages": [HumanMessage, AIMessage(도구 호출 요청), ToolMessage(도구 작동), AIMessage(요약 생성)] 
final_state = agent.invoke(initial_state)

# 결과 확인
print(f"\n최종 상태: \n{final_state}")
print('-'*80)
print(f"최종 상태 메시지: {final_state['messages']}")
print('-'*80)
print(f"최종 LLM 출력: {final_state['messages'][-1].content}")
````

````python
from pydantic import BaseModel, Field
from langchain_core.messages import BaseMessage, HumanMessage, AIMessage
from langchain_openai import ChatOpenAI
from langchain_tavily import TavilySearch
from langgraph.graph import StateGraph, END, add_messages
````

````python
# ChatOpenAI 클래스 -> llm 생성
llm = ChatOpenAI(model="gpt-5-nano")
````

````python
# TavilySearch 클래스 -> Tavily 검색 도구 생성 
search_tool = TavilySearch(max_results=3)
````

## 상태 정의

````python
# Pydanctic 라이브러리의 BaseModel 클래스 상속 -> 커스텀 자료형 생성
class AgentState(BaseModel):
    # 상태 메시지 누적 구조 생성
    messages: Annotated[list[BaseMessage], add_messages] = []
    # 웹 검색 필요 여부에 대한 표시 (True: 검색 필요, False: 불필요)
    needs_search: bool = False
    # Tavily 검색 결과 저장
    search_results: str = ""
````

## 노드 함수 정의

````python
# 사용자 질문 의도 분류 노드 함수 정의
"""
# 사용자의 질문을 분석 후 실시간 웹 검색이 필요한지를 단답형(SEARCH / NO_SEARCH)으로 생성
"""

def classifier_node(state: AgentState) -> dict:
    print("[Node 1: 질문 의도 분류] 실행 중...")
    print('-'*80)

    # 상태 state 출력
    print(f'state: {state}')
    print('-'*80)

    # 대화 스토리 -> 사용자 질문 추출 -> 상태 -> messages 
    user_question = state.messages[-1].content
    print(f"[사용자 질문 원본]: {user_question}")

    print('-'*80)

    # 프롬프트 생성: LLM에게 SEARCH 또는 NO_SEARCH 단어 하나만 반환하도록 명확하게 지시

    ## 시스템 프롬프트
    system_prompt="""
    당신은 사용자의 질문을 분석하는 전문가입니다.
    사용자의 질문에 답변하기 위해 실시간 웹 검색이 반드시 필요한지 판단하세요.

    - 최신 시사 뉴스, 실시간 정보, 특정 연도 실적(예: 2026년 실적) 등 최신 데이터 조회가 필요한 경우: 'SEARCH'
    - 일반 상식, 개념 설명, 프로그래밍 문법, 단순 인사말 등 모델 자체 지식으로 충분한 경우: 'NO_SEARCH'

    반드시 'SEARCH' 또는 'NO_SEARCH' 단어 하나만 정확하게 출력하세요. 다른 설명은 일체 포함하지 마세요.
    """    

    ## 전체 프롬프트 작성
    messages = [
        ("system",system_prompt),
        ("user", f"질문: {user_question}")
    ]
    response = llm.invoke(messages) 
    ## llm의 답변 -> response -> AIMessage(content="답변")
    print(f'llm의 답변: \n{response}')

    print('-'*80)

    ## llm의 답변 -> 텍스트('SEARCH ' 또는 'NO_SEARCH') 추출
    decision = response.content.strip().upper()

    ## 단답형 응답 문자열 -> 블리언 변환
    needs_search = (decision=="SEARCH")
    print(f"LLM 판단 결과: {decision} (검색 필요 여부: {needs_search})")

    ## needs_search 값을 반환 -> 상태 업데이트
    return {"needs_search": needs_search}
````

````python
# Tavily 웹 검색 노드 함수 정의
def tavily_search_node(state: AgentState) -> dict: 
    """
    # 기능: TavilySearch 도구를 실행하여 최신 웹 정보를 수집하고 텍스트로 가공한다.
    """
    print("[Node 2: Tavily 웹 검색 도구 실행] 실행 중...")
    print('-'*80)

    # 대화 히스토뢰 -> 사용자 최신 질문 추출 -> state.messages[-1] 
    query = state.messages[-1].content
    print(f"[검색 쿼리]: {query}")

    print('-'*80)

    # TavilySearch 검색 도구 실행
    try:
        # TavilySearch 검색 도구 실행 
        # 실행 결과: 딕셔너리 -> 키: results -> 값: 딕셔너리의 리스트 (검색 결과 3개: 딕셔너리 3개) 
        raw_results = search_tool.invoke({"query": query})

        ## 1. 도구 실행 결과 -> 딕셔너리 -> 'reuslts'키 -> 값: 리스트 추출, 없으면 빈 리스트
        if isinstance(raw_results, dict):
            results = raw_results.get("results", [])
        else:
            results = []

        ## 2. 리스트([{검색결과1}, ...]) -> 각 검색 결과(dict) -> 출처(url)와 본문(content) 추출
        context_parts = []
        for item in results:
            if isinstance(item, dict):
                url = item.get("url", "")
                content = item.get("content", "")
                context_parts.append(f"출처: {url}\n내용: {content}")
            else:
                context_parts.append(str(item)) 

        ## 3. 3개의 검색 결과 문자열(출처+본문) -> 하나의 문자열로 병합 -> "\n\n".join()
        if context_parts:
            combined_results = "\n\n".join(context_parts)
            print(f"웹 검색 성공 (총 {len(context_parts)}개 문서 수집 완료)")
        else:
            combined_results = "검색 결과가 없습니다."
            print("검색 결과가 비어 있습니다.")

    except Exception as e:
        error_message=f"검색 중 오류 발생: {e}"
        print(error_message)

    # 정제된 검색 결과(combined_results) 반환 -> LangGraph -> 상태 업데이트
    return {"search_results": combined_results}
````

````python
# 최종 답변 생성 노드 함수 정의
def answer_node(state: AgentState) -> dict:
    """
    # 기능: 검색 결과 유무에 따라 시스템 프롬프트를 동적으로 변경하여 최적의 답변을 생성
    """
    print("[Node 3: 최종 답변 생성] 실행 중...")
    print('-'*80)

    # 대화 히스토리 -> 사용자 질문 추출 -> HumanMessage -> state.messages[-1]
    user_question = state.messages[-1].content

    # 검색을 거쳐온 경우와 거치지 않은 경우에 맞춰 시스템 프롬프트를 다르게 구성
    if state.needs_search and state.search_results:
        # 웹 검색 결과 -> 프롬프트에 반영
        system_prompt = f"""
        당신은 정확한 최신 정보를 제공하는 전문 AI 비서입니다.
        아래 제공된 [웹 검색 결과]를 바탕으로 사용자의 질문에 친절하고 명확하게 답변해 주세요.

        [웹 검색 결과]
        {state.search_results}
        """
    else:
        # 일반적인 질문인 경우: 일반적인 친절한 AI 비서로서의 지침 부여
        system_prompt = """당신은 친절하고 똑똑한 AI 비서입니다. 사용자의 질문에
        명쾌하게 답변해 주세요."""

    # 전체 프롬프트 작성
    messages=[
        ("system", system_prompt),
        ("user", user_question)
    ]

    # llm.invoke() 함수 호출 -> llm 답변 -> AIMessage(content="텍스트")
    response=llm.invoke(messages)
    # print(f"[최종 AI 답변 원본]: \n{response}")

    print('-'*80)

    # 생성된 답변을 AIMessage 형태로 전달(return) -> LangGraph 시스템 -> 상태 업데이트
    return {"messages":[AIMessage(content=response.content)]}
````

## 라우팅 함수 정의

````python
# 기능: state의 'needs_search' 값을 읽고, classifier_node 다음에 실행할 노드의 이름을 반환

def route_by_search_need(state: AgentState) -> str:
    if state.needs_search:
        print("[Router] 최신 정보 검색이 필요하므로 'search_node'로 이동합니다.")
        return "search_node"  
    else:
        print("[Router] 검색이 불필요하므로 즉시 'answer_node'로 직행합니다.")
        return "answer_node"     
````

## 그래프 구성 

````python
# '상태' 기반 빈 그래프(노드X, 엣지X) 생성 -> StateGraph 클래스 호출 
graph = StateGraph(AgentState) 

# 그래프에 노드 추가 -> StateGraph 클래스 -> 전용 함수 -> add_node 함수 호츨
graph.add_node("classifier_node", classifier_node)
graph.add_node("search_node", tavily_search_node)
graph.add_node("answer_node", answer_node)

# 시작점 설정: 질문이 들어오면 항상 분류 노드부터 시작 -> set_entry_point 함수 호출
graph.set_entry_point("classifier_node")

# 그래프에 엣지 추가

## 조건부 엣지 연결 -> add_conditional_edges() 함수 호출
"""
# classifier_node 실행 후, route_by_search_need 함수의 반환값에 따라 분기(edge 생성)
# 결과: classifier_node->search_node 또는 classifier_node->answer_node
"""
graph.add_conditional_edges(
    "classifier_node",
    route_by_search_need,
    {
        "search_node": "search_node",
        "answer_node": "answer_node"
    }
)

## 검색 노드 실행 후에는 최종 답변 생성 노드 이동
graph.add_edge("search_node", "answer_node")

## 최종 답변 생성이 끝나면 워크플로우 종료
graph.add_edge("answer_node", END)
````

````python
# 사용자 질문 -> 모호한 질문 -> 명확한 질문 -> 검색 기능 -> 검색 -> 답변의 근거 -> 답변 생성
# [답변 생성 체인] 검색된 문서를 바탕으로 최종 답변 작성

## system 프롬프트 생성
qa_system_prompt = """
당신은 코드크래프터스 아카데미의 친절한 고객 지원 담당자입니다.
아래의 [Context] 정보를 사용하여 사용자의 질문에 답변하세요.
정보가 없다면 '죄송하지만 매뉴얼에서 정보를 찾을 수 없습니다'라고 답하세요.

[Context]
{context}
"""

## chain 전달 -> 전체 프롬프트 생성 -> ChatPromptTemplate.from_messages()
qa_prompt=ChatPromptTemplate.from_messages(
    [
        ("system", qa_system_prompt),
        MessagesPlaceholder(variable_name="chat_history"),
        ("human", "{input}")
    ]
)

print(qa_prompt)

## chain 생성 -> LLM 답변 -> AIMessage 생성 -> 상태 업데이트 
qa_chain = qa_prompt | llm 
````

````python
# pydantic 라이브러리 -> BaseModel 클래스 상속 -> 상태 정의
class AgentState(BaseModel):
    # 대화 내역 누적 항목
    messages: Annotated[list[BaseMessage], add_messages] = []
    # 질문 재구성 에이전트가 생성한 검색어 입력 항목
    rephrased_query: str = ""
    # 문서 검색 에이전트가 찾은 매뉴얼 내용 입력 항목
    context: str = ""
````

````python
# 1. 질문 재구성 노드 함수 정의(설계: 사용자 질문 -> 질문 재구성 노드)

# 내용 정리
"""
# 1턴 진행 중: 사용자 첫 번째 질문 -> messages: [HumanMessage]
# 2턴 진행 중: 사용자 두 번째 질문 -> messages: [HumanMessage, AIMessage, HumanMessage]
# 3턴 진행 중: 사용자 세 번째 질문 -> messages: [HumanMessage, AIMessage, HumanMessage, AIMessage, HumanMessage]
"""

def rephrase_node(state: AgentState) -> dict:
    print("에이전트 1: 질문 맥락 분석 및 재구성")

    print('-'*80)

    # 상태 state 출력    
    print(f'state: \n{state}')

    print('-'*80)

    # 대화 히스토리 -> 사용자 최신 질문 추출
    latest_question = state.messages[-1].content

    # 대화 히스토리 -> 사용자 최신 질문 이전 대화 내역 -> 리스트 자료형
    chat_history = state.messages[:-1]

    # 이전 대화 기록이 존재할 때만 재구성 체인 실행 
    if chat_history:
        query = rephrase_chain.invoke({
            "chat_history": chat_history, 
            "input": latest_question
        })
    else:
        # 사용자 질문 -> 첫 번째 -> 재구성된 질문 = 사용자 질문
        query = latest_question

    print(f"[결과] 재구성된 쿼리: {query}")

    # 답변 생성(return) -> dict 자료형 -> LangGraph 시스템 -> 상태 업데이트
    return {"rephrased_query": query}
````

````python
# 2. 검색 노드 함수 정의
def retriever_node(state: AgentState) -> dict:
    print("에이전트 2: 관련 매뉴얼 검색")

    # 상태 -> 재구성된 질문 추출
    query = state.rephrased_query

    # 검색어: 재구성된 질문 -> 검색 도구 실행 -> 결과: [Document]
    retrieved_docs = retriever.invoke(query)

    # 검색된 문서 리스트를 순회하며 텍스트(page_content) 추출
    contents=[]
    for doc in retrieved_docs:
        text=doc.page_content
        contents.append(text)

    # 리스트(contents)의 모든 문자열 -> join 함수 -> "\n\n" 기준으로 연결
    context_text = "\n\n".join(contents)
    print(f'답변의 근거: \n{context_text}') 

    print('-'*80)

    # return -> dict 자료형 -> LangGraph 시스템 -> 상태 업데이트
    return {'context' : context_text}
````

````python
def answer_node(state : AgentState) -> dict :
    latest_question = state.messages[-1].content
    chat_history = state.messages[:-1]
    context = state.context
    response = qa_chain.invoke({
        'context' : context,
        'chat_history' : chat_history,
        'input' : latest_question
    })

    return {'messages' : [response]}
````

````python
graph = StateGraph(AgentState)

graph.add_node('rephrase_node', rephrase_node)
graph.add_node('retriever_node', retriever_node)
graph.add_node('answer_node', answer_node)

graph.set_entry_point('rephrase_node')

graph.add_edge('rephrase_node', 'retriever_node')
graph.add_edge('retriever_node', 'answer_node')
graph.add_edge('answer_node', END)

multi_agent = graph.compile()
````

````python
print("안녕하세요! 코드크래프터스 아카데미 지원 챗봇입니다. 무엇이든 물어보세요! (종료: exit)")
print("=" * 80)

# 1턴의 대화 -> END 도달 -> 상태(state) -> 기존 대화 기록 -> 메모리에서 소멸(누적 X)
# 이전의 대화 기록을 유지한 멀티턴 대화 가능 -> 대화 기록과 상태를 계속해서 누적해서 관리할 딕셔너리 필요
# 1턴 시작: inputs = {"messages": []}
# 2턴 시작: inputs = {messages:[HumanMessage, AIMessage]}
# 3턴 시작: inputs = {messages:[HumanMessage, AIMessage, HumanMessage, AIMessage]}
inputs = {"messages": []}

# while 반복문 작성
while True:
    try:
        # 사용자로부터 실시간 질문 입력 받기
        user_input = input("사용자: ")

        # 사용자 질문 -> 종료 조건 만족 -> 종료
        if user_input.lower() == 'exit':
            print("챗봇을 안전하게 종료합니다.")
            break

        # 사용자 입력(text) -> dict 자료형 -> invoke 함수 실행 시 -> 
        # LangGraph 시스템 -> AgentState 자료형 변환 -> 상태 업데이트(messages=[HumanMessage])
        # 정리: 모든 노드 함수의 실행의 결과(return), 초기 상태(사용자 입력) -> dict 자료형 표현

        # 사용자 입력 -> inputs(메시지 목록)에 추가 -> inputs -> {'messages':[HumanMessage]}
        # inputs -> {'messages':[HumanMessage]} -> 초기 상태(initial_state)
        inputs["messages"].append(HumanMessage(content=user_input))

        # agent 실행 -> 최종 상태 생성 -> AgentState 자료형 -> LangGraph 시스템 -> dict 자료형 변환
        # 최종 상태: final_state = {messages:[HumanMessage, AIMessage]}
        final_state = multi_agent.invoke(inputs)

        # 가장 최신 메시지(AIMessage) 추출 -> 텍스트 추출
        ai_answer = final_state['messages'][-1].content
        print(f"AI 에이전트:\n{ai_answer}")
        print("-" * 80) 

        # 다음 대화 턴에서 과거 대화를 기억할 수 있도록 상태를 갱신: final_state -> input
        inputs = final_state
        
        # inputs -> {'messages':[]} -> {'messages': [HumanMessage]} -> {messages:[HumanMessage, AIMessage]}

    except Exception as e:
        print(f'오류가 발생했습니다.: {e}')
````

````python
# 파일 경로 설정
csv_path = "faq_manual.csv"

# CSV 파일 로드 및 변환 도구 생성 -> CSVLoader 클래스 호출
loader = CSVLoader(
    file_path=csv_path,
    encoding="utf-8",
    # 본문(page_content)에 포함할 컬럼 지정
    content_columns=["question", "answer"]
)

# load() 함수 호출 -> Document로 변환
docs = loader.load()

print(f"로드된 총 문서 수: {len(docs)}개")
print("-" * 60)
print(f"[첫 번째 Document 본문(page_content)]\n{docs[0].page_content}")
print("-" * 60)
print(f"[첫 번째 Document 메타데이터(metadata)]\n{docs[0].metadata}")
````

### 텍스트 임베딩 모델 생성

````python
# 모델 이름 설정
model_name = "text-embedding-3-small"

# OpenAI 임베딩 모델 생성
model = OpenAIEmbeddings(
    model=model_name
)
````

### 백터 저장소 및 검색 도구 생성 

````python
# 벡터 저장소 저장 경로 설정
chroma_path = 'chroma_db_multi'

# 이미 생성된 벡터 저장소가 있다면 불러오고, 없다면 새로 생성
if os.path.exists(chroma_path) and os.listdir(chroma_path):
    vectorstore = Chroma(
        persist_directory=chroma_path,
        embedding_function=model
    )
    print("기존 벡터 저장소를 성공적으로 불러왔습니다.")
else:
    vectorstore = Chroma.from_documents(
        documents=docs,
        embedding=model,
        persist_directory=chroma_path
    )
    print("신규 벡터 저장소를 생성하고 저장했습니다.")
````

````python
# 검색 도구 생성 -> as_retriever() 함수 호출
retriever = vectorstore.as_retriever(search_kwargs={'k': 1})
````

````python
# 검색 도구 성능 테스트
user_query = "환불은"
output = retriever.invoke(user_query)
print(f"검색 결과: \n{output}")
````

## 에이전트별 전용 체인 생성

````python
# [질문 재구성 체인] 모호한 질문을 명확한 질문으로 변환

## system 프롬프트 생성
rephrase_system_prompt = """
주어진 채팅 기록(chat history)과 최신 사용자 질문이 있을 때,
이 질문이 채팅 기록의 맥락을 참조하고 있다면 모호한 질문(예: "그거 전액 다 받으려면 조건이 뭐야?"),
채팅 기록 없이도 이해할 수 있는 명확한 질문(예: 수강 신청한 강의를 전액 환불 받으려면 조건이 뭐야?)으로 다시 작성하세요.
질문에 답하지 말고, 오직 질문을 재구성하거나 그대로 반환하기만 하세요.
"""

## 전체 프롬프트 생성
rephrase_prompt = ChatPromptTemplate.from_messages([
    ("system", rephrase_system_prompt),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{input}")
])
print(rephrase_prompt)

## chain 생성
rephrase_chain = rephrase_prompt | llm | StrOutputParser()
````

````python
# [답변 생성 체인] 검색된 문서와 맥락을 바탕으로 최종 답변 작성

## system 프롬프트 생성
qa_system_prompt = """
당신은 코드크래프터스 아카데미의 친절한 고객 지원 담당자입니다.
아래의 [Context] 정보를 사용하여 사용자의 질문에 답변하세요.
정보가 없다면 '죄송하지만 매뉴얼에서 정보를 찾을 수 없습니다'라고 답하세요.

[Context]
{context}
"""

## 전체 프롬프트 생성
qa_prompt = ChatPromptTemplate.from_messages([
    ("system", qa_system_prompt),
    MessagesPlaceholder("chat_history"),
    ("human", "{input}")
])
print(qa_prompt)

## chain 생성
qa_chain = qa_prompt | llm
````

## 그래프 상태 정의

````python
# pydantic 라이브러리 -> BaseModel 클래스 상속 -> 커스텀 자료형 생성
class AgentState(BaseModel):
    # 대화 내역 누적 항목
    messages: Annotated[list[BaseMessage], add_messages] = []
    # 재구성된 질문 저장
    rephrased_query: str = ""
    # 검색된 매뉴얼 내용 저장
    context: str = ""
````

## 그래프 노드 함수 정의

````python
# 1. 질문 재구성 노드 함수 정의(비동기 처리 방식)
async def rephrase_node(state: AgentState) -> dict:
    print("에이전트 1: 질문 맥락 분석 및 재구성")
    print('-'*80)    

    # 대화 히스토리 -> 사용자 최신 질문 추출
    latest_question = state.messages[-1].content

    # 대화 히스토리 -> 사용자 최신 질문 이전 대화 내역 -> 리스트 자료형
    chat_history = state.messages[:-1]

    # 이전 대화 기록이 존재할 때만 재구성 체인 실행 
    # 비동기 처리 방식 chain 실행 -> invoke() 함수 -> await 예약어, ainvoke() 함수  
    if chat_history:
        query = await rephrase_chain.ainvoke({
            "chat_history": chat_history, 
            "input": latest_question
        })
    else:
        # 사용자 질문 -> 첫 번째 -> 재구성된 질문 = 사용자 질문
        query = latest_question

    print(f"[결과] 재구성된 쿼리: {query}")

    # 답변 생성(return) -> dict 자료형 -> LangGraph 시스템 -> AgentState 자료형 변환 -> 상태 업데이트
    return {"rephrased_query": query}
````

````python
# 2. 검색 노드 함수 정의 (비동기 처리 방식)
async def retriever_node(state: AgentState) -> dict:
    print("에이전트 2: 관련 매뉴얼 검색")

    # 상태 -> 재구성된 질문 추출
    query = state.rephrased_query

    # 검색어: 재구성된 질문 -> 검색 도구 비동기 처리 방식 실행 -> 결과: [Document]
    # 비동기 처리 방식 실행 -> invoke() 함수 -> await, ainvoke() 함수
    retrieved_docs = await retriever.ainvoke(query)

    # 검색된 문서 리스트를 순회하며 텍스트(page_content) 추출
    contents=[]
    for doc in retrieved_docs:
        text=doc.page_content
        contents.append(text)

    # 리스트(contents)의 모든 문자열 -> join 함수 -> "\n\n" 기준으로 연결
    context_text = "\n\n".join(contents)
    print(f'답변의 근거: \n{context_text}')
    print('-'*80)

    # 함수 실행의 결과(return) -> dict 자료형 -> LangGraph 시스템 -> AgentState 자료형 변환 -> 상태 업데이트
    return {"context": context_text}
````

````python
# 3. 답변 생성 노드 정의 (비동기 처리 방식)
async def answer_node(state: AgentState) -> dict:
    print("에이전트 3: 최종 답변 작성")
    print('-'*80)

    # 대화 스토리 -> 사용자 최신 질문 추출 -> HumanMessage=state.messages[-1]
    latest_question = state.messages[-1].content 

    # 대화 스토리 -> 사용자 최신 질문 이전 대화 내역 추출
    chat_history = state.messages[:-1]

    # 상태 -> context 추출(답변의 근거)
    context = state.context

    # qa_chain 호출 -> 비동기 처리 방식 -> 답변 생성(AIMessage)
    # 비동기 처리 -> invoke() -> await, ainvoke() 
    response = await qa_chain.ainvoke({
        "context": context,
        "chat_history": chat_history,
        "input": latest_question
    })

    # 함수 실행 결과(return) -> dict 자료형 -> LangGraph 시스템 -> AgentState 자료형 변환 -> 상태 업데이트
    return {"messages": [response]} 
````

# 25-3. MCP (Model Context Protocol)

## 가. MCP 개요 및 아키텍처
* **정의**: LLM이 외부 세계와 지능적으로 상호작용할 수 있도록 컨텍스트(도구, 리소스, 프롬프트)를 표준화된 방식으로 주고받는 통신 규약 ("AI 시대의 USB-C")
* **3계층 구조**:
  * **MCP 호스트 (Host)**: AI 애플리케이션 실행 환경 (LLM 오케스트레이션 및 보안 통제)
  * **MCP 클라이언트 (Client)**: 여러 MCP 서버와의 세션 연결 및 메시지 라우팅 전담
  * **MCP 서버 (Server)**: 특정 데이터나 기능을 표준 규격으로 제공하는 독립형 프로그램
* **통신 규약**: JSON-RPC 2.0 기반의 표준 요청/응답 처리

## 나. FastMCP와 멀티 서버 구현
* **FastMCP**: Python 함수를 MCP 서버의 도구(Tool)로 쉽게 노출하고 `try-except` 등으로 안정적인 에러 반환 로직 구축
* **멀티 서버 통합**: 시장 조사 전문가 서버, 보고서 작성 전문가 서버 등 독립된 서버들을 `MultiServerMCPClient`와 LangGraph로 연동하여 확장성 있는 마이크로서비스 아키텍처 구축
