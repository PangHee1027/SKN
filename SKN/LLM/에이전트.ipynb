## 나. OpenAI API 사용법과 토큰 (Token)
### 1. **OpenAI API 개요 및 필요성**
* **필요성**: 모든 기능을 직접 구현할 필요 없이, OpenAI의 대규모 언어 모델(LLM)을 가져와 기업 내 전용 Agentic AI 애플리케이션에 결합
* **OpenAI**: 인공지능 연구 개발 기업으로 ChatGPT를 개발
* **OpenAI API**: 텍스트 생성(사용자 요청에 대한 응답), 코드 생성 등 다양한 기능 제공
* **API 키 관리**: 민감한 정보는 코드와 분리하여 `.env` 파일에 안전하게 저장 (예: `OPENAI_API_KEY="sk-..."`)

### 2. **토큰(Token)과 서브 워드 토큰화 (Subword Tokenization)**
* **핵심 단위**: 토큰(Token)
* **서브 워드 토큰화 개념**: 단어를 고정된 단위로 자르지 않고, 의미를 가진 더 작은 단위인 '서브워드'로 분리하는 방식
* **필요성**: 컴퓨터가 신조어를 포함한 수억 개의 다양한 문장을 적은 양의 단어 조각 조합으로 이해할 수 있도록 지원
* **핵심 아이디어**: 자주 사용되는 단어는 하나의 토큰으로 처리하고, 자주 사용되지 않는 단어나 신조어는 여러 개의 서브워드로 분리 (예: Chatbot $\rightarrow$ Chat + bot)

### 3. **영어와 한글 토큰 처리 비교**
| 구분 | 영어 | 한글 | 차이점 |
| :--- | :--- | :--- | :--- |
| **1 토큰당 글자 수** | 평균 4글자 (알파벳 기준) | 평균 1~2글자 (공백 포함) | 한글은 음절이나 자모 단위로 쪼개짐 |
| **단어당 토큰 수** | 1단어 $\approx 0.75$ 토큰 | 1단어 $\approx 2\sim3$ 토큰 | 한글은 조사와 어미 때문에 토큰이 증가함 |
| **비용 및 속도** | 기준점 (1배) | 약 2.5배~3배 부담 | 한글 텍스트 처리가 훨씬 비싸고 느림 |

## 다. 프롬프트 엔지니어링 (Prompt Engineering)
### 1. **프롬프트 엔지니어링의 중요성**
* **프롬프트**: 생성형 AI 모델에게 내리는 명령어 또는 질문
* **프롬프트 엔지니어링**: 효과적인 프롬프트 작성 방법을 연구하여 AI 모델 성능을 극대화하고 원하는 결과를 얻기 위해 필수적임
* **기법 비교**:
  * **Standard Prompting**: 질문에 대한 직접적인 결과 도출
  * **Chain-of-Thought (CoT) Prompting**: 단계별 사고 과정(풀이 과정)을 함께 유도하여 추론 정확도 향상

### 2. **효과적인 프롬프트 작성 방법**
* **구체적인 지시**: 명확하고 자세하게 요구사항 명시 (예: "소설을 써줘" $\rightarrow$ "등장인물 3명, 갈등 구조를 가진 로맨스 소설 1000자 내외로 써줘")
* **역할 부여**: AI 모델에게 특정 역할 부여 (예: "너는 IT 전문가야. IT 트렌드를 요약해줘.")
* **맥락 제시**: 필요한 배경 정보 제공 (예: "최근 경제 상황을 고려하여, 2026년 하반기 IT 시장 전망을 분석해줘.")
* **예시 활용**: 원하는 결과물의 예시 제시
* **반복 및 개선**: 결과물을 확인하며 프롬프트 수정 및 개선

# LangChain 프레임워크 활용

## 가. LangChain 개요
### 1. **LangChain의 정의와 목표**
* **정의**: 대규모 언어 모델(LLM)을 기반으로 한 AI 애플리케이션 개발을 빠르고 효율적으로 완성해주는 프레임워크
* **목표**: LLM의 추론 능력과 외부 데이터 소스 또는 기타 계산 도구를 결합하여 더욱 강력하고 확장 가능한 애플리케이션 생성

### 2. **핵심 아키텍처 및 구성 요소**
* **Model I/O**: 언어 모델과의 상호작용 관리 (프롬프트 템플릿, 모델 호출, 출력 파서)
* **Data Connection (인덱스)**: 외부 데이터 연동 (도큐먼트 로더, 텍스트 스플리터, 벡터 DB)
* **Memory**: 대화 기록 및 상태 유지
* **Chains**: 여러 컴포넌트를 연결하여 일련의 작업 흐름 구축
* **Agents**: 모델이 직접 도구를 선택하고 실행하는 자율형 에이전트 구조

## 나. Model I/O와 Prompts
### 1. **Prompts (프롬프트 관리)**
* **PromptTemplate**: 변수를 동적으로 주입하여 재사용 가능한 프롬프트 구조 생성
* **ChatPromptTemplate**: 시스템, 사용자, 어시스턴트 등 역할별 메시지 구조를 체계적으로 관리

### 2. **LLMs와 ChatModels**
* **LLMs**: 텍스트를 입력받아 텍스트를 완성하는 모델 인터페이스
* **ChatModels**: 메시지 리스트를 입력받아 대화형 응답을 생성하는 모델 인터페이스 (OpenAI, Anthropic 등 다양한 공급자 지원)

### 3. **Output Parsers (출력 파서)**
* **개념**: LLM의 텍스트 출력을 파이썬 객체나 특정 구조(JSON, Pydantic 모델 등)로 변환하여 후속 처리가 용이하도록 지원

## 다. Chains와 Memory
### 1. **Chains (체인)**
* **개념**: LangChain의 핵심 빌딩 블록으로, 여러 구성 요소(프롬프트 $\rightarrow$ 모델 $\rightarrow$ 파서)를 파이프라인 형태로 순차적 연결
* **LCEL (LangChain Expression Language)**: 파이프(`|`) 연산자를 사용하여 선언적이고 직관적으로 체인을 조합하는 최신 문법

### 2. **Memory (메모리)**
* **필요성**: LLM은 기본적으로 상태가 없는(Stateless) 특성이 있으므로, 이전 대화 맥락을 기억하기 위해 필요
* **주요 종류**:
  * `ConversationBufferMemory`: 대화 기록을 버퍼에 그대로 유지
  * `ConversationSummaryMemory`: 대화 기록이 길어질 경우 핵심 내용을 요약하여 유지

````python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser
from langchain_core.messages import HumanMessage, AIMessage
````

````python
env_file_path='.env'
output = load_dotenv(env_file_path)
print(output)
````

````python
llm = ChatOpenAI(model='gpt-5-nano')
````

````python
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "너는 친절하고 전문적인 AI 비서야. 이전 대화 맥락을 정확하게 기억하여 사용자의 질문에 일관성 있게 답변해야 해"),
        MessagesPlaceholder(variable_name="chat_history")
    ]
)

print(prompt)
````

````python
output_parser = StrOutputParser()
````

````python
chatbot_chain = prompt | llm | output_parser
print(chatbot_chain)
````

````python
chat_history = []

print("대화형 AI 챗봇이 시작되었습니다!. \n")


# 종료 조건이 충족되기전까지 계속 실행 -> while 반복문
while True:
    # 사용자 입력 함수 호출 -> 사용자 입력 생성
    user_input = input("사용자 입력: ")

    # 종료 조건 설정: 사용자 입력이 ["q", "quit", "exit", "종료"]에 속하면 종료
    if user_input.lower() in ["q", "quit", "exit", "종료"]:
        break

    # chain을 호출하기 전, 사용자 질문(입력)을 chat_history에 추가한다.
    """
    # user_input -> 자료형: 문자열 -> LangChain 활용 -> HumanMessage 자료형
    """
    chat_history.append(HumanMessage(content=user_input))

    # invoke() 함수 호출, AI 답변 생성
    response = chatbot_chain.invoke({"chat_history":chat_history})
    print(f"AI 챗봇 답변: {response}\n")

    # LLM 답변 생성이 끝난 후, LLM의 답변을 AIMessage 클래스를 이용, chat_history 추가
    chat_history.append(AIMessage(content=response))
````

````python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_tavily import TavilySearch
from langchain.agents import create_agent
````

````python
env_file_path='.env'
output = load_dotenv(env_file_path)
print(output)
````

````python
search_tool = TavilySearch(max_results=3, search_depth='advanced')
````

````python
tools = [search_tool]
print(f'도구 목록: {tools}')
````

````python
llm = ChatOpenAI(model='gpt-5-nano')
````

````python
system_prompt = """
너는 웹 검색 도구를 적극 활용하여 사용자 질문에 정확하고 객관적인 답변을 제공하는 전문 AI 에이전트야.
전달받은 웹 검색 결과에 기반하여 환각(Hallucination) 없이 사실만을 정리해 답변해야 해. 
그리고 답변은 반드시 한국어로 작성할 것
"""
````

````python
agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt=system_prompt
)
````

````python
response = agent.invoke({
    "messages": [{"role":"user", "content":"2026년 생성형 AI 시장 트렌드 및 최신 AI 에이전트 프레임워크 동향을 간단하게 정리해 줘"}]
})

# 결과 확인
print(f"[최종 에이전트 답변]: \n{response}")
````

````python
text = response['messages'][-1].content
print(text)
````

## 라. Indexes와 RAG (검색 증강 생성)
### 1. **Document Loaders (도큐먼트 로더)**
* **개념**: PDF, TXT, CSV, 웹페이지 등 다양한 소스에서 비정 estructed 데이터를 불러오는 기능

### 2. **Text Splitters (텍스트 스플리터)**
* **개념**: LLM의 입력 토큰 제한을 고려하고 검색 정확도를 높이기 위해 긴 문서를 적절한 크기(Chunk)로 분할

### 3. **Vector Stores & Retrievers (벡터 저장소 및 리트리버)**
* **Embeddings**: 텍스트를 고차원 숫자 벡터로 변환하여 의미적 유사성 비교
* **Vector Store**: 임베디드된 벡터 데이터를 효율적으로 저장하고사 검색하는 데이터베이스 (Chroma, FAISS 등)
* **Retriever**: 사용자 질문과 가장 관련성이 높은 문맥(Context)을 벡터 스토어에서 빠르게 찾아내는 컴포넌트
