# 자연어 처리 (Natural Language Processing, NLP)

## 가. 자연어 처리(NLP) 개요
### 1. **자연어(Natural Language)와 인공 언어**
* **자연어**: 일상생활에서 인간이 사용하는 언어 (예: 한글, 영어 등)
* **반대말**: 인공 언어 또는 프로그래밍 언어

### 2. **자연어 처리 (NLP) 정의**
* 인공지능의 한 분야로, 컴퓨터가 인간의 언어를 이해하고 생성하도록 하는 기술

### 3. **언어 모델 (Language Model, LM)**
* NLP 기술을 구현하며, 문맥을 파악해 다음 단어를 맞추는 알고리즘
* **대표적인 모델**: ChatGPT, Gemini, Claude

## 나. 자연어 처리 과정 및 데이터 수집
### 1. **자연어 처리 파이프라인 (5단계)**
* 텍스트 데이터 수집 $\rightarrow$ 텍스트 전처리 $\rightarrow$ 토큰화 $\rightarrow$ 임베딩 $\rightarrow$ 모델 학습 (목적 달성)

### 2. **텍스트 데이터 수집 방법**
* **공공데이터**: 오픈 API를 이용한 데이터 수집
* **웹 크롤링**: 웹 데이터를 수집 (SNS, 블로그, 카페 등)

## 다. 텍스트 전처리 및 데이터 구조 (형태소와 불용어)
### 1. **텍스트 전처리 (Text Preprocessing)**
* 텍스트 데이터에서 불필요한 데이터를 처리하는 작업
* **특수문자 제거**: 문장 부호 등 특수문자 제거
* **불용어(Stopword) 제거**: 데이터 분석에서 큰 의미가 없는 단어 제거 (예: 영어의 be 동사 등)

### 2. **텍스트 데이터의 구조와 구성 성분**
* 계층 구조: 형태소 $\rightarrow$ 단어(word) $\rightarrow$ 문장(sentence) $\rightarrow$ 문단(paragraph) $\rightarrow$ 문서(document) $\rightarrow$ 말뭉치(corpus)

### 3. **형태소(Morpheme)의 개념과 예시**
* **개념**: 뜻을 가지고 있는 말 중 가장 작은 단위, 더 나눌 수 없는 단위
* **영어 예시**: kindness = kind(형용사) + -ness(접사), useful = use(동사) + -ful(접사)
* **한글 예시**: 책가방 = 책(명사) + 가방(명사), 풋사과 = 풋(접사) + 사과(명사)

### 4. **영어와 한글의 불용어 차이**
* **영어**: 불용어 $\rightarrow$ 단어 단위 (대명사, 관사, 접속사, 조동사 등)
* **한글**: 불용어 $\rightarrow$ 조사, 어미 등 형태소 단위 (예: "꽃밭에 꽃이 많이 피었다" $\rightarrow$ `피었다`는 `피`(어간) + `었`(어미) + `다`(어미)로 구성)

## 라. 토큰화 (Tokenization)
### 1. **토큰화의 개념과 종류**
* **토큰화 (Tokenization)**: 텍스트를 의미 있는 단위(토큰)로 나누는 작업
* **문장 토큰화**: 문장의 끝을 나타내는 구두점 기호(`.`, `!`, `?`)를 기준으로 분할
* **단어 토큰화**: 띄어쓰기 기준으로 분할
* **형태소 토큰화**: 형태소 단위로 분할

### 2. **영어와 한글의 토큰화 차이**
| 언어 구분 | 토큰화 기준 |
| :--- | :--- |
| **영어 토큰화** | 문장 / 띄어쓰기(단어) 기준 |
| **한글 토큰화** | 문장 / 형태소(Morpheme) 기준 |

### 3. **NLTK 라이브러리 활용**
* 문장 단위 토큰화: `sent_tokenize()`
* 단어 단위 토큰화: `word_tokenize()`

## 마. 단어 단위 토큰화의 한계와 서브워드 토큰화 (Subword Tokenization)

### 1. **단어 단위 토큰화의 문제점: OOV (Out-Of-Vocabulary)**
* **OOV 문제**: 컴퓨터의 단어 사전에 존재하지 않는 단어(신조어, 오타, 감정 표현 등)가 입력될 경우 처리가 불가능한 현상
* **예시**: 감성 분류 가사 중 '씹다', '디스' 같은 비속어나 신조어가 OOV로 처리되어 올바른 감정('분노' 등)으로 분류되지 못할 위험 발생

### 2. **해결책: 서브워드 단위 토큰화 (Subword Tokenization)**
* **서브워드(Subword)의 정의**: 단어를 더 쪼갠 의미를 가진 최소한의 글자 조각 (레고 블록과 같은 역할)
* **필요성**: 세상의 모든 단어를 사전에 담는 것은 불가능하므로, 적은 수의 서브워드 조각만 가지고도 조합하여 수억 개의 다양한 문장을 처리할 수 있도록 지원
* **영어 예시**: `unhappiness` $\rightarrow$ `un`(부정) + `happi`(어근) + `ness`(명사 접사) 조합으로 의미 유추 가능
* **대표 알고리즘**: BPE (Byte Pair Encoding), WordPiece (워드피스)

## 바. BPE (Byte Pair Encoding) 알고리즘
### 1. **BPE의 개념과 구축 과정**
* **구축 방식**: 빈도수가 가장 큰 바이그램(Bigram, 2글자 쌍)을 반복적으로 병합하여 단어 사전을 구축하는 방식
* **작동 순서**:
  1. 모든 단어를 문자(Character) 단위로 분리하여 초기 단어 사전 구성
  2. 각 토큰을 두 개씩 묶어 바이그램 쌍의 빈도수 계산
  3. 가장 많이 등장한 바이그램 쌍을 병합하여 새로운 토큰으로 단어 사전에 추가
  4. 원하는 어휘 집합(Vocabulary) 크기가 될 때까지 위 과정을 반복
* **효과**: 자주 등장하지 않는 단어(예: `pug`, `bun`, `hugs`)도 서브워드로 분해하여 단어 사전에 포함함으로써 OOV 문제 완화

## 사. WordPiece 알고리즘과 서브워드 토큰화의 장점
### 1. **WordPiece 알고리즘 및 서브워드 표시 방식**
* **해시태그(`##`) 활용**: 서브워드가 단어의 중간이나 접미어로 사용될 때 앞 토큰에 종속됨을 표시하기 위해 두 개의 해시태그(`##`)를 붙임
* **예시 ("너와 함께한 시간 모두 눈부셨다")**: `너`, `##와`, `함께`, `##한`, `시간`, `모두`, `눈부`, `##셔`, `##ㅆ다`
* **복원 규칙**:
  * `##`이 없는 토큰을 만나면 $\rightarrow$ 앞 토큰과 한 칸 띄우고 작성
  * `##`이 있는 토큰을 만나면 $\rightarrow$ `##`을 지우고 앞 토큰에 바짝 붙여서 작성

### 2. **서브워드 토큰화의 3대 장점**
1. **OOV (Out-of-Vocabulary) 완벽 해결**: 사전에 없는 단어나 오타, 신조어도 글자 조각으로 분해하여 처리 가능
2. **가벼운 모델 용량**: 수백만 개의 단어를 모두 저장하는 대신, 수만 개의 핵심 서브워드 조각만 저장하여 메모리 극대화 절약
3. **의미적 연관성 학습**: `walk`, `walking`, `walked`가 공통 서브워드(`walk`)를 공유함으로써 단어 간의 어근과 의미적 뿌리를 컴퓨터가 자연스럽게 이해

````python
sample_texts = [
    "SK네트웍스 Family AI 캠프 자연어 처리 수업입니다.",
    "딥러닝과 자연어 처리를 열심히 공부하고 있습니다.",
    "서브워드 토큰화와 WordPiece 알고리즘을 실습합니다.",
    "한국어 형태소 분석과 토큰화 과정은 인공지능 학습에 매우 중요합니다.",
    "신조어나 사전에 없는 단어도 서브워드로 쪼개어 거뜬히 처리할 수 있습니다."
]

with open("./데이터/sample_corpus.txt", "w", encoding="utf-8") as f:
    for text in sample_texts:
        f.write(text + "\n")

print("sample_corpus.txt 파일이 성공적으로 생성되었습니다!")
````

````python
# 필요한 라이브러리 임포트
import os
from tokenizers import BertWordPieceTokenizer
from transformers import BertTokenizerFast

# -------------------------------------------------------------------------
# 1단계: 실습용 소스 말뭉치(Corpus) 텍스트 파일 생성
# -------------------------------------------------------------------------
corpus_file = "./데이터sample_corpus.txt"
sample_texts = [
    "SK네트웍스 Family AI 캠프 자연어 처리 수업입니다.",
    "딥러닝과 자연어 처리를 열심히 공부하고 있습니다.",
    "서브워드 토큰화와 WordPiece 알고리즘을 실습합니다.",
    "한국어 형태소 분석과 토큰화 과정은 인공지능 학습에 매우 중요합니다.",
    "신조어나 사전에 없는 단어도 서브워드로 쪼개어 거뜬히 처리할 수 있습니다."
]

# 텍스트 파일을 UTF-8 인코딩으로 저장
with open(corpus_file, "w", encoding="utf-8") as f:
    for text in sample_texts:
        f.write(text + "\n")

print(f"[{corpus_file}] 샘플 말뭉치 파일 생성 완료!")


# -------------------------------------------------------------------------
# 2단계: BertWordPieceTokenizer 객체 생성 및 학습(Train)
# -------------------------------------------------------------------------
# WordPiece 알고리즘을 구현하는 학습 전용 토크나이저 생성
wordpiece_tokenizer = BertWordPieceTokenizer(
    clean_text=True,
    handle_chinese_chars=True,
    strip_accents=False,
    lowercase=False  # 한글 및 영어 대소문자 구분 여부
)

# 말뭉치 파일로부터 최적의 서브워드 단어 사전(vocab) 학습 진행
wordpiece_tokenizer.train(
    files=[corpus_file],
    vocab_size=500,          # 예시를 위한 작은 어휘 집합 크기 설정
    min_frequency=1,         # 최소 등장 빈도
    special_tokens=["[PAD]", "[UNK]", "[CLS]", "[SEP]", "[MASK]"], # 특수 토큰 정의
    wordpieces_prefix="##"   # 서브워드 접두사 설정
)

# 학습된 단어 사전을 저장할 디렉토리 생성 후 vocab.txt 저장
os.makedirs("model_vocab", exist_ok=True)
wordpiece_tokenizer.save_model("model_vocab")
print("WordPiece 단어 사전(vocab.txt) 학습 및 'model_vocab' 폴더 저장 완료!")


# -------------------------------------------------------------------------
# 3단계: BertTokenizerFast를 이용한 실제 한글 문장 토큰화 및 복원 실습
# -------------------------------------------------------------------------
# 방금 학습시킨 vocab.txt 파일을 기반으로 BERT 모델용 Fast 토크나이저 로드
fast_tokenizer = BertTokenizerFast.from_pretrained(
    pretrained_model_name_or_path="bert-base-uncased",  # 또는 "klue/bert-base" 등
    vocab_file="model_vocab/vocab.txt", 
    lowercase=False
)

# 토큰화 테스트 대상 문장 (사전에 없을 법한 단어 포함)
test_sentence = "SK네트웍스 AI 캠프에서 자연어 처리를 공부합니다."

# 1. 텍스트 토큰화 수행 (tokenize)
tokens = fast_tokenizer.tokenize(test_sentence)
print("\n--- 1. 토큰화 결과 (Tokens) ---")
print(tokens)

# 2. 정수 인코딩 (Token IDs 변환)
input_ids = fast_tokenizer.encode(test_sentence)
print("\n--- 2. 정수 인코딩 결과 (Input IDs) ---")
print(input_ids)

# 3. 토큰들을 다시 원래 문장으로 복원 (Decode)
decoded_text = fast_tokenizer.decode(input_ids)
print("\n--- 3. 문장 복원 결과 (Decode) ---")
print(decoded_text)
````

````python
# 내장된 단어 사전 -> 추출 -> get_vocab()

## get_vocab() 함수 실행
vocab = tokenizer.get_vocab()

## 단어 사전 출력
print(vocab)

print('-'*80)

## 단어 사전의 자료형 확인
print(f'단어 사전의 자료형: {type(vocab)}')
````

## 마. 한글 형태소 단위 토큰화와 Konlpy 라이브러리

### 1. **Konlpy 라이브러리 개요**
* 파이썬 기반의 대표적인 한국어 자연어 처리(NLP) 및 형태소 분석 라이브러리
* 주요 형태소 분석기: **Okt(Open Korea Text)**, **Mecab**, Komoran, Hannanum, Kkma 등

## 바. Okt 형태소 분석기

### 1. **설치 및 임포트**
* 설치 명령어: `pip install konlpy`
* 도구 임포트: `from konlpy.tag import Okt`
* 객체 생성: `okt = Okt()`

### 2. **Okt 주요 클래스 함수**
* **`okt.morphs(text)`**: 입력 텍스트를 형태소 단위로 분리하여 리스트로 반환
* **`okt.pos(text)`**: 형태소 분석과 함께 품사(Part-of-Speech) 태깅을 수행하여 튜플 형태의 리스트(`[(형태소, 품사), ...]`) 반환
* **`okt.nouns(text)`**: 텍스트에서 명사만 추출하여 리스트로 반환

### 3. **Okt 주요 품사(POS) 태그**
* **Noun**: 명사, 대명사, 수사 포함
* **Verb / Adjective**: 동사 / 형용사
* **Josa**: 조사
* **Punctuation**: 구두점 (마침표, 물음표, 쉼표 등)

````python
# Okt 형태소 분석기 실습
from konlpy.tag import Okt

okt = Okt()
sample_sentence = "SK네트웍스 Family AI 캠프 자연어 처리 수업 정말 유익하고 재미있습니다!"

# 1. 형태소 분석
morphs = okt.morphs(sample_sentence)
print("형태소 분석 결과:", morphs)

# 2. 품사 태깅
pos_result = okt.pos(sample_sentence)
print("\n품사 태깅 결과:", pos_result)

# 3. 명사 추출
nouns = okt.nouns(sample_sentence)
print("\n명사 추출 결과:", nouns)
````

## 사. Mecab 형태소 분석기

### 1. **설치 및 임포트**
* 설치 명령어: `pip install python-mecab-ko`
* 도구 임포트: `from mecab import MeCab as Mecab`
* 객체 생성: `mecab = Mecab()`

### 2. **Mecab 주요 클래스 함수**
* **`mecab.morphs(text)`**: 형태소 단위 토큰화
* **`mecab.pos(text)`**: 형태소와 세종(Sejong) 품사 태그 부착
* **`mecab.nouns(text)`**: 명사 추출

### 3. **주요 세종 품사 태그 (Sejong Tag)**
* **체언**: `NNG`(일반 명사), `NNP`(고유 명사), `NNB`(의존 명사), `NR`(수사), `NP`(대명사)
* **용언**: `VV`(동사), `VA`(형용사), `VX`(보조 용언), `VCP`(긍정 지정사)
* **수식언 / 독립언 / 조사**: `MM`(관형사), `MAG`(일반 부사), `IC`(감탄사), `JK*`(조사 종류)
* **기타 부호 및 기호**: `SF`(마침표/물음표/느낌표), `SL`(외국어), `SN`(숫자)

````python
# Mecab 형태소 분석기 실습
from mecab import MeCab

mecab = MeCab()
sample_sentence = "대한민국 헌법 제1조 1항은 대한민국은 민주공화국이다."

# 1. 형태소 분석
morphs_mecab = mecab.morphs(sample_sentence)
print("Mecab 형태소 분석:", morphs_mecab)

# 2. 품사 태깅
pos_mecab = mecab.pos(sample_sentence)
print("\nMecab 품사 태깅:", pos_mecab)

# 3. 명사 추출
nouns_mecab = mecab.nouns(sample_sentence)
print("\nMecab 명사 추출:", nouns_mecab)
````

## 마. 단어 빈도수 분석 및 시각화 (Word Frequency & WordCloud)

### 1. **실습 개요 및 파이프라인**
* 한글 텍스트 데이터 $\rightarrow$ 한글 형태소 분석기(Mecab 등) $\rightarrow$ 형태소 단위 토큰화 $\rightarrow$ 명사(Noun) 추출
* `python` 내장 패키지 `collections`의 `Counter()` 함수를 이용한 빈도수 분석
* 빈도수 분석 결과를 바탕으로 `wordcloud` 라이브러리를 활용한 시각화

## 바. `collections` 패키지의 `Counter()` 함수

### 1. **기능 및 특징**
* 문자열, 리스트, 튜플 등 여러 성분 원소를 저장하고 있는 자료형에서 각 성분 원소의 빈도수(출현 횟수)를 계산해 주는 딕셔너리 형태의 객체 반환 함수

### 2. **사용 예시**
```python
from collections import Counter
Counter(["hi", "hey", "hi", "hi", "hello", "hey"])
# 결과: Counter({'hi': 3, 'hey': 2, 'hello': 1})
```

## 사. `wordcloud` 라이브러리를 이용한 시각화

### 1. **개념**
* 텍스트에 등장하는 단어들의 빈도수를 기반으로 주요 단어들을 시각적으로 크고 작게 표현해 주는 워드클라우드 시각화 라이브러리

### 2. **사용 방법 (6단계 절차)**
1. 필요한 라이브러리 및 함수 임포트 (`WordCloud`, `matplotlib.pyplot` 등)
2. 한글 폰트 설정 (폰트 파일 경로 지정)
3. `WordCloud` 클래스 함수 객체 생성 (배경색, 폰트 경로, 최대 단어 수 등 설정)
4. 워드클라우드 이미지 생성 (`generate_from_frequencies` 또는 `generate`)
5. 이미지 저장 (`to_file`)
6. 이미지 출력 (`matplotlib`을 이용한 시각화 출력)
