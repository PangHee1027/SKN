# 회귀
## 가. 회귀 (Regression)
  * 학습 데이터를 이용해 데이터의 특성과 상관관계를 파악한 후, 이를 바탕으로 미지의 데이터에 대해 연속적인 숫자 값을 예측하는 지도학습 기법

## 나. 선형 회귀(Linear Regression)
### 1. 입력 데이터($x$)와 출력 데이터($y$)의 관계를 $y = Wx + b$ 형태의 직선으로 나타내어 최적의 예측 선을 그리는 알고리즘
### 2. 선형 회귀의 성능 평가 지표
#### 가) 오차(Error/Loss)가 작을수록 우수한 모델로 평가
#### 나) MAE (Mean Absolute Error, 평균 절대값 오차): 실제 값과 예측 값 차이의 절대값 평균
  * $$MAE = \frac{\sum \vert{}y - \hat{y}\vert{}}{n}$$
#### 다) MSE (Mean Squared Error, 평균 제곱 오차): 실제 값과 예측 값 차이의 제곱 평균
  * $$MSE = \frac{\sum (y - \hat{y})^2}{n}$$
#### 라) RMSE (Root Mean Squared Error, 평균 제곱근 오차): MSE에 제곱근을 씌운 지표
  * $$RMSE = \sqrt{\frac{\sum_{i=1}^{n} (\hat{y}_i - y_i)^2}{n}}$$

```python
# 수업 실습
import numpy as np
import pandas as pd

file_path1 = "./실습데이터/house_price.csv"
df1 = pd.read_csv(file_path1)
display(df1)
```

```python
# 데이터 전처리
## 누락데이터 확인
print(df1.isnull().sum())
```

```python
## feature/label 분할
X1 = df1.drop(columns=["medv"])
y1 = df1["medv"]
display(X1)
display(y1)
```

```python
## 학습용/평가용 데이터 분할
from sklearn.model_selection import train_test_split

X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size = 0.25, random_state = 0)

display(X1_train)
display(y1_train)
display(X1_test)
display(y1_test)
```

```python
## scaling -> 선형 회귀임으로 표준화
from sklearn.preprocessing import StandardScaler

standard_scaler = StandardScaler()

# 학습용 = fit_transform()
X1_train_standard = standard_scaler.fit_transform(X1_train)
# 평가용 = transform()
X1_test_standard = standard_scaler.transform(X1_test)

df1_X1_train_standard = pd.DataFrame(data = X1_train_standard, index = X1_train.index, columns = X1_train.columns)
df1_X1_test_standard = pd.DataFrame(data = X1_test_standard, index = X1_test.index, columns = X1_test.columns)

display(df1_X1_train_standard)
display(df1_X1_test_standard)
```

```python
# 선형회귀
from sklearn.linear_model import LinearRegression

lireg1 = LinearRegression()

lireg1.fit(X1_train_standard, y1_train)

print() # fit() 함수가 마지막 줄에 오는 것 방지 (에러 발생 차단)
```

```python
# 결과 확인
print(f"가중치(W) : {lireg1.coef_}")
print(f"편향(b) : {lireg1.intercept_}")
```

```python
from sklearn.metrics import root_mean_squared_error as rmse

y_pred_lr1 = lireg1.predict(X1_test_standard)

rmse_lr1 = rmse(y1_test, y_pred_lr1)

print(rmse_lr1)
```

### 3. 경사 하강법 (Gradient Descent)
#### 가) 개념 : 손실 함수(Loss Function) $L(w,b)$의 값을 최소화하는 가중치($W$)와 편향($b$)을 찾아 나가는 최적화 알고리즘
#### 나) 수식 설계를 위한 명확한 변수 정의
##### 1) 종속 변수 : 손실 함수 (L, Loss Function / Cost Function)
  * 모델의 예측 값과 실제 정답 간의 오차 크기를 나타내는 단 하나의 실수 값
##### 2) 독립 변수: 가중치 (W, Weights) 및 편향 (b, Bias)
  * 모델 내부에서 입력 값(데이터)의 영향력을 조절하며, 궁극적으로 손실 함수의 출력값(L)을 결정짓는 조절 가능핚 독립 변수 매개변수들
##### 3) $$L(w,b) = \frac{1}{n} \sum_{i=1}^{n} (y_i - (wx_i + b))^2$$
#### 다) 가중치 업데이트 수식 : $$W_{new} = W_{old} - \eta \frac{\partial L}{\partial W}$$
#### 라) 파라미터 vs 하이퍼파라미터 :
  * 파라미터 (Parameter) : 모델이 데이터를 통해 스스로 학습하여 최적화하는 변수 ($W, b$)
  * 하이퍼파라미터 (Hyperparameter) : 사람이 학습 시작 전 직접 설정하는 변수 (학습률 $\eta$)
### 4. 모델 정규화 (Regularization)
#### 가) 모델이 과대적합(Overfitting)되지 않도록 특성의 수를 줄이지 않으면서 가중치($W$)의 크기를 제한하는 기법
#### 나) L1 규제 (Lasso) : 모든 가중치를 같은 크기로 감소시켜, 중요도가 낮은 특성의 가중치를 $0$으로 만들어 제거하는 효과

```python
from sklearn.linear_model import Lasso

```

#### 다) L2 규제 (Ridge) : 모든 가중치를 같은 비율로 감소시켜, 큰 가중치를 크게 줄이고 전체 가중치 크기를 $0$에 가깝게 만듦

```python
from sklearn.linear_model import Ridge

```

#### 라) 제미나이 생성 예제

```python
# 1. 가상 데이터 생성 (독립변수 X 3개, 종속변수 y)
np.random.seed(42)
X2 = np.random.rand(100, 3) * 10
y2 = 2.5 * X2[:, 0] + 1.2 * X2[:, 1] - 0.8 * X2[:, 2] + 5 + np.random.randn(100) * 2
```

```python
# 2. 학습용(Train)과 평가용(Test) 데이터 분할
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.2, random_state=42)
```

```python
# -------------------------------------------------------------
# 3-1. 기본 선형 회귀 (LinearRegression)
# -------------------------------------------------------------
lireg2 = LinearRegression()
lireg2.fit(X2_train, y2_train)
y_pred_lr2 = lireg2.predict(X2_test)
rmse_lr2 = rmse(y2_test, y_pred_lr2)

print("=== Linear Regression ===")
print(print_weights := f"가중치(W): {lireg2.coef_}, 편향(b): {lireg2.intercept_:.4f}")
print(f"RMSE: {rmse_lr2:.4f}\n")
```

```python
# -------------------------------------------------------------
# 3-2. L1 규제 선형 회귀 (Lasso)
# -------------------------------------------------------------

lasso = Lasso(alpha=0.1, random_state=0)
lasso.fit(X2_train, y2_train)
y_pred_lasso = lasso.predict(X2_test)
rmse_lasso = rmse(y2_test, y_pred_lasso)

print("=== Lasso (L1 Regularization) ===")
print(f"가중치(W): {lasso.coef_}")
print(f"RMSE: {rmse_lasso:.4f}\n")
```

```python
# -------------------------------------------------------------
# 3-3. L2 규제 선형 회귀 (Ridge)
# -------------------------------------------------------------

ridge = Ridge(alpha=1.0, random_state=0)
ridge.fit(X2_train, y2_train)
y_pred_ridge = ridge.predict(X2_test)
rmse_ridge = rmse(y2_test, y_pred_ridge)

print("=== Ridge (L2 Regularization) ===")
print(f"가중치(W): {ridge.coef_}")
print(f"RMSE: {rmse_ridge:.4f}")
```
