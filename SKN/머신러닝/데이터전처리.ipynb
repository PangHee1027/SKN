# 데이터전처리
## 가. 개요
### 1. 개념 : 수집한 데이터를 분석 및 머신러닝 모델 학습에 적합한 형태로 변환하는 작업
### 2. 목적 : 데이터 내 노이즈 및 불균형을 정제하여 모델 학습의 안정성과 예측 성능 향상
## 나. 특성 공학 (Feature Engineering) 및 주요 전처리 기법
### 1. 파생 변수 생성 : 기존 변수들을 조합하거나 가공하여 유의미한 새 변수 생성
### 2. 범주형 변수 변환 (Encoding) :

```python
import pandas as pd
import numpy as np

file_path1 = "./실습데이터/iris_dataset.csv"
df1 = pd.read_csv(file_path1)
display(df1)
```

#### 가) 원-핫 인코딩 (One-Hot Encoding) : 범주별로 고유 컬럼을 생성해 0과 1로 표현 (pd.get_dummies())

```python
print(df1['label'].value_counts())
df_encoded1 = pd.get_dummies(df1, columns=['label'], dtype=int)
display(df_encoded1)
```

#### 나) 레이블 인코딩 (Label Encoding) : 범주형 데이터를 0부터 $n-1$까지의 정수 수치로 변환 (df.replace())

```python
df_encoded2 = df1.replace({'setosa' : 0, 'versicolor' : 1, 'virginica' : 2})
display(df_encoded2)
```

### 3. 연속형 변수 변환 (Scaling) :
#### 가) 정규화 (Normalization): 데이터의 최소-최대 범위를 [0, 1] 사이 값으로 변환 ($x_{new} = \frac{x-x_{min}}{x_{max}-x_{min}}$)

```python
# 데이터 생성
data1 = {
    "age" : [22, 25, 47, 38, 29],
    "income" : [3000, 3500, 8500, 6000, 4200]
}

df2 = pd.DataFrame(data1)

display(df2)
```

```python
from sklearn.preprocessing import MinMaxScaler

minmax_scaler = MinMaxScaler()

data_minmax = minmax_scaler.fit_transform(df2)
data_index = df2.index

df2_nomalized = pd.DataFrame(data = data_minmax, index = data_index, columns = ['nomalized_age', 'nomalized_income'])

display(df2_nomalized)
```

#### 나) 표준화 (Standardization) : 평균 0, 표준편차 1인 정규분포 형태로 변환 ($x_{new} = \frac{x-\mu}{\sigma}$)

```python
from sklearn.preprocessing import StandardScaler

standard_scaler = StandardScaler()

data_standard = standard_scaler.fit_transform(df2)

df2_standard = pd.DataFrame(data = data_standard, index = data_index, columns = ['standard_age', 'standard_income'])

display(df2_standard)
```

#### 다) 머신러닝 알고리즘별 Scaling 가이드

| 알고리즘 분류 | 대표 모델 | 추천 전처리 방법 | 핵심 이유 |
| :--- | :--- | :--- | :--- |
| **선형 & 규제 모델** | Linear / Logistic Regression, Ridge, Lasso | `StandardScaler` | L1/L2 규제 패널티 계산 시 스케일 영향 차단 |
| **SVM & 차원 축소** | SVM, PCA, LDA | `StandardScaler` | 초평면 마진 및 분산 최적화 영향 최소화 |
| **딥러닝 (신경망)** | MLP, CNN, RNN 등 | `StandardScaler` 또는<br>(이상치 clip 후) `MinMaxScaler` | 스케일러 1개만 적용 (경사하강법 수렴 속도 최적화) |
| **거리 기반 모델** | K-Means 등 | 이상치 clip 후<br>`StandardScaler` 또는 `RobustScaler` | K-Means는 이상치에 극도로 취약 |
| **트리 기반 모델** | Decision Tree, Random Forest, XGBoost, LightGBM 등 | **스케일링 불필요** | 변수의 크기가 아닌 대소 비교를 통한 조건 분할 방식 사용 |

### 3. 구간화 (Binning) : 연속형 수치 데이터를 일정 기준 구간으로 나누어 범주형 데이터로 변환 (pd.cut())

```python
df2_binned = pd.cut(df2["age"], bins = [0, 10, 20, 30, 40, 50, 60, 70], labels = ['0s', "10s", "20s", "30s", '40s', '50s', '60s'])
display(df2_binned)
```

### 2. 데이터 병합 (Concat) : pd.concat()
#### 가) 개념 및 특징
- 여러 데이터프레임(DataFrame)이나 시리즈(Series)를 축(axis)을 기준으로 연결하여 하나의 데이터프레임으로 병합하는 함수
- 기본값(`axis=0`)은 **행 방향(위아래)**으로 병합하며, 공통된 컬럼을 기준으로 데이터를 쌓음
- `axis=1` 지정 시 **열 방향(좌우)**으로 병합
- `ignore_index=True` 옵션을 설정하면 기존 인덱스를 무시하고 `0`부터 시작하는 새로운 인덱스를 재할당

```python
# 샘플 데이터프레임 생성
df1 = pd.DataFrame({'A': ['A0', 'A1'], 'B': ['B0', 'B1']})
df2 = pd.DataFrame({'A': ['A2', 'A3'], 'B': ['B2', 'B3']})
df3 = pd.DataFrame({'A': ['A4', 'A5'], 'B': ['B4', 'B5']})

# axis=0 방향으로 병합 (기본값, 위아래 병합)
# ignore_index=True를 통해 인덱스를 0, 1, 2, 3, 4, 5로 재배열
df_concat = pd.concat([df1, df2, df3], ignore_index=True)

display(df_concat)
```
