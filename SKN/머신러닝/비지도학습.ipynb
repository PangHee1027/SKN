# 비지도 학습 (Unsupervised Learning)

## 가. 군집화 (Clustering) 개요
### 1. **개념** : 유사한 속성들을 갖는 데이터를 묶어 전체 데이터를 몇 개의 군집(그룹)으로 나누는 비지도 학습 방법
### 2. **군집화 기준** :
#### 가) **군집 간 분산 (Inter-cluster variance)** : 최대화 (집단 간의 이질성 최대)
#### 나) **군집 내 분산 (Inner-cluster variance)** : 최소화 (집단 내의 유사성 최대)
### 3. **분류(Classification) vs 군집화(Clustering)** :
#### 가) **분류** : 사전에 정의된 범주가 있는(labeled) 데이터를 학습 후 미지의 데이터에 대해 범주 예측
#### 나) **군집화** : 사전에 정의된 범주가 없는(unlabeled) 데이터로부터 최적의 군집을 탐색

## 가. K-Means 알고리즘
### 1. **개념** : 군집화에서 가장 일반적인 알고리즘으로, 임의의 $K$개 만큼 클러스터를 묶는 방식이며 유클리디안 거리를 측정하여 군집화 진행
### 2. **작동 원리** :
#### 가) 초기 중심점(Centroid) 설정
#### 나) 각 데이터를 가장 가까운 중심점에 할당하여 초기 클러스터 생성
#### 다) 클러스터에 할당된 데이터 간 거리의 평균을 계산하여 새로운 중심점(중심 이동) 지정
#### 라) 중심점 변경과 이동이 없을 때까지 반복 종료
### 3. **최적의 클러스터 개수 설정 (엘보우 Method)** :
#### 가) 클러스터 내 오차제곱합(SSE)이 최소가 되도록 설정
#### 나) 클러스터 개수를 1부터 $n$까지 증가시키며 SSE 변화 관찰
#### 다) SSE의 변화가 가장 크게 꺾이는(감소 폭이 줄어드는) 구간을 최적의 개수로 선정

```python
# [실습] K-Means를 이용한 붓꽃(Iris) 데이터 군집화
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import pandas as pd

# 1. 붓꽃 데이터 불러오기
iris = pd.read_csv("./실습데이터/iris_dataset.csv")
X = iris.drop(columns=['label'])

# 2. KMeans 모델 생성 및 학습 (클러스터 개수 k=3)
model = KMeans(n_clusters=3, random_state=0)
cluster_labels = model.fit_predict(X)

# 3. 결과 확인
print("K-Means 군집화 예측 결과 레이블:\n", cluster_labels)
```

## 가. GMM (Gaussian Mixture Model)
### 1. **개념** : 데이터가 여러 개의 다른 모양의 정규 분포(가우시안 분포)로 결합되어 있다는 가정 하에, 개별 데이터를 가우시안 분포별로 묶어주는 확률 기반 군집화 알고리즘
### 2. **특징** : 데이터가 원형이 아닌 타원형 분포를 가지거나, 클러스터의 크기와 모양이 다양할 때 K-Means보다 유연하게 군집화 수행 가능

```python
# [실습] GMM을 이용한 붓꽃(Iris) 데이터 군집화
from sklearn.datasets import load_iris
from sklearn.mixture import GaussianMixture
import pandas as pd

# 1. 붓꽃 데이터 불러오기
iris = pd.read_csv("./실습데이터/iris_dataset.csv")
X = iris.drop(columns=['label'])

# 2. GaussianMixture 모델 생성 및 학습 (컴포넌트 개수 n_components=3)
gmm_model = GaussianMixture(n_components=3, random_state=0)
gmm_labels = gmm_model.fit_predict(X)

# 3. 결과 확인
print("GMM 군집화 예측 결과 레이블:\n", gmm_labels)
```
