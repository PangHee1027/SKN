# 붓꽃 품종 분류

## 붓꽃 데이터 EDA

```python
# 필요한 라이브러리 임포트
import pandas as pd
import plotly.express as px
```

```python
# 경고 무시 설정
import warnings
warnings.filterwarnings('ignore')
```

### 데이터 불러오기

```python
# 파일 경로 설정
file_path='./실습데이터/iris_dataset.csv'

# DataFrame 생성
df = pd.read_csv(file_path)

# 결과 확인
display(df)
```

### 데이터 전처리

```python
# 누락 데이터 처리

# 각 컬럼별 누락 데이터의 수 확인
num_nulls = df.isnull().sum()

# 결과 확인
print(f'각 컬럼별 누락 데이터의 수 : \n{num_nulls}')
```

### 특성별 품종 간의 차이 시각화

```python
# 정답 컬럼 -> 품종별 개수 확인 -> 정답(품종)의 분포(균형 또는 불균형) 확인
counts = df['label'].value_counts()
print(f'붓꽃 데이터의 품종별 개수 : \n{counts}')
```

```python
# 컬럼 이름 추출
names = df.columns
print(f'붓꽃 데이터의 컬럼 이름 : \n{names}')
```

#### 꽃받침의 길이와 품종 간의 차이 시각화

```python
# 꽃받침의 길이 분포 시각화
fig1 = px.histogram(
    data_frame=df,
    x=names[0],
    nbins=30,
    title='꽃받침의 길이 분포'
)

## 모든 막대에 일괄적으로 검은색 경계선 넣기
fig1.update_traces(marker_line_color='black', marker_line_width=1)
fig1.show()
```

```python
# 품종별 꽃받침의 길이 분포 시각화
fig2 = px.histogram(
    data_frame=df,
    x=names[0],
    nbins=30,
    title='품종별 꽃받침의 길이 분포',
    color=names[-1],
    barmode='overlay',
    opacity=0.6
)

fig2.show()
```

#### 꽃받침의 너비와 품종 간의 차이 시각화

```python
# 꽃받침의 너비 분포 시각화
fig3 = px.histogram(
    data_frame=df,
    x=names[1],
    nbins=30,
    title='꽃받침의 너비 분포'
)

## 모든 막대에 일괄적으로 검은색 경계선 넣기
fig3.update_traces(marker_line_color='black', marker_line_width=1)
fig3.show()
```

```python
# 품종별 꽃받침의 너비 시각화
fig4 = px.histogram(
    data_frame=df,
    x=names[1],
    nbins=30,
    title='품종별 꽃받침의 너비 분포',
    color=names[-1],
    barmode='overlay',
    opacity=0.6
)

fig4.show()
```
