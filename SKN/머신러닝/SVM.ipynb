# SVM
## 가. 핵심 개념 및 원리
### 1. 개념 : 클래스 간의 안전한 거리(마진, Margin)를 최대화하여 모델의 일반화 성능을 높이는 대표적 분류/회귀 알고리즘
### 2. 용어
#### 가) 서포트 벡터 (Support Vector) : 결정 경계와 가장 가까운 위치에 존재하는 데이터 포인트
#### 나) 마진 (Margin) : 결정 경계(Decision Boundary)와 서포트 벡터 사이의 폭(거리)을 의미
### 3. 학습 목표 : 데이터를 두 클래스로 완벽하게 분리하면서 마진이 최대화되는 최적의 초평면(Hyperplane)을 탐색
## 나. 장단점
### 1. 장점
#### 가) 소규모 데이터셋(수십~수백 개)에서도 일반화 성능이 탁월
#### 나) 피처(Feature) 수가 샘플 수보다 많은 고차원 데이터에서도 안정적인 성능을 유지
### 2. 단점
#### 가) 데이터 크기가 커지면 연산 복잡도가 증가하여 학습 속도가 감소
#### 나) 로지스틱 회귀와 달리 각 특성이 결과에 미치는 영향력을 직관적으로 해석 어려움
## 다. SVM의 종류
### 1. 선형 SVM (Linear SVM) : 선형 결정 경계를 생성하여 데이터를 분리
### 2. 비선형 SVM (Non-linear SVM) : 커널 함수(Kernel Trick)를 이용해 데이터를 고차원 공간으로 변환한 후 최적의 초평면을 설정
  * 주요 커널 : RBF (Radial Basis Function, 방사 기저 함수)가 기본값(default)으로 널리 사용
## 라. 데이터 전처리 유의사항
### 1. 스케일링 필수 : SVM은 데이터 간 거리 기반 알고리즘이므로 특성 간 범위를 맞추는 StandardScaler 전처리가 성능 향상에 필수적

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import time
```

### 사용 데이터 : iris_dataset.csv

```python
# 데이터 불러오기
df1 = pd.read_csv('./실습데이터/iris_dataset.csv')

# 독립변수(X)와 종속변수(y) 분리
X1 = df1.drop(columns=['label'])
y1 = df1['label']
```

```python
X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.2)
```

```python
scaler1 = StandardScaler()
X1_train_scaled = scaler1.fit_transform(X1_train)
X1_test_scaled = scaler1.transform(X1_test)
```

```python
svc1 = SVC(random_state=0)
svc1.fit(X1_train_scaled, y1_train)
svc1_pred = svc1.predict(X1_test_scaled)

print("=== [1] SVM 모델 평가 (Test Set) ===")
print(f"Accuracy: {accuracy_score(y1_test, svc1_pred):.4f}")
print(confusion_matrix(y1_test, svc1_pred))
print(classification_report(y1_test, svc1_pred))
```

```python
lr1 = LogisticRegression(random_state=0)
lr1.fit(X1_train_scaled, y1_train)
lr1_pred = lr1.predict(X1_test_scaled)

print("=== [2] 로지스틱 회귀 모델 평가 (Test Set) ===")
print(f"Accuracy: {accuracy_score(y1_test, lr1_pred):.4f}")
print(confusion_matrix(y1_test, lr1_pred))
print(classification_report(y1_test, lr1_pred))
```

```python
X1_scaled = scaler1.fit_transform(X1)
print("=== [3] 전체 데이터셋 기준 정확도 비교 ===")
print(f"SVM 전체 정확도: {accuracy_score(y1, svc1.predict(X1_scaled)):.4f}")
print(f"Logistic Regression 전체 정확도: {accuracy_score(y1, lr1.predict(X1_scaled)):.4f}")
```

### 사용 데이터 : diabetes_binary_5050split_health_indicators_BRFSS2015.csv

```python
# 1. 데이터 로드
df2 = pd.read_csv('./실습데이터/diabetes_binary_5050split_health_indicators_BRFSS2015.csv')
X2 = df2.drop(columns=['Diabetes_binary'])
y2 = df2['Diabetes_binary']
```

```python
# 2. 학습용(Train)과 평가용(Test) 데이터 분할 (8:2 비율)
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.2)
```

```python
# 3. 데이터 표준화 (StandardScaler) - 선형모델과 SVM 모두 필수
scaler2 = StandardScaler()
X2_train_scaled = scaler2.fit_transform(X2_train)
X2_test_scaled = scaler2.transform(X2_test)
```

```python
# ==========================================
# 4. 로지스틱 회귀 (Logistic Regression) 모델
# ==========================================
print("=== [1] 로지스틱 회귀 모델 학습 및 평가 ===")
start_time = time.time()

lr2 = LogisticRegression(max_iter=1000, random_state=0)
lr2.fit(X2_train_scaled, y2_train)
lr2_pred = lr2.predict(X2_test_scaled)

print(f"학습 소요 시간: {time.time() - start_time:.2f}초")
print(f"정확도 (Accuracy): {accuracy_score(y2_test, lr2_pred):.4f}")
print("\n[Classification Report]")
print(classification_report(y2_test, lr2_pred))
```

```python
# ==========================================
# 5. SVM (Support Vector Classifier) 모델
# ==========================================
print("\n=== [2] SVM 모델 학습 및 평가 (속도를 위해 일부 샘플 사용) ===")
# 대용량 데이터이므로 학습용 데이터 중 5,000개만 샘플링하여 SVM 학습 진행
X_train_sub = X2_train_scaled[:5000]
y_train_sub = y2_train.iloc[:5000]
```

```python
start_time = time.time()
svc2 = SVC(random_state=0)
svc2.fit(X_train_sub, y_train_sub)
svc2_pred = svc2.predict(X2_test_scaled)
```

```python
print(f"학습 소요 시간: {time.time() - start_time:.2f}초")
print(f"정확도 (Accuracy): {accuracy_score(y2_test, svc2_pred):.4f}")
print("\n[Classification Report]")
print(classification_report(y2_test, svc2_pred))
```
